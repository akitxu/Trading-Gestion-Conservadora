```python
"""
MÓDULO 1: CLASES BASE Y UTILIDADES
Ejecutar en Celda 1 de Jupyter Lab

Contiene:
- NormalizadorFechas: Gestión de conversión de fechas
- ValidadorEntradas: Validación de inputs del usuario
- GestorArchivos: Manejo de archivos CSV y Excel
"""

import pandas as pd
import os
from datetime import datetime
from pathlib import Path
import numpy as np
from scipy.optimize import minimize
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ============================================================================
# CLASE 1: NORMALIZADOR DE FECHAS
# ============================================================================

class NormalizadorFechas:
    """
    Especializado en normalización y validación de fechas.
    Garantiza que todas las fechas sean datetime64[ns] consistentemente.
    """
    
    # Formatos de fecha soportados
    FORMATOS_FECHA = [
        '%Y-%m-%d',    # 2023-01-15
        '%d/%m/%Y',    # 15/01/2023
        '%d-%m-%Y',    # 15-01-2023
        '%Y/%m/%d',    # 2023/01/15
        '%d.%m.%Y',    # 15.01.2023
        '%d/%m/%y',    # 15/01/23
        '%d-%m-%y',    # 15-01-23
    ]
    
    # Nombres de columnas de fecha comunes
    COLUMNAS_FECHA_COMUNES = [
        'Fecha_Inversion', 'Fecha_Compra', 'Fecha_Baja',
        'Date', 'date', 'Fecha', 'fecha', 'DATE',
        'Ultima_Actualizacion', 'fecha_inversion', 
        'fecha_ultimo_precio', 'fecha_compra'
    ]
    
    @staticmethod
    def normalizar_dataframe(df, columnas_fecha=None, verbose=True):
        """
        Normaliza todas las columnas de fecha en un DataFrame.
        
        Args:
            df: DataFrame a normalizar
            columnas_fecha: Lista de columnas a convertir. Si None, usa las comunes.
            verbose: Si True, muestra mensajes de progreso
            
        Returns:
            DataFrame con fechas normalizadas
        """
        if df is None or df.empty:
            return df
        
        df = df.copy()  # Evitar modificar el original
        
        # Determinar columnas a procesar
        if columnas_fecha is None:
            columnas_fecha = NormalizadorFechas.COLUMNAS_FECHA_COMUNES
        
        columnas_procesadas = 0
        
        for col in columnas_fecha:
            if col in df.columns:
                # Verificar si ya es datetime
                if not pd.api.types.is_datetime64_any_dtype(df[col]):
                    # Convertir a datetime
                    df[col] = pd.to_datetime(df[col], errors='coerce')
                    columnas_procesadas += 1
                    
                    # Contar NaT generados
                    num_nat = df[col].isna().sum()
                    if num_nat > 0 and verbose:
                        print(f"   ⚠️  '{col}': {num_nat} fechas inválidas → NaT")
        
        if verbose and columnas_procesadas > 0:
            print(f"   ✅ {columnas_procesadas} columnas de fecha normalizadas")
        
        return df
    
    @staticmethod
    def validar_fecha(fecha):
        """
        Valida que una fecha sea tipo datetime.
        
        Args:
            fecha: Fecha a validar
            
        Returns:
            True si es datetime válido, False en caso contrario
        """
        return pd.api.types.is_datetime64_any_dtype(pd.Series([fecha]))
    
    @staticmethod
    def convertir_a_datetime(fecha_input):
        """
        Convierte un string o valor a datetime con múltiples formatos.
        
        Args:
            fecha_input: String, datetime u otro tipo
            
        Returns:
            pd.Timestamp o None si falla la conversión
        """
        # Si ya es datetime, retornar
        if isinstance(fecha_input, (pd.Timestamp, datetime)):
            return pd.Timestamp(fecha_input)
        
        # Si es None o vacío
        if fecha_input is None or (isinstance(fecha_input, str) and not fecha_input.strip()):
            return None
        
        # Intentar conversión con pandas (más flexible)
        try:
            return pd.to_datetime(fecha_input, errors='coerce')
        except:
            pass
        
        # Intentar con formatos específicos
        if isinstance(fecha_input, str):
            for formato in NormalizadorFechas.FORMATOS_FECHA:
                try:
                    return pd.Timestamp(datetime.strptime(fecha_input, formato))
                except ValueError:
                    continue
        
        return None
    
    @staticmethod
    def solicitar_fecha_usuario(mensaje="Fecha (DD/MM/YYYY) o Enter para hoy: "):
        """
        Solicita una fecha al usuario con validación.
        
        Args:
            mensaje: Mensaje a mostrar
            
        Returns:
            pd.Timestamp con la fecha válida
        """
        while True:
            fecha_input = input(mensaje).strip()
            
            # Si vacío, usar fecha actual
            if not fecha_input:
                return pd.Timestamp.now()
            
            # Intentar convertir
            fecha = NormalizadorFechas.convertir_a_datetime(fecha_input)
            
            if fecha is not None and not pd.isna(fecha):
                print(f"   ✅ Fecha reconocida: {fecha.strftime('%Y-%m-%d')}")
                return fecha
            else:
                print("   ❌ Formato inválido. Formatos aceptados:")
                print("      • DD/MM/YYYY (ej: 15/01/2023)")
                print("      • DD-MM-YYYY (ej: 15-01-2023)")
                print("      • YYYY-MM-DD (ej: 2023-01-15)")
                print("      • O Enter para fecha actual")


# ============================================================================
# CLASE 2: VALIDADOR DE ENTRADAS
# ============================================================================

class ValidadorEntradas:
    """
    Maneja todas las validaciones de entrada del usuario.
    Simplifica el código de interfaz.
    """
    
    @staticmethod
    def solicitar_entero(mensaje, min_val=None, max_val=None):
        """
        Solicita un número entero con validación de rango.
        
        Args:
            mensaje: Mensaje a mostrar
            min_val: Valor mínimo permitido (opcional)
            max_val: Valor máximo permitido (opcional)
            
        Returns:
            int: Número válido
        """
        while True:
            try:
                valor = int(input(mensaje))
                
                if (min_val is None or valor >= min_val) and \
                   (max_val is None or valor <= max_val):
                    return valor
                else:
                    rango = f"{min_val if min_val is not None else '∞'}-{max_val if max_val is not None else '∞'}"
                    print(f"   ❌ Valor fuera de rango ({rango})")
            except ValueError:
                print("   ❌ Debe ingresar un número entero válido")
    
    @staticmethod
    def solicitar_decimal(mensaje, min_val=None):
        """
        Solicita un número decimal con validación.
        
        Args:
            mensaje: Mensaje a mostrar
            min_val: Valor mínimo permitido (opcional)
            
        Returns:
            float: Número válido
        """
        while True:
            try:
                valor = float(input(mensaje))
                
                if min_val is None or valor >= min_val:
                    return valor
                else:
                    print(f"   ❌ El valor debe ser ≥ {min_val}")
            except ValueError:
                print("   ❌ Debe ingresar un número válido")
    
    @staticmethod
    def solicitar_confirmacion(mensaje="¿Continuar? (s/n): "):
        """
        Solicita confirmación sí/no.
        
        Args:
            mensaje: Mensaje a mostrar
            
        Returns:
            bool: True si 's', False si 'n'
        """
        while True:
            respuesta = input(mensaje).strip().lower()
            if respuesta in ['s', 'si', 'sí', 'y', 'yes']:
                return True
            elif respuesta in ['n', 'no']:
                return False
            else:
                print("   ❌ Responda 's' o 'n'")
    
    @staticmethod
    def solicitar_texto(mensaje, no_vacio=True, max_length=None):
        """
        Solicita texto con validación.
        
        Args:
            mensaje: Mensaje a mostrar
            no_vacio: Si True, no permite strings vacíos
            max_length: Longitud máxima permitida
            
        Returns:
            str: Texto validado
        """
        while True:
            texto = input(mensaje).strip()
            
            if no_vacio and not texto:
                print("   ❌ El texto no puede estar vacío")
                continue
            
            if max_length and len(texto) > max_length:
                print(f"   ❌ Máximo {max_length} caracteres")
                continue
            
            return texto


# ============================================================================
# CLASE 3: GESTOR DE ARCHIVOS
# ============================================================================

class GestorArchivos:
    """
    Maneja toda la I/O de archivos: carga y guardado de CSVs y Excel.
    Centraliza el acceso a archivos para mejor mantenimiento.
    """
    
    def __init__(self, DIR_DATOS=None, DIR_CARTERAS=None):
        """
        Inicializa el gestor de archivos.
        
        Args:
            DIR_DATOS: Ruta al directorio de datos (opcional, usa CONFIGURACION si no se proporciona)
            DIR_CARTERAS: Ruta al directorio de carteras (opcional)
        """
        # Si no se proporcionan, usar CONFIGURACION
        if DIR_DATOS is None:
            if 'CONFIGURACION' in globals():
                DIR_DATOS = CONFIGURACION['dir_datos']
            else:
                raise ValueError("DIR_DATOS no proporcionado y CONFIGURACION no encontrada")
        
        if DIR_CARTERAS is None:
            if 'CONFIGURACION' in globals():
                DIR_CARTERAS = CONFIGURACION['dir_carteras']
            else:
                raise ValueError("DIR_CARTERAS no proporcionado y CONFIGURACION no encontrada")
        
        self.DIR_DATOS = Path(DIR_DATOS)
        self.DIR_CARTERAS = Path(DIR_CARTERAS)
        self.normalizador = NormalizadorFechas()
        
        # Crear directorios si no existen
        self.DIR_DATOS.mkdir(parents=True, exist_ok=True)
        self.DIR_CARTERAS.mkdir(parents=True, exist_ok=True)
    
    def cargar_datos_historicos(self, archivo_csv, verbose=True):
        """
        Carga datos históricos desde CSV con normalización de fechas.
        
        Args:
            archivo_csv: Nombre del archivo CSV
            verbose: Si True, muestra mensajes de progreso
            
        Returns:
            DataFrame con datos históricos o None si falla
        """
        # ✅ NO modificar el nombre del archivo, usarlo tal cual
        if not archivo_csv.endswith('.csv'):
            archivo_csv = f"{archivo_csv}.csv"
        
        # Buscar en múltiples ubicaciones
        rutas_posibles = [
            self.DIR_DATOS / archivo_csv,
            self.DIR_CARTERAS / archivo_csv,
            Path(archivo_csv)
        ]
        
        ruta = None
        for ruta_prueba in rutas_posibles:
            if ruta_prueba.exists():
                ruta = ruta_prueba
                if verbose:
                    print(f"   📍 Encontrado en: {ruta_prueba.parent.name}/{archivo_csv}")
                break
        
        if ruta is None:
            if verbose:
                print(f"   ❌ No encontrado: {archivo_csv}")
                print(f"   🔍 Buscado en:")
                for ruta_prueba in rutas_posibles:
                    print(f"      - {ruta_prueba}")
            return None
        
        try:
            # Leer CSV
            df = pd.read_csv(ruta)
            
            if df.empty:
                if verbose:
                    print(f"   ⚠️  Archivo vacío: {archivo_csv}")
                return None
            
            # Normalizar fechas
            df = self.normalizador.normalizar_dataframe(df, verbose=False)  # ← verbose=False para menos ruido
            
            # Verificar columnas necesarias
            if 'Date' not in df.columns:
                if verbose:
                    print(f"   ❌ Falta columna 'Date'")
                    print(f"      Columnas disponibles: {df.columns.tolist()}")
                return None
            
            # Ordenar por fecha
            df = df.sort_values('Date').reset_index(drop=True)
            
            if verbose:
                print(f"   ✅ {len(df)} registros: {df['Date'].min().date()} → {df['Date'].max().date()}")
            
            return df
            
        except Exception as e:
            if verbose:
                print(f"   ❌ Error al cargar: {e}")
                import traceback
                traceback.print_exc()
            return None
    
    def guardar_cartera_excel(self, df_cartera, ruta, df_adicionales=None):
        """
        Guarda la cartera en formato Excel con múltiples hojas.
        
        Args:
            df_cartera: DataFrame principal de la cartera
            ruta: Ruta del archivo a guardar
            df_adicionales: Dict con DataFrames adicionales {nombre_hoja: df}
            
        Returns:
            bool: True si se guardó correctamente
        """
        try:
            with pd.ExcelWriter(ruta, engine='openpyxl') as writer:
                # Hoja principal
                df_cartera.to_excel(writer, sheet_name='Sheet1', index=False)
                
                # Hojas adicionales
                if df_adicionales:
                    for nombre_hoja, df in df_adicionales.items():
                        if df is not None and not df.empty:
                            df.to_excel(writer, sheet_name=nombre_hoja, index=False)
            
            print(f"   ✅ Guardado: {Path(ruta).name}")
            return True
            
        except Exception as e:
            print(f"   ❌ Error al guardar: {e}")
            return False
    
    def cargar_cartera_excel(self, ruta, normalizar_fechas=True):
        """
        Carga una cartera desde Excel.
        
        Args:
            ruta: Ruta del archivo a cargar
            normalizar_fechas: Si True, normaliza fechas automáticamente
            
        Returns:
            dict: {'cartera': df, 'hojas_adicionales': {nombre: df}}
        """
        try:
            # Cargar hoja principal
            df_cartera = pd.read_excel(ruta, sheet_name='Sheet1')
            
            if normalizar_fechas:
                df_cartera = self.normalizador.normalizar_dataframe(df_cartera)
            
            # Intentar cargar otras hojas
            xls = pd.ExcelFile(ruta)
            hojas_adicionales = {}
            
            for nombre_hoja in xls.sheet_names:
                if nombre_hoja != 'Sheet1':
                    df = pd.read_excel(ruta, sheet_name=nombre_hoja)
                    if normalizar_fechas:
                        df = self.normalizador.normalizar_dataframe(df)
                    hojas_adicionales[nombre_hoja] = df
            
            return {
                'cartera': df_cartera,
                'hojas_adicionales': hojas_adicionales
            }
            
        except Exception as e:
            print(f"   ❌ Error al cargar cartera: {e}")
            return None
    
    def listar_carteras_disponibles(self):
        """
        Lista todas las carteras disponibles en el directorio.
        
        Returns:
            list: Lista de tuplas (nombre_cartera, ruta_completa)
        """
        patron = self.DIR_CARTERAS / "*.xlsx"
        archivos = list(self.DIR_CARTERAS.glob("*.xlsx"))
        
        carteras = []
        for archivo in archivos:
            nombre = archivo.stem
            # Excluir archivos de historial de bajas
            if not nombre.startswith('historial_bajas_'):
                carteras.append((nombre, str(archivo)))
        
        return sorted(carteras)
    
    def listar_archivos_csv_disponibles(self):
        """
        Lista todos los archivos CSV en el directorio de datos.
        
        Returns:
            list: Lista de nombres de archivos CSV
        """
        return sorted([f.name for f in self.DIR_DATOS.glob("*.csv")])


print("✅ Módulo 1 cargado: NormalizadorFechas, ValidadorEntradas, GestorArchivos")
```

# MÓDULO 2: CALCULADORES Y ANALIZADORES

```python
"""
MÓDULO 2: CALCULADORES Y ANALIZADORES
Ejecutar en Celda 2 de Jupyter Lab (después de Celda 1)

Contiene:
- CalculadorRentabilidades: Cálculo de rentabilidades de activos y benchmarks
- AnalizadorRiesgo: Análisis de riesgo, volatilidad, diversificación
"""

import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime


# ============================================================================
# CLASE 4: CALCULADOR DE RENTABILIDADES
# ============================================================================

class CalculadorRentabilidades:
    """
    Especializado en cálculo de rentabilidades de activos individuales,
    carteras completas y benchmarks.
    """
    
    def __init__(self, gestor_archivos):
        """
        Inicializa el calculador.
        
        Args:
            gestor_archivos: Instancia de GestorArchivos para cargar datos
        """
        self.gestor_archivos = gestor_archivos
    
    def calcular_rentabilidad_activo(self, row):
        """
        Calcula la rentabilidad de un activo individual.
        
        Args:
            row: Fila del DataFrame de cartera con datos del activo
            
        Returns:
            dict: Métricas de rentabilidad o None si falla
        """
        try:
            # Extraer datos
            ticker = row.get('ISIN', row.get('Ticker', 'Unknown'))
            archivo_csv = row['Fichero_CSV']
            fecha_inversion = row['Fecha_Inversion']
            cantidad_invertida = row['Cantidad_Invertida']
            
            # Validar fecha (ya debería ser datetime)
            if not isinstance(fecha_inversion, pd.Timestamp):
                fecha_inversion = pd.to_datetime(fecha_inversion, errors='coerce')
            
            if pd.isna(fecha_inversion):
                print(f"     ❌ Fecha inválida")
                return None
            
            # Cargar datos históricos
            df_historico = self.gestor_archivos.cargar_datos_historicos(
                archivo_csv, 
                verbose=False
            )
            
            if df_historico is None or df_historico.empty:
                print(f"     ❌ Sin datos históricos")
                return None
            
            # Filtrar desde fecha de inversión
            df_desde_inversion = df_historico[
                df_historico['Date'] >= fecha_inversion
            ].copy()
            
            if df_desde_inversion.empty:
                print(f"     ⚠️  Sin datos desde {fecha_inversion.date()}")
                return None
            
            # Obtener precios
            precio_inversion = df_desde_inversion['Close'].iloc[0]
            precio_actual = df_desde_inversion['Close'].iloc[-1]
            fecha_ultimo_precio = df_desde_inversion['Date'].iloc[-1]
            
            # Calcular métricas
            num_participaciones = cantidad_invertida / precio_inversion
            valor_actual = num_participaciones * precio_actual
            ganancia = valor_actual - cantidad_invertida
            rentabilidad_porcentual = (ganancia / cantidad_invertida * 100) if cantidad_invertida > 0 else 0
            
            return {
                'ticker': ticker,
                'precio_inversion': precio_inversion,
                'precio_actual': precio_actual,
                'num_participaciones': num_participaciones,
                'cantidad_invertida': cantidad_invertida,
                'valor_actual': valor_actual,
                'ganancia': ganancia,
                'rentabilidad_porcentual': rentabilidad_porcentual,
                'fecha_inversion': fecha_inversion,
                'fecha_ultimo_precio': fecha_ultimo_precio
            }
            
        except Exception as e:
            print(f"     ❌ Error: {e}")
            return None
    
    def calcular_rentabilidad_cartera(self, df_cartera, verbose=True):
        """
        Calcula rentabilidades de toda la cartera.
        
        Args:
            df_cartera: DataFrame con la cartera
            verbose: Si True, muestra progreso
            
        Returns:
            DataFrame con rentabilidades o None
        """
        if df_cartera.empty:
            if verbose:
                print("\n⚠️  La cartera está vacía")
            return None
        
        if verbose:
            print("\n" + "="*110)
            print("📊 ANÁLISIS DE RENTABILIDAD DE LA CARTERA")
            print("="*110)
            print("\n🔄 Calculando rentabilidades...")
        
        resultados = []
        total_invertido = 0
        total_valor_actual = 0
        activos_con_datos = 0
        activos_sin_datos = 0
        
        for idx, row in df_cartera.iterrows():
            if verbose:
                print(f"\n  📊 {row['Valor'][:35]}...")
                print(f"     📄 {row['Fichero_CSV']}")
                print(f"     📅 {row['Fecha_Inversion']}")
                print(f"     💰 {row['Cantidad_Invertida']:,.2f} €")
            
            rentabilidad = self.calcular_rentabilidad_activo(row)
            
            if rentabilidad:
                resultados.append({
                    'Valor': row['Valor'],
                    'Categoria': row['Categoria'],
                    'Invertido': row['Cantidad_Invertida'],
                    'Valor_Actual': rentabilidad['valor_actual'],
                    'Ganancia': rentabilidad['ganancia'],
                    'Rentabilidad_%': rentabilidad['rentabilidad_porcentual'],
                    'Fecha_Inversion': row['Fecha_Inversion'],
                    'Ultima_Actualizacion': rentabilidad['fecha_ultimo_precio']
                })
                
                total_invertido += row['Cantidad_Invertida']
                total_valor_actual += rentabilidad['valor_actual']
                activos_con_datos += 1
                
                if verbose:
                    print(f"     ✅ Rentabilidad: {rentabilidad['rentabilidad_porcentual']:+.2f}%")
            else:
                activos_sin_datos += 1
        
        if verbose:
            print(f"\n📈 Procesados: {activos_con_datos} ✅ / {activos_sin_datos} ❌")
        
        if not resultados:
            if verbose:
                print("\n⚠️  No se calcularon rentabilidades")
            return None
        
        df_rentabilidades = pd.DataFrame(resultados)
        
        # Mostrar resumen por categoría
        if verbose:
            for categoria in df_rentabilidades['Categoria'].unique():
                df_cat = df_rentabilidades[df_rentabilidades['Categoria'] == categoria]
                total_cat = df_cat['Invertido'].sum()
                rent_cat = df_cat['Rentabilidad_%'].mean()
                
                print(f"\n📊 {categoria.upper()}")
                print(f"   Invertido: {total_cat:,.2f} € | Rent. Media: {rent_cat:+.2f}%")
                print("-" * 110)
                
                for _, r in df_cat.iterrows():
                    simbolo = "🟢" if r['Rentabilidad_%'] >= 0 else "🔴"
                    print(f"  {simbolo} {r['Valor'][:40]:<40} | "
                          f"{r['Invertido']:>10,.2f} € → {r['Valor_Actual']:>10,.2f} € | "
                          f"{r['Ganancia']:>+10,.2f} € | {r['Rentabilidad_%']:>+7.2f}%")
            
            # Resumen total
            ganancia_total = total_valor_actual - total_invertido
            rent_total = (ganancia_total / total_invertido * 100) if total_invertido > 0 else 0
            
            print("\n" + "="*110)
            print("💰 RESUMEN TOTAL DE LA CARTERA")
            print("-" * 110)
            print(f"   Capital Invertido:  {total_invertido:>15,.2f} €")
            print(f"   Valor Actual:       {total_valor_actual:>15,.2f} €")
            print(f"   Ganancia/Pérdida:   {ganancia_total:>+15,.2f} €")
            print(f"   Rentabilidad:       {rent_total:>+15.2f} %")
            print("="*110)
        
        return df_rentabilidades
    
    def calcular_rentabilidad_benchmark(self, ticker, fecha_inicio, fecha_fin=None):
        """
        Calcula la rentabilidad de un benchmark.
        
        Args:
            ticker: Símbolo del benchmark (ej: '^STOXX50E')
            fecha_inicio: Fecha de inicio (pd.Timestamp)
            fecha_fin: Fecha final (pd.Timestamp). Si None, usa hoy.
            
        Returns:
            dict con rentabilidad del benchmark o None
        """
        if fecha_fin is None:
            fecha_fin = pd.Timestamp.now()
        
        # Asegurar que son Timestamps
        if not isinstance(fecha_inicio, pd.Timestamp):
            fecha_inicio = pd.to_datetime(fecha_inicio)
        if not isinstance(fecha_fin, pd.Timestamp):
            fecha_fin = pd.to_datetime(fecha_fin)
        
        print(f"\n📊 Calculando benchmark {ticker}...")
        print(f"📅 Período: {fecha_inicio.date()} → {fecha_fin.date()}")
        
        try:
            # Descargar datos
            print(f"⬇️  Descargando datos...")
            benchmark_data = yf.download(
                ticker,
                start=fecha_inicio,
                end=fecha_fin,
                progress=False,
                auto_adjust=True
            )
            
            if benchmark_data.empty:
                print(f"❌ Sin datos para {ticker}")
                return None
            
            # Obtener precios (manejar MultiIndex)
            if isinstance(benchmark_data.columns, pd.MultiIndex):
                if 'Adj Close' in benchmark_data.columns.levels[0]:
                    precios = benchmark_data['Adj Close'].iloc[:, 0]
                else:
                    precios = benchmark_data['Close'].iloc[:, 0]
            else:
                precios = benchmark_data.get('Adj Close', benchmark_data['Close'])
            
            # Calcular rentabilidad
            precio_inicial = float(precios.iloc[0])
            precio_final = float(precios.iloc[-1])
            rentabilidad = ((precio_final - precio_inicial) / precio_inicial) * 100
            
            print(f"✅ Rentabilidad {ticker}: {rentabilidad:+.2f}%")
            print(f"   Precio inicial: {precio_inicial:.2f}")
            print(f"   Precio final: {precio_final:.2f}")
            
            return {
                'ticker': ticker,
                'rentabilidad': rentabilidad,
                'precio_inicial': precio_inicial,
                'precio_final': precio_final,
                'fecha_inicio': fecha_inicio,
                'fecha_fin': fecha_fin
            }
            
        except Exception as e:
            print(f"❌ Error: {e}")
            return None


# ============================================================================
# CLASE 5: ANALIZADOR DE RIESGO
# ============================================================================

class AnalizadorRiesgo:
    """
    Análisis de riesgo: concentración, diversificación, volatilidad, Sharpe.
    """
    
    def __init__(self, tasa_libre_riesgo=0.02):
        """
        Args:
            tasa_libre_riesgo: Tasa anual libre de riesgo (default 2%)
        """
        self.tasa_libre_riesgo = tasa_libre_riesgo
    
    def analizar_concentracion(self, df_cartera):
        """
        Analiza la concentración de la cartera.
        
        Args:
            df_cartera: DataFrame con la cartera
            
        Returns:
            dict con métricas de concentración
        """
        total_invertido = df_cartera['Cantidad_Invertida'].sum()
        
        # Top 5 posiciones
        df_top5 = df_cartera.nlargest(5, 'Cantidad_Invertida')
        peso_top5 = (df_top5['Cantidad_Invertida'].sum() / total_invertido * 100)
        
        # Calcular índice Herfindahl
        pesos = (df_cartera['Cantidad_Invertida'] / total_invertido) * 100
        herfindahl = (pesos ** 2).sum()
        
        return {
            'peso_top5': peso_top5,
            'num_posiciones': len(df_cartera),
            'herfindahl': herfindahl,
            'advertencia_concentracion': peso_top5 > 50
        }
    
    def analizar_diversificacion(self, df_cartera):
        """
        Analiza la diversificación por categorías.
        
        Args:
            df_cartera: DataFrame con la cartera
            
        Returns:
            DataFrame con distribución por categoría
        """
        total_invertido = df_cartera['Cantidad_Invertida'].sum()
        
        dist_categoria = df_cartera.groupby('Categoria').agg({
            'Cantidad_Invertida': ['sum', 'count']
        })
        
        dist_categoria.columns = ['Total_Invertido', 'Num_Posiciones']
        dist_categoria['Peso_%'] = (dist_categoria['Total_Invertido'] / total_invertido * 100)
        dist_categoria = dist_categoria.sort_values('Peso_%', ascending=False)
        
        return dist_categoria
    
    def calcular_volatilidad(self, df_rentabilidades):
        """
        Calcula la volatilidad de la cartera.
        
        Args:
            df_rentabilidades: DataFrame con rentabilidades
            
        Returns:
            dict con métricas de volatilidad
        """
        volatilidad = df_rentabilidades['Rentabilidad_%'].std()
        rent_media = df_rentabilidades['Rentabilidad_%'].mean()
        rent_max = df_rentabilidades['Rentabilidad_%'].max()
        rent_min = df_rentabilidades['Rentabilidad_%'].min()
        
        return {
            'volatilidad': volatilidad,
            'rentabilidad_media': rent_media,
            'rentabilidad_max': rent_max,
            'rentabilidad_min': rent_min,
            'rango': rent_max - rent_min
        }
    
    def calcular_sharpe(self, rentabilidad_anual, volatilidad_anual):
        """
        Calcula el ratio de Sharpe.
        
        Args:
            rentabilidad_anual: Rentabilidad anualizada (%)
            volatilidad_anual: Volatilidad anualizada (%)
            
        Returns:
            float: Ratio de Sharpe
        """
        if volatilidad_anual == 0:
            return 0
        
        exceso_rentabilidad = rentabilidad_anual - (self.tasa_libre_riesgo * 100)
        return exceso_rentabilidad / volatilidad_anual
    
    def analisis_completo(self, df_cartera, df_rentabilidades):
        """
        Realiza un análisis de riesgo completo.
        
        Args:
            df_cartera: DataFrame con la cartera
            df_rentabilidades: DataFrame con rentabilidades
            
        Returns:
            dict con todos los análisis
        """
        concentracion = self.analizar_concentracion(df_cartera)
        diversificacion = self.analizar_diversificacion(df_cartera)
        volatilidad = self.calcular_volatilidad(df_rentabilidades)
        
        sharpe = self.calcular_sharpe(
            volatilidad['rentabilidad_media'],
            volatilidad['volatilidad']
        ) if volatilidad['volatilidad'] > 0 else 0
        
        # Contar posiciones negativas
        posiciones_negativas = len(df_rentabilidades[df_rentabilidades['Rentabilidad_%'] < 0])
        pct_negativas = (posiciones_negativas / len(df_rentabilidades) * 100)
        
        return {
            'concentracion': concentracion,
            'diversificacion': diversificacion,
            'volatilidad': volatilidad,
            'sharpe': sharpe,
            'posiciones_negativas': posiciones_negativas,
            'pct_posiciones_negativas': pct_negativas
        }


print("✅ Módulo 2 cargado: CalculadorRentabilidades, AnalizadorRiesgo")
```

# MÓDULO 3: GENERADORES DE GRÁFICOS Y REPORTES

```python
"""
MÓDULO 3: GENERADORES DE GRÁFICOS Y REPORTES
Ejecutar en Celda 3 de Jupyter Lab (después de Celda 2)

Contiene:
- GeneradorGraficos: Visualizaciones de la cartera
- GeneradorReportes: Generación de informes y exportaciones
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
import json
from pathlib import Path


# ============================================================================
# CLASE 6: GENERADOR DE GRÁFICOS
# ============================================================================

class GeneradorGraficos:
    """
    Genera todas las visualizaciones de la cartera:
    distribución, rentabilidades, comparaciones, etc.
    """
    
    def __init__(self, directorio_salida=None):
        """
        Args:
            directorio_salida: Directorio donde guardar gráficos
        """
        self.directorio_salida = Path(directorio_salida) if directorio_salida else Path.cwd()
        self.directorio_salida.mkdir(parents=True, exist_ok=True)
    
    def grafico_distribucion_categoria(self, df_cartera, mostrar=True, guardar=True):
        """
        Gráfico de distribución por categoría (pie chart).
        
        Args:
            df_cartera: DataFrame con la cartera
            mostrar: Si True, muestra el gráfico
            guardar: Si True, guarda el gráfico
            
        Returns:
            Path del archivo guardado o None
        """
        fig, ax = plt.subplots(figsize=(10, 7))
        
        distribucion = df_cartera.groupby('Categoria')['Cantidad_Invertida'].sum()
        colors = plt.cm.Set3(range(len(distribucion)))
        
        ax.pie(distribucion.values, labels=distribucion.index, autopct='%1.1f%%', 
               colors=colors, startangle=90)
        ax.set_title('Distribución de la Cartera por Categoría', 
                     fontsize=14, fontweight='bold')
        
        ruta = None
        if guardar:
            nombre = f'distribucion_categoria_{datetime.now().strftime("%Y%m%d")}.png'
            ruta = self.directorio_salida / nombre
            plt.savefig(ruta, dpi=300, bbox_inches='tight')
            print(f"   ✅ Guardado: {nombre}")
        
        if mostrar:
            plt.show()
        else:
            plt.close()
        
        return ruta
    
    def grafico_top_inversiones(self, df_cartera, top_n=10, mostrar=True, guardar=True):
        """
        Gráfico de barras horizontales con top inversiones.
        
        Args:
            df_cartera: DataFrame con la cartera
            top_n: Número de activos a mostrar
            mostrar: Si True, muestra el gráfico
            guardar: Si True, guarda el gráfico
            
        Returns:
            Path del archivo guardado o None
        """
        fig, ax = plt.subplots(figsize=(12, 8))
        
        top = df_cartera.nlargest(top_n, 'Cantidad_Invertida')
        valores_cortos = [v[:30] for v in top['Valor']]
        
        ax.barh(valores_cortos, top['Cantidad_Invertida'], color='steelblue')
        ax.set_xlabel('Cantidad Invertida (€)', fontsize=11)
        ax.set_title(f'Top {top_n} Activos por Inversión', 
                     fontsize=14, fontweight='bold')
        ax.invert_yaxis()
        ax.grid(axis='x', alpha=0.3)
        
        ruta = None
        if guardar:
            nombre = f'top_inversiones_{datetime.now().strftime("%Y%m%d")}.png'
            ruta = self.directorio_salida / nombre
            plt.savefig(ruta, dpi=300, bbox_inches='tight')
            print(f"   ✅ Guardado: {nombre}")
        
        if mostrar:
            plt.show()
        else:
            plt.close()
        
        return ruta
    
    def grafico_rentabilidades(self, df_rentabilidades, top_n=10, mostrar=True, guardar=True):
        """
        Gráfico de rentabilidades por activo.
        
        Args:
            df_rentabilidades: DataFrame con rentabilidades
            top_n: Número de activos a mostrar
            mostrar: Si True, muestra el gráfico
            guardar: Si True, guarda el gráfico
            
        Returns:
            Path del archivo guardado o None
        """
        fig, ax = plt.subplots(figsize=(12, 8))
        
        # Ordenar por rentabilidad
        df_sorted = df_rentabilidades.sort_values('Rentabilidad_%', ascending=False).head(top_n)
        colores = ['green' if x >= 0 else 'red' for x in df_sorted['Rentabilidad_%']]
        valores_cortos = [v[:30] for v in df_sorted['Valor']]
        
        ax.barh(valores_cortos, df_sorted['Rentabilidad_%'], color=colores, alpha=0.7)
        ax.set_xlabel('Rentabilidad (%)', fontsize=11)
        ax.set_title(f'Top {top_n} Rentabilidades', fontsize=14, fontweight='bold')
        ax.axvline(x=0, color='black', linestyle='-', linewidth=0.8)
        ax.invert_yaxis()
        ax.grid(axis='x', alpha=0.3)
        
        ruta = None
        if guardar:
            nombre = f'rentabilidades_{datetime.now().strftime("%Y%m%d")}.png'
            ruta = self.directorio_salida / nombre
            plt.savefig(ruta, dpi=300, bbox_inches='tight')
            print(f"   ✅ Guardado: {nombre}")
        
        if mostrar:
            plt.show()
        else:
            plt.close()
        
        return ruta
    
    def grafico_valor_actual_vs_invertido(self, df_rentabilidades, mostrar=True, guardar=True):
        """
        Scatter plot: valor invertido vs valor actual.
        
        Args:
            df_rentabilidades: DataFrame con rentabilidades
            mostrar: Si True, muestra el gráfico
            guardar: Si True, guarda el gráfico
            
        Returns:
            Path del archivo guardado o None
        """
        fig, ax = plt.subplots(figsize=(10, 8))
        
        scatter = ax.scatter(df_rentabilidades['Invertido'], 
                           df_rentabilidades['Valor_Actual'],
                           c=df_rentabilidades['Rentabilidad_%'], 
                           cmap='RdYlGn', s=100, alpha=0.6, edgecolors='black')
        
        # Línea de referencia (sin cambios)
        max_val = max(df_rentabilidades['Invertido'].max(), 
                     df_rentabilidades['Valor_Actual'].max())
        ax.plot([0, max_val], [0, max_val], 'k--', alpha=0.3, label='Sin cambios')
        
        ax.set_xlabel('Capital Invertido (€)', fontsize=11)
        ax.set_ylabel('Valor Actual (€)', fontsize=11)
        ax.set_title('Capital Invertido vs Valor Actual', fontsize=14, fontweight='bold')
        ax.legend()
        ax.grid(True, alpha=0.3)
        
        # Colorbar
        cbar = plt.colorbar(scatter, ax=ax)
        cbar.set_label('Rentabilidad (%)', fontsize=10)
        
        ruta = None
        if guardar:
            nombre = f'valor_invertido_vs_actual_{datetime.now().strftime("%Y%m%d")}.png'
            ruta = self.directorio_salida / nombre
            plt.savefig(ruta, dpi=300, bbox_inches='tight')
            print(f"   ✅ Guardado: {nombre}")
        
        if mostrar:
            plt.show()
        else:
            plt.close()
        
        return ruta
    
    def grafico_distribucion_rentabilidades(self, df_rentabilidades, mostrar=True, guardar=True):
        """
        Histograma de distribución de rentabilidades.
        
        Args:
            df_rentabilidades: DataFrame con rentabilidades
            mostrar: Si True, muestra el gráfico
            guardar: Si True, guarda el gráfico
            
        Returns:
            Path del archivo guardado o None
        """
        fig, ax = plt.subplots(figsize=(10, 7))
        
        ax.hist(df_rentabilidades['Rentabilidad_%'], bins=15, 
                color='skyblue', edgecolor='black', alpha=0.7)
        ax.axvline(x=0, color='red', linestyle='--', linewidth=2, label='Break-even')
        ax.axvline(x=df_rentabilidades['Rentabilidad_%'].mean(), color='green',
                  linestyle='--', linewidth=2, 
                  label=f'Media: {df_rentabilidades["Rentabilidad_%"].mean():.2f}%')
        
        ax.set_xlabel('Rentabilidad (%)', fontsize=11)
        ax.set_ylabel('Número de Activos', fontsize=11)
        ax.set_title('Distribución de Rentabilidades', fontsize=14, fontweight='bold')
        ax.legend()
        ax.grid(True, alpha=0.3)
        
        ruta = None
        if guardar:
            nombre = f'distribucion_rentabilidades_{datetime.now().strftime("%Y%m%d")}.png'
            ruta = self.directorio_salida / nombre
            plt.savefig(ruta, dpi=300, bbox_inches='tight')
            print(f"   ✅ Guardado: {nombre}")
        
        if mostrar:
            plt.show()
        else:
            plt.close()
        
        return ruta
    
    def dashboard_completo(self, df_cartera, df_rentabilidades, mostrar=True, guardar=True):
        """
        Dashboard completo con 6 gráficos.
        
        Args:
            df_cartera: DataFrame con la cartera
            df_rentabilidades: DataFrame con rentabilidades
            mostrar: Si True, muestra el gráfico
            guardar: Si True, guarda el gráfico
            
        Returns:
            Path del archivo guardado o None
        """
        print("\n📊 Generando dashboard completo...")
        
        fig = plt.figure(figsize=(16, 10))
        
        # 1. Distribución por categoría
        ax1 = plt.subplot(2, 3, 1)
        distribucion = df_cartera.groupby('Categoria')['Cantidad_Invertida'].sum()
        colors = plt.cm.Set3(range(len(distribucion)))
        ax1.pie(distribucion.values, labels=distribucion.index, autopct='%1.1f%%',
               colors=colors, startangle=90)
        ax1.set_title('Distribución por Categoría', fontsize=12, fontweight='bold')
        
        # 2. Top inversiones
        ax2 = plt.subplot(2, 3, 2)
        top = df_cartera.nlargest(10, 'Cantidad_Invertida')
        valores_cortos = [v[:25] for v in top['Valor']]
        ax2.barh(valores_cortos, top['Cantidad_Invertida'], color='steelblue')
        ax2.set_xlabel('€')
        ax2.set_title('Top 10 Inversiones', fontsize=12, fontweight='bold')
        ax2.invert_yaxis()
        
        # 3. Rentabilidades
        ax3 = plt.subplot(2, 3, 3)
        df_sorted = df_rentabilidades.sort_values('Rentabilidad_%', ascending=False).head(10)
        colores = ['green' if x >= 0 else 'red' for x in df_sorted['Rentabilidad_%']]
        valores_cortos = [v[:25] for v in df_sorted['Valor']]
        ax3.barh(valores_cortos, df_sorted['Rentabilidad_%'], color=colores, alpha=0.7)
        ax3.set_xlabel('%')
        ax3.set_title('Top 10 Rentabilidades', fontsize=12, fontweight='bold')
        ax3.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
        ax3.invert_yaxis()
        
        # 4. Valor invertido vs actual
        ax4 = plt.subplot(2, 3, 4)
        ax4.scatter(df_rentabilidades['Invertido'], df_rentabilidades['Valor_Actual'],
                   c=df_rentabilidades['Rentabilidad_%'], cmap='RdYlGn', s=80, alpha=0.6)
        max_val = max(df_rentabilidades['Invertido'].max(), 
                     df_rentabilidades['Valor_Actual'].max())
        ax4.plot([0, max_val], [0, max_val], 'k--', alpha=0.3)
        ax4.set_xlabel('Invertido (€)')
        ax4.set_ylabel('Valor Actual (€)')
        ax4.set_title('Invertido vs Actual', fontsize=12, fontweight='bold')
        ax4.grid(True, alpha=0.3)
        
        # 5. Distribución rentabilidades
        ax5 = plt.subplot(2, 3, 5)
        ax5.hist(df_rentabilidades['Rentabilidad_%'], bins=15, 
                color='skyblue', edgecolor='black', alpha=0.7)
        ax5.axvline(x=0, color='red', linestyle='--', linewidth=1.5)
        ax5.axvline(x=df_rentabilidades['Rentabilidad_%'].mean(), 
                   color='green', linestyle='--', linewidth=1.5)
        ax5.set_xlabel('Rentabilidad (%)')
        ax5.set_ylabel('Frecuencia')
        ax5.set_title('Distribución Rentabilidades', fontsize=12, fontweight='bold')
        
        # 6. Ganancias absolutas
        ax6 = plt.subplot(2, 3, 6)
        df_ganancia = df_rentabilidades.sort_values('Ganancia', ascending=False).head(10)
        colores = ['green' if x >= 0 else 'red' for x in df_ganancia['Ganancia']]
        valores_cortos = [v[:25] for v in df_ganancia['Valor']]
        ax6.barh(valores_cortos, df_ganancia['Ganancia'], color=colores, alpha=0.7)
        ax6.set_xlabel('Ganancia (€)')
        ax6.set_title('Top 10 Ganancias Absolutas', fontsize=12, fontweight='bold')
        ax6.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
        ax6.invert_yaxis()
        
        plt.tight_layout()
        
        ruta = None
        if guardar:
            nombre = f'dashboard_completo_{datetime.now().strftime("%Y%m%d_%H%M%S")}.png'
            ruta = self.directorio_salida / nombre
            plt.savefig(ruta, dpi=300, bbox_inches='tight')
            print(f"   ✅ Dashboard guardado: {nombre}")
        
        if mostrar:
            plt.show()
        else:
            plt.close()
        
        return ruta


# ============================================================================
# CLASE 7: GENERADOR DE REPORTES
# ============================================================================

class GeneradorReportes:
    """
    Genera reportes y exportaciones en diferentes formatos:
    Excel, CSV, JSON, texto.
    """
    
    def __init__(self, directorio_salida=None):
        """
        Args:
            directorio_salida: Directorio donde guardar reportes
        """
        self.directorio_salida = Path(directorio_salida) if directorio_salida else Path.cwd()
        self.directorio_salida.mkdir(parents=True, exist_ok=True)
    
    def generar_resumen_ejecutivo_texto(self, df_cartera, df_rentabilidades, 
                                       guardar=True, mostrar=True):
        """
        Genera resumen ejecutivo en formato texto.
        
        Args:
            df_cartera: DataFrame con la cartera
            df_rentabilidades: DataFrame con rentabilidades
            guardar: Si True, guarda el archivo
            mostrar: Si True, imprime en consola
            
        Returns:
            str con el contenido del resumen
        """
        # Cálculos
        total_invertido = df_rentabilidades['Invertido'].sum()
        valor_actual = df_rentabilidades['Valor_Actual'].sum()
        ganancia = valor_actual - total_invertido
        rent_total = (ganancia / total_invertido * 100) if total_invertido > 0 else 0
        
        posiciones_pos = len(df_rentabilidades[df_rentabilidades['Rentabilidad_%'] >= 0])
        posiciones_neg = len(df_rentabilidades[df_rentabilidades['Rentabilidad_%'] < 0])
        tasa_exito = (posiciones_pos / len(df_rentabilidades) * 100) if len(df_rentabilidades) > 0 else 0
        
        # Generar texto
        resumen = f"""
{'='*90}
📊 RESUMEN EJECUTIVO DE LA CARTERA
{'='*90}

Fecha del reporte: {datetime.now().strftime('%d/%m/%Y %H:%M')}
{'-'*90}

1️⃣  VISIÓN GENERAL
   Capital Invertido:        {total_invertido:>15,.2f} €
   Valor Actual:             {valor_actual:>15,.2f} €
   Ganancia/Pérdida:         {ganancia:>+15,.2f} €
   Rentabilidad Total:       {rent_total:>+15.2f} %
   Número de Posiciones:     {len(df_rentabilidades):>15}

2️⃣  DISTRIBUCIÓN POR CATEGORÍA
"""
        for categoria in df_rentabilidades['Categoria'].unique():
            df_cat = df_rentabilidades[df_rentabilidades['Categoria'] == categoria]
            peso = (df_cat['Invertido'].sum() / total_invertido * 100)
            rent_cat = df_cat['Rentabilidad_%'].mean()
            resumen += f"   {categoria:<25} {peso:>6.2f}%  (Rent: {rent_cat:>+7.2f}%)\n"
        
        resumen += f"""
3️⃣  ANÁLISIS DE PERFORMANCE
   Posiciones Positivas:     {posiciones_pos:>15} ({tasa_exito:.1f}%)
   Posiciones Negativas:     {posiciones_neg:>15}
   Mejor Rentabilidad:       {df_rentabilidades['Rentabilidad_%'].max():>+15.2f} %
   Peor Rentabilidad:        {df_rentabilidades['Rentabilidad_%'].min():>+15.2f} %
   Rentabilidad Media:       {df_rentabilidades['Rentabilidad_%'].mean():>+15.2f} %

4️⃣  TOP 3 MEJORES POSICIONES
"""
        for _, row in df_rentabilidades.nlargest(3, 'Rentabilidad_%').iterrows():
            resumen += f"   • {row['Valor'][:40]:<40} {row['Rentabilidad_%']:>+7.2f}% ({row['Ganancia']:>+10,.2f} €)\n"
        
        resumen += "\n5️⃣  TOP 3 PEORES POSICIONES\n"
        for _, row in df_rentabilidades.nsmallest(3, 'Rentabilidad_%').iterrows():
            resumen += f"   • {row['Valor'][:40]:<40} {row['Rentabilidad_%']:>+7.2f}% ({row['Ganancia']:>+10,.2f} €)\n"
        
        resumen += f"\n{'='*90}\n"
        
        if mostrar:
            print(resumen)
        
        if guardar:
            nombre = f'resumen_ejecutivo_{datetime.now().strftime("%Y%m%d_%H%M%S")}.txt'
            ruta = self.directorio_salida / nombre
            with open(ruta, 'w', encoding='utf-8') as f:
                f.write(resumen)
            print(f"\n✅ Resumen guardado: {nombre}")
        
        return resumen
    
    def exportar_excel_completo(self, df_cartera, df_rentabilidades, 
                                df_bajas=None, nombre_archivo=None):
        """
        Exporta análisis completo a Excel con múltiples hojas.
        
        Args:
            df_cartera: DataFrame con la cartera
            df_rentabilidades: DataFrame con rentabilidades
            df_bajas: DataFrame con bajas (opcional)
            nombre_archivo: Nombre del archivo (si None, genera automático)
            
        Returns:
            Path del archivo guardado
        """
        if nombre_archivo is None:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            nombre_archivo = f'cartera_completa_{timestamp}.xlsx'
        
        ruta = self.directorio_salida / nombre_archivo
        
        print(f"\n📊 Exportando a Excel...")
        
        with pd.ExcelWriter(ruta, engine='openpyxl') as writer:
            # Hoja 1: Cartera
            df_cartera.to_excel(writer, sheet_name='Cartera', index=False)
            
            # Hoja 2: Rentabilidades
            if df_rentabilidades is not None and not df_rentabilidades.empty:
                df_rentabilidades.to_excel(writer, sheet_name='Rentabilidades', index=False)
            
            # Hoja 3: Resumen por categoría
            if not df_cartera.empty:
                resumen = df_cartera.groupby('Categoria').agg({
                    'Cantidad_Invertida': ['sum', 'count', 'mean']
                }).round(2)
                resumen.to_excel(writer, sheet_name='Resumen_Categoria')
            
            # Hoja 4: Bajas
            if df_bajas is not None and not df_bajas.empty:
                df_bajas.to_excel(writer, sheet_name='Bajas', index=False)
            
            # Hoja 5: Estadísticas
            if df_rentabilidades is not None and not df_rentabilidades.empty:
                total_inv = df_rentabilidades['Invertido'].sum()
                total_actual = df_rentabilidades['Valor_Actual'].sum()
                
                estadisticas = pd.DataFrame({
                    'Métrica': [
                        'Total Invertido', 'Valor Actual', 'Ganancia Total',
                        'Rentabilidad Media', 'Mejor Rentabilidad', 'Peor Rentabilidad',
                        'Número de Posiciones', 'Posiciones Positivas', 'Posiciones Negativas'
                    ],
                    'Valor': [
                        f"{total_inv:.2f} €",
                        f"{total_actual:.2f} €",
                        f"{(total_actual - total_inv):.2f} €",
                        f"{df_rentabilidades['Rentabilidad_%'].mean():.2f}%",
                        f"{df_rentabilidades['Rentabilidad_%'].max():.2f}%",
                        f"{df_rentabilidades['Rentabilidad_%'].min():.2f}%",
                        len(df_rentabilidades),
                        len(df_rentabilidades[df_rentabilidades['Rentabilidad_%'] >= 0]),
                        len(df_rentabilidades[df_rentabilidades['Rentabilidad_%'] < 0])
                    ]
                })
                estadisticas.to_excel(writer, sheet_name='Estadisticas', index=False)
        
        print(f"✅ Excel exportado: {nombre_archivo}")
        return ruta
    
    def exportar_json(self, df_cartera, df_rentabilidades=None, 
                     df_bajas=None, nombre_archivo=None):
        """
        Exporta datos a JSON.
        
        Args:
            df_cartera: DataFrame con la cartera
            df_rentabilidades: DataFrame con rentabilidades (opcional)
            df_bajas: DataFrame con bajas (opcional)
            nombre_archivo: Nombre del archivo (si None, genera automático)
            
        Returns:
            Path del archivo guardado
        """
        if nombre_archivo is None:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            nombre_archivo = f'cartera_{timestamp}.json'
        
        ruta = self.directorio_salida / nombre_archivo
        
        datos = {
            'fecha_exportacion': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
            'cartera': df_cartera.to_dict('records')
        }
        
        if df_rentabilidades is not None and not df_rentabilidades.empty:
            datos['rentabilidades'] = df_rentabilidades.to_dict('records')
        
        if df_bajas is not None and not df_bajas.empty:
            datos['bajas'] = df_bajas.to_dict('records')
        
        with open(ruta, 'w', encoding='utf-8') as f:
            json.dump(datos, f, ensure_ascii=False, indent=2, default=str)
        
        print(f"✅ JSON exportado: {nombre_archivo}")
        return ruta


print("✅ Módulo 3 cargado: GeneradorGraficos, GeneradorReportes")
```

# MÓDULO 4: GESTOR PRINCIPAL E INTERFAZ DE USUARIO

```python
"""
MÓDULO 4: GESTOR PRINCIPAL E INTERFAZ DE USUARIO
Ejecutar en Celda 4 de Jupyter Lab (después de Celda 3)

Contiene:
- GestorCartera: Clase principal que coordina todos los componentes
- InterfazUsuario: Menú principal y gestión de interacciones
"""
# ============================================================================
# CLASE 8: GESTOR DE CARTERA (COORDINADOR PRINCIPAL)
# ============================================================================

class GestorCartera:
    """
    Clase principal que coordina todos los componentes del sistema.
    Orquesta las operaciones entre GestorArchivos, Calculadores, 
    Analizadores y Generadores.
    """
    
    def __init__(self, directorio_datos=None, directorio_carteras=None, 
                 directorio_exports=None, directorio_diagnosticos=None,
                 benchmark_ticker='^STOXX50E'):
        """
        Inicializa el gestor de cartera con todos sus componentes.
        
        Args:
            directorio_datos: Ruta a datos históricos (opcional, usa CONFIGURACION)
            directorio_carteras: Ruta a carteras guardadas (opcional, usa CONFIGURACION)
            directorio_exports: Ruta para guardar informes (opcional, usa CONFIGURACION)
            directorio_diagnosticos: Ruta para diagnósticos (opcional)
            benchmark_ticker: Ticker del benchmark (default: ^STOXX50E)
        """
        # Si no se proporcionan, usar CONFIGURACION
        if directorio_datos is None:
            if 'CONFIGURACION' in globals():
                directorio_datos = CONFIGURACION['dir_datos']
            else:
                raise ValueError("directorio_datos no proporcionado y CONFIGURACION no encontrada")
        
        if directorio_carteras is None:
            if 'CONFIGURACION' in globals():
                directorio_carteras = CONFIGURACION['dir_carteras']
            else:
                raise ValueError("directorio_carteras no proporcionado y CONFIGURACION no encontrada")
        
        if directorio_exports is None:
            if 'CONFIGURACION' in globals():
                directorio_exports = CONFIGURACION['dir_exports']
            else:
                directorio_exports = Path(directorio_carteras) / 'exports'
        
        if directorio_diagnosticos is None:
            directorio_diagnosticos = Path(directorio_carteras) / 'diagnosticos'
        
        # Directorios
        self.directorio_datos = Path(directorio_datos)
        self.directorio_carteras = Path(directorio_carteras)
        self.DIR_DATOS = self.directorio_datos  # Alias para compatibilidad
        self.DIR_CARTERAS = self.directorio_carteras  # Alias para compatibilidad
        self.DIR_EXPORTS = Path(directorio_exports)
        self.directorio_diagnosticos = Path(directorio_diagnosticos)
        
        # Crear directorios
        for directorio in [self.DIR_DATOS, self.DIR_CARTERAS, 
                          self.DIR_EXPORTS, self.directorio_diagnosticos]:
            directorio.mkdir(parents=True, exist_ok=True)
        
        # Componentes del sistema
        self.gestor_archivos = GestorArchivos(
            DIR_DATOS=self.DIR_DATOS,
            DIR_CARTERAS=self.DIR_CARTERAS
        )
        self.normalizador = NormalizadorFechas()
        self.validador = ValidadorEntradas()
        self.calculador = CalculadorRentabilidades(self.gestor_archivos)
        self.analizador_riesgo = AnalizadorRiesgo()
        self.generador_graficos = GeneradorGraficos(str(self.DIR_EXPORTS))
        self.generador_reportes = GeneradorReportes(str(self.DIR_EXPORTS))
        
        # DataFrames principales
        self.df_cartera = pd.DataFrame()
        self.df_bajas = pd.DataFrame()
        
        # Archivos actuales
        self.archivo_cartera_actual = None
        self.archivo_bajas_actual = None
        self.nombre_cartera_actual = None
        
        # Datos
        self.asset_data = {}
        self.lista_df = []
        
        # Configuración
        self.benchmark_ticker = benchmark_ticker
        
    def normalizar_todas_fechas(self):
        """Normaliza fechas en todos los DataFrames."""
        if not self.df_cartera.empty:
            self.df_cartera = self.normalizador.normalizar_dataframe(
                self.df_cartera, verbose=True
            )
        
        if not self.df_bajas.empty:
            self.df_bajas = self.normalizador.normalizar_dataframe(
                self.df_bajas, verbose=True
            )
    
    # ============================================================================
    # MODIFICAR cargar_cartera() para limpiar al cargar
    # ============================================================================
    
    def cargar_cartera(self, nombre_cartera):
        """
        Carga una cartera desde Excel.
        
        Características:
        - No convierte primera columna en índice
        - Limpia columnas duplicadas automáticamente
        - NO actualiza valores (usar opción 13 manualmente)
        """
        # Construir ruta del archivo
        archivo = self.DIR_CARTERAS / f'{nombre_cartera}.xlsx'
        
        if not archivo.exists():
            print(f"\n❌ No existe la cartera '{nombre_cartera}'")
            return False
        
        try:
            # ================================================================
            # CARGAR EXCEL (intentar diferentes nombres de hoja)
            # ================================================================
            df_loaded = None
            
            try:
                df_loaded = pd.read_excel(
                    archivo,
                    sheet_name='Cartera'
                )
                print(f"   📄 Hoja encontrada: 'Cartera'")
            except ValueError:
                try:
                    df_loaded = pd.read_excel(
                        archivo,
                        sheet_name='Sheet1'
                    )
                    print(f"   📄 Hoja encontrada: 'Sheet1'")
                except ValueError:
                    df_loaded = pd.read_excel(archivo)
                    print(f"   📄 Usando primera hoja disponible")
            
            if df_loaded is None:
                raise ValueError("No se pudo cargar ninguna hoja del archivo")
            
            self.df_cartera = df_loaded
            
            # ================================================================
            # NORMALIZAR FECHAS
            # ================================================================
            for col in ['Fecha', 'Fecha_Inversion', 'Fecha_actual']:
                if col in self.df_cartera.columns:
                    self.df_cartera[col] = pd.to_datetime(
                        self.df_cartera[col], 
                        errors='coerce'
                    )
            
            # ================================================================
            # ASIGNAR VARIABLES CRÍTICAS (antes de cualquier operación)
            # ================================================================
            self.nombre_cartera_actual = nombre_cartera
            self.archivo_cartera_actual = archivo
            self.archivo_bajas_actual = self.DIR_CARTERAS / f'{nombre_cartera}_bajas.xlsx'
            
            # ================================================================
            # CARGAR BAJAS (si existen)
            # ================================================================
            try:
                if self.archivo_bajas_actual.exists():
                    self.df_bajas = pd.read_excel(
                        self.archivo_bajas_actual,
                        sheet_name='Bajas'
                    )
                    
                    # Normalizar fechas en bajas
                    for col in ['Fecha_Inversion', 'Fecha_Baja']:
                        if col in self.df_bajas.columns:
                            self.df_bajas[col] = pd.to_datetime(
                                self.df_bajas[col],
                                errors='coerce'
                            )
                else:
                    self.df_bajas = pd.DataFrame()
            except Exception as e:
                print(f"   ⚠️  Error cargando bajas: {e}")
                self.df_bajas = pd.DataFrame()
            
            # ================================================================
            # INFORMACIÓN DE CARGA
            # ================================================================
            print(f"\n✅ Cartera '{nombre_cartera}' cargada")
            print(f"📊 {len(self.df_cartera)} activos")
            
            # ================================================================
            # LIMPIAR COLUMNAS DUPLICADAS
            # ================================================================
            self.limpiar_columnas_duplicadas()
            
            # ================================================================
            # AVISO: NO actualiza automáticamente
            # ================================================================
            print("\n💡 Usa la opción 13 del menú para actualizar valores actuales")
            
            return True
            
        except Exception as e:
            print(f"\n❌ Error al cargar cartera: {e}")
            import traceback
            traceback.print_exc()
            return False
    
    def _limpiar_cartera(self):
        """Limpia la cartera de datos inválidos."""
        if self.df_cartera.empty:
            return
        
        # Eliminar filas con datos críticos nulos
        filas_antes = len(self.df_cartera)
        self.df_cartera.dropna(subset=['ISIN', 'Categoria'], inplace=True)
        
        # Convertir columnas numéricas
        for col in ['Cantidad_Invertida', 'Num_Acciones']:
            if col in self.df_cartera.columns:
                self.df_cartera[col] = pd.to_numeric(
                    self.df_cartera[col], 
                    errors='coerce'
                ).fillna(0)
        
        # Eliminar activos con inversión cero
        self.df_cartera = self.df_cartera[
            self.df_cartera['Cantidad_Invertida'] > 0.01
        ].copy()
        
        filas_despues = len(self.df_cartera)
        
        if filas_antes != filas_despues:
            print(f"🔄 Limpiados: {filas_antes - filas_despues} registros inválidos")
    
    def guardar_cartera(self):
        """
        Guarda la cartera actual.
        
        VERSIÓN CON MEJOR DIAGNÓSTICO
        """
        # Verificar que existe archivo_cartera_actual
        if not hasattr(self, 'archivo_cartera_actual'):
            print("❌ No se ha definido archivo_cartera_actual")
            return False
        
        if self.archivo_cartera_actual is None:
            print("❌ No hay cartera cargada (archivo_cartera_actual es None)")
            return False
        
        # Debug: Mostrar qué se va a guardar
        print(f"   💾 Guardando en: {self.archivo_cartera_actual.name}")
        
        return self.gestor_archivos.guardar_cartera_excel(
            self.df_cartera,
            self.archivo_cartera_actual
        )
    
    def guardar_bajas(self):
        """Guarda el historial de bajas."""
        if self.archivo_bajas_actual is None or self.df_bajas.empty:
            return False
        
        try:
            self.df_bajas.to_excel(self.archivo_bajas_actual, index=False)
            print(f"✅ Bajas guardadas")
            return True
        except Exception as e:
            print(f"❌ Error al guardar bajas: {e}")
            return False
    
    def _agregar_activo_interactivo(self):
        """
        Proceso interactivo para agregar activo a la cartera.
        
        Características:
        - Busca automáticamente en asset_data
        - Usa nombre, ticker y categoría si los encuentra
        - Solo pide cantidad y fecha manualmente
        - Muestra proyección de rentabilidad
        """
        print("\n" + "="*100)
        print("➕ AGREGAR ACTIVO A LA CARTERA")
        print("="*100)
        
        # ========================================================================
        # PASO 1: Listar archivos CSV disponibles
        # ========================================================================
        print("\n📂 ARCHIVOS CSV DISPONIBLES:")
        print("-"*100)
        
        # CORREGIDO: Verificar que existe lista_df
        if not hasattr(self.gestor, 'lista_df'):
            print("\n❌ Error: No se ha cargado la lista de archivos CSV")
            print("   Verifica que se ejecutó FondosLoader correctamente")
            input("\nPresiona ENTER para continuar...")
            return
        
        archivos_disponibles = sorted(self.gestor.lista_df)
        
        if not archivos_disponibles:
            print("\n❌ No hay archivos CSV disponibles en el directorio de datos")
            print(f"   Directorio: {self.gestor.DIR_DATOS}")
            input("\nPresiona ENTER para continuar...")
            return
        
        # Mostrar archivos con índice
        for idx, archivo in enumerate(archivos_disponibles):
            print(f"  [{idx:3d}] {archivo}")
        
        print("-"*100)
        print(f"📊 Total de archivos: {len(archivos_disponibles)}")
        
        # ========================================================================
        # PASO 2: Seleccionar archivo
        # ========================================================================
        indice_csv = self.validador.solicitar_entero(
            "\nÍndice del archivo CSV (-1 para cancelar): ",
            -1,
            len(archivos_disponibles) - 1
        )
        
        if indice_csv == -1:
            print("\n🚫 Operación cancelada")
            input("\nPresiona ENTER para continuar...")
            return
        
        archivo_seleccionado = archivos_disponibles[indice_csv]
        
        # ========================================================================
        # PASO 3: Cargar datos históricos del archivo
        # ========================================================================
        print(f"\n📥 Cargando datos de: {archivo_seleccionado}...")
        df_hist = self.gestor.gestor_archivos.cargar_datos_historicos(
            archivo_seleccionado,
            verbose=False
        )
        
        if df_hist is None or df_hist.empty:
            print(f"\n❌ Error al cargar el archivo {archivo_seleccionado}")
            print("   Verifica que el archivo existe y tiene el formato correcto")
            input("\nPresiona ENTER para continuar...")
            return
        
        # Asegurar que el índice es DatetimeIndex
        try:
            if not isinstance(df_hist.index, pd.DatetimeIndex):
                # Buscar columna de fecha
                if 'Date' in df_hist.columns:
                    df_hist = df_hist.set_index('Date')
                elif 'date' in df_hist.columns:
                    df_hist = df_hist.set_index('date')
                
                # Convertir índice a datetime
                df_hist.index = pd.to_datetime(df_hist.index)
            
            # Verificar que ahora sí es DatetimeIndex
            if not isinstance(df_hist.index, pd.DatetimeIndex):
                raise ValueError("No se pudo convertir el índice a fechas")
                
        except Exception as e:
            print(f"\n❌ Error al procesar fechas del archivo: {e}")
            print("   El archivo debe tener fechas válidas en el índice o columna 'Date'")
            input("\nPresiona ENTER para continuar...")
            return
        
        # ========================================================================
        # PASO 4: BUSCAR ACTIVO EN ASSET_DATA (AUTOMÁTICO)
        # ========================================================================
        print("\n🔍 Buscando activo en la base de datos...")
        
        # Extraer posible ticker del nombre de archivo
        ticker_archivo = archivo_seleccionado.replace('.csv', '').replace('vliq_', '')
        
        nombre_encontrado = None
        ticker_encontrado = None
        categoria_encontrada = None
        
        # CORREGIDO: Verificar que existe asset_data
        if hasattr(self.gestor, 'asset_data') and self.gestor.asset_data:
            # Buscar en asset_data
            for categoria, activos in self.gestor.asset_data.items():
                for nombre_activo, datos in activos.items():
                    # Buscar por ticker o ISIN
                    ticker_dato = datos.get('ticker', '')
                    isin_dato = datos.get('ISIN', '')
                    
                    if (ticker_dato == ticker_archivo or 
                        isin_dato == ticker_archivo or
                        ticker_archivo in ticker_dato or
                        ticker_archivo in isin_dato):
                        
                        nombre_encontrado = nombre_activo
                        ticker_encontrado = isin_dato if isin_dato else ticker_dato
                        categoria_encontrada = categoria
                        break
                if nombre_encontrado:
                    break
        
        # ========================================================================
        # PASO 5: Mostrar información del activo
        # ========================================================================
        print("\n" + "="*100)
        print("📊 INFORMACIÓN DEL ACTIVO")
        print("="*100)
        
        if nombre_encontrado:
            # USAR AUTOMÁTICAMENTE los datos encontrados
            print(f"\n✅ Activo identificado automáticamente en asset_data:")
            print(f"   📛 Nombre:     {nombre_encontrado}")
            print(f"   🏷️  Ticker:    {ticker_encontrado}")
            print(f"   📂 Categoría:  {categoria_encontrada}")
            
            nombre = nombre_encontrado
            ticker = ticker_encontrado
            categoria = categoria_encontrada
            
        else:
            # NO encontrado - Pedir datos manualmente
            print(f"\n⚠️  Activo NO encontrado en asset_data")
            print(f"   Deberás introducir los datos manualmente")
            
            # Pedir nombre
            nombre = self.validador.solicitar_texto(
                "\n   📛 Nombre del activo: ", 
                no_vacio=True
            )
            
            # Pedir ticker
            print(f"   💡 Sugerencia basada en archivo: {ticker_archivo}")
            ticker_input = self.validador.solicitar_texto(
                "   🏷️  Ticker/ISIN (ENTER para usar sugerencia): ", 
                no_vacio=False
            )
            ticker = ticker_input if ticker_input else ticker_archivo
            
            # CORREGIDO: Seleccionar categoría verificando columna
            categorias_existentes = []
            if not self.gestor.df_cartera.empty and 'Categoria' in self.gestor.df_cartera.columns:
                categorias_existentes = list(self.gestor.df_cartera['Categoria'].unique())
            
            if categorias_existentes:
                print("\n   📂 Categorías existentes en la cartera:")
                for idx, cat in enumerate(categorias_existentes):
                    num_activos = len(self.gestor.df_cartera[
                        self.gestor.df_cartera['Categoria'] == cat
                    ])
                    print(f"      [{idx}] {cat} ({num_activos} activos)")
                print("      [N] Nueva categoría")
                
                opcion_cat = input("\n   Selecciona categoría: ").strip().upper()
                
                if opcion_cat == 'N':
                    categoria = self.validador.solicitar_texto(
                        "   Nueva categoría: ", 
                        no_vacio=True
                    )
                elif opcion_cat.isdigit() and 0 <= int(opcion_cat) < len(categorias_existentes):
                    categoria = categorias_existentes[int(opcion_cat)]
                else:
                    # Si escribe texto, usarlo como nueva categoría
                    categoria = opcion_cat if opcion_cat else self.validador.solicitar_texto(
                        "   Categoría: ", 
                        no_vacio=True
                    )
            else:
                print("   📝 No hay categorías existentes. Crea la primera.")
                categoria = self.validador.solicitar_texto(
                    "   📂 Categoría: ", 
                    no_vacio=True
                )
        
        # Mostrar información del archivo
        print(f"\n📄 Archivo:        {archivo_seleccionado}")
        
        try:
            fecha_inicio = df_hist.index.min()
            fecha_fin = df_hist.index.max()
            print(f"📅 Primer dato:    {fecha_inicio.date()}")
            print(f"📅 Último dato:    {fecha_fin.date()}")
            print(f"📈 Días de datos:  {len(df_hist)}")
            
            if 'Close' in df_hist.columns:
                precio_inicial = df_hist['Close'].iloc[0]
                precio_actual = df_hist['Close'].iloc[-1]
                cambio_pct = ((precio_actual - precio_inicial) / precio_inicial * 100)
                print(f"💰 Precio inicial: {precio_inicial:.4f}")
                print(f"💰 Precio actual:  {precio_actual:.4f}")
                print(f"📊 Cambio total:   {cambio_pct:+.2f}%")
        except Exception as e:
            print(f"⚠️  Error mostrando detalles: {e}")
        
        print("="*100)
        
        # ========================================================================
        # PASO 6: CANTIDAD INVERTIDA (siempre se pide)
        # ========================================================================
        print("\n💰 CANTIDAD A INVERTIR:")
        cantidad = self.validador.solicitar_decimal("   Cantidad (€): ", min_val=0.01)
        
        # ========================================================================
        # PASO 7: FECHA DE INVERSIÓN (siempre se pide)
        # ========================================================================
        print("\n📅 FECHA DE INVERSIÓN:")
        try:
            print(f"   Rango disponible: {df_hist.index.min().date()} a {df_hist.index.max().date()}")
        except:
            print(f"   Rango disponible: {df_hist.index.min()} a {df_hist.index.max()}")
        
        fecha_valida = False
        fecha = None
        
        while not fecha_valida:
            fecha = self.normalizador.solicitar_fecha_usuario()
            
            # Convertir fecha a timestamp para comparar
            fecha_ts = pd.Timestamp(fecha)
            
            # Validar que esté dentro del rango
            if fecha_ts < df_hist.index.min() or fecha_ts > df_hist.index.max():
                try:
                    print(f"   ⚠️  La fecha debe estar entre {df_hist.index.min().date()} "
                          f"y {df_hist.index.max().date()}")
                except:
                    print(f"   ⚠️  La fecha debe estar entre {df_hist.index.min()} "
                          f"y {df_hist.index.max()}")
                print("   Intenta de nuevo.")
            else:
                fecha_valida = True
        
        # ========================================================================
        # PASO 8: CONFIRMACIÓN CON PROYECCIÓN
        # ========================================================================
        print("\n" + "="*100)
        print("✅ CONFIRMAR DATOS DEL NUEVO ACTIVO")
        print("="*100)
        print(f"\n📄 Archivo CSV:       {archivo_seleccionado}")
        print(f"📛 Nombre:            {nombre}")
        print(f"🏷️  Ticker/ISIN:      {ticker}")
        print(f"📂 Categoría:         {categoria}")
        print(f"💰 Cantidad:          {cantidad:,.2f} €")
        
        try:
            print(f"📅 Fecha inversión:   {fecha.date()}")
        except:
            print(f"📅 Fecha inversión:   {fecha}")
        
        # Calcular proyección
        try:
            fecha_ts = pd.Timestamp(fecha)
            df_fecha = df_hist[df_hist.index >= fecha_ts]
            if not df_fecha.empty and 'Close' in df_fecha.columns:
                precio_en_fecha = df_fecha['Close'].iloc[0]
                participaciones = cantidad / precio_en_fecha
                
                print(f"\n💵 Precio en {fecha_ts.date()}:  {precio_en_fecha:.4f}")
                print(f"📊 Participaciones:      {participaciones:.4f}")
                
                # Valor actual y rentabilidad
                precio_actual = df_hist['Close'].iloc[-1]
                valor_actual = participaciones * precio_actual
                ganancia = valor_actual - cantidad
                rentabilidad = (ganancia / cantidad * 100) if cantidad > 0 else 0
                
                print(f"\n📈 PROYECCIÓN SI HUBIERAS INVERTIDO EN {fecha_ts.date()}:")
                print(f"   Valor actual:      {valor_actual:,.2f} €")
                print(f"   Ganancia:          {ganancia:+,.2f} €")
                print(f"   Rentabilidad:      {rentabilidad:+.2f}%")
                
                # Calcular CAGR si hay más de 1 año
                from datetime import datetime
                dias = (df_hist.index.max() - fecha_ts).days
                años = dias / 365.25
                
                if años >= 1:
                    cagr = ((valor_actual / cantidad) ** (1 / años) - 1) * 100
                    print(f"   CAGR:              {cagr:+.2f}%")
        except Exception as e:
            print(f"   ⚠️  No se pudo calcular proyección: {e}")
        
        print("="*100)
        
        # ========================================================================
        # PASO 9: AGREGAR ACTIVO A LA CARTERA
        # ========================================================================
        if self.validador.solicitar_confirmacion("\n¿Agregar este activo a la cartera? (s/n): "):
            # CORREGIDO: Try-except al agregar
            try:
                resultado = self.gestor.agregar_activo(
                    ticker=ticker,
                    nombre=nombre,
                    categoria=categoria,
                    cantidad_invertida=cantidad,
                    fecha_inversion=fecha,
                    archivo_csv=archivo_seleccionado
                )
                
                if resultado:
                    print("\n" + "="*100)
                    print("✅ ACTIVO AGREGADO CORRECTAMENTE")
                    print("="*100)
                    print(f"\n🎉 {nombre} ha sido agregado a tu cartera")
                    print(f"📊 Total de activos en cartera: {len(self.gestor.df_cartera)}")
                    
                    # Mostrar resumen de la categoría
                    if 'Categoria' in self.gestor.df_cartera.columns:
                        df_cat = self.gestor.df_cartera[
                            self.gestor.df_cartera['Categoria'] == categoria
                        ]
                        total_categoria = df_cat['Cantidad_Invertida'].sum()
                        num_activos_cat = len(df_cat)
                        print(f"💰 Total invertido en {categoria}: {total_categoria:,.2f} € ({num_activos_cat} activos)")
                    
                    # Mostrar total de la cartera
                    total_cartera = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                    print(f"💼 Total en cartera: {total_cartera:,.2f} €")
                    
                    # Porcentaje de la inversión
                    porcentaje = (cantidad / total_cartera * 100) if total_cartera > 0 else 0
                    print(f"📊 Esta inversión representa el {porcentaje:.2f}% de tu cartera")
                else:
                    print("\n" + "="*100)
                    print("❌ ERROR AL AGREGAR ACTIVO")
                    print("="*100)
                    print("\n⚠️  No se pudo agregar el activo a la cartera")
                    print("   Posibles causas:")
                    print("   - El activo ya existe en la cartera")
                    print("   - Error en los datos introducidos")
                    print("   - Problema al guardar el archivo")
            
            except Exception as e:
                print("\n" + "="*100)
                print("❌ ERROR AL AGREGAR ACTIVO")
                print("="*100)
                print(f"\n⚠️  Error: {e}")
                import traceback
                traceback.print_exc()
        else:
            print("\n🚫 Operación cancelada - El activo NO fue agregado")
        
        input("\nPresiona ENTER para continuar...")
        
    def _recalcular_pesos(self):
        """Recalcula los pesos de todos los activos."""
        if self.df_cartera.empty:
            return
        
        capital_total = self.df_cartera['Cantidad_Invertida'].sum()
        
        if capital_total > 0:
            self.df_cartera['Peso_Objetivo'] = (
                self.df_cartera['Cantidad_Invertida'] / capital_total
            ) * 100
            self.df_cartera['Peso_Actual'] = (
                self.df_cartera['Cantidad_Invertida'] / capital_total
            ) * 100
    
    def eliminar_activo(self, indice, motivo="No especificado"):
        """
        Elimina un activo de la cartera y lo registra en bajas.
        
        Args:
            indice: Índice del activo a eliminar
            motivo: Motivo de la baja
            
        Returns:
            bool: True si se eliminó correctamente
        """
        if indice not in self.df_cartera.index:
            print("❌ Índice inválido")
            return False
        
        # Obtener datos del activo
        activo = self.df_cartera.loc[indice].copy()
        
        # Registrar en bajas
        baja = pd.DataFrame({
            'Valor': [activo['Valor']],
            'ISIN': [activo['ISIN']],
            'Fichero_CSV': [activo['Fichero_CSV']],
            'Fecha_Inversion': [activo['Fecha_Inversion']],
            'Cantidad_Invertida': [activo['Cantidad_Invertida']],
            'Fecha_Baja': [pd.Timestamp.now()],
            'Motivo': [motivo],
            'Categoria': [activo['Categoria']]
        })
        
        self.df_bajas = pd.concat([self.df_bajas, baja], ignore_index=True)
        
        # Eliminar de cartera
        self.df_cartera = self.df_cartera.drop(indice).reset_index(drop=True)
        
        # Recalcular pesos
        self._recalcular_pesos()
        
        print(f"✅ Activo eliminado: {activo['Valor']}")
        return True
    
    def calcular_rentabilidades(self, verbose=True):
        """
        Calcula rentabilidades de la cartera.
        
        Returns:
            DataFrame con rentabilidades
        """
        return self.calculador.calcular_rentabilidad_cartera(
            self.df_cartera, 
            verbose=verbose
        )
    
    def comparar_con_benchmark(self, ticker_benchmark=None):
        """
        Compara la cartera con un benchmark.
        
        Args:
            ticker_benchmark: Ticker del benchmark (si None, usa el configurado)
            
        Returns:
            dict con resultados de comparación
        """
        if ticker_benchmark is None:
            ticker_benchmark = self.benchmark_ticker
        
        # Calcular rentabilidades de la cartera
        df_rent = self.calcular_rentabilidades(verbose=False)
        
        if df_rent is None:
            return None
        
        # Obtener fecha de inicio
        fecha_inicio = self.df_cartera['Fecha_Inversion'].min()
        
        # Calcular rentabilidad del benchmark
        bench_result = self.calculador.calcular_rentabilidad_benchmark(
            ticker_benchmark,
            fecha_inicio
        )
        
        if bench_result is None:
            return None
        
        # Calcular métricas de comparación
        rent_cartera = ((df_rent['Valor_Actual'].sum() - df_rent['Invertido'].sum()) / 
                       df_rent['Invertido'].sum() * 100)
        
        alpha = rent_cartera - bench_result['rentabilidad']
        
        return {
            'rentabilidad_cartera': rent_cartera,
            'rentabilidad_benchmark': bench_result['rentabilidad'],
            'alpha': alpha,
            'ticker_benchmark': ticker_benchmark,
            'df_rentabilidades': df_rent
        }
    
    def analizar_riesgo_completo(self):
        """
        Realiza análisis de riesgo completo.
        
        Returns:
            dict con análisis de riesgo
        """
        df_rent = self.calcular_rentabilidades(verbose=False)
        
        if df_rent is None:
            return None
        
        return self.analizador_riesgo.analisis_completo(
            self.df_cartera,
            df_rent
        )
    
    def generar_dashboard(self):
        """Genera dashboard completo de gráficos."""
        df_rent = self.calcular_rentabilidades(verbose=False)
        
        if df_rent is None:
            return None
        
        return self.generador_graficos.dashboard_completo(
            self.df_cartera,
            df_rent,
            mostrar=True,
            guardar=True
        )
    
    def exportar_completo(self):
        """Exporta análisis completo a Excel."""
        df_rent = self.calcular_rentabilidades(verbose=False)
        
        return self.generador_reportes.exportar_excel_completo(
            self.df_cartera,
            df_rent,
            self.df_bajas,
            f'cartera_completa_{self.nombre_cartera_actual}.xlsx'
        )
    

    def actualizar_valores_cartera(self, guardar=True):
        """
        Actualiza valores usando nombres de columna ESTÁNDAR.
        
        CORREGIDO: Maneja tipos de datos correctamente para evitar FutureWarning
        """
        if self.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            return False
        
        print("\n🔄 Actualizando valores de la cartera...")
        
        # ========================================================================
        # ASEGURAR TIPOS DE DATOS CORRECTOS EN LAS COLUMNAS
        # ========================================================================
        
        # Crear columnas si no existen CON EL TIPO CORRECTO
        if 'Valor_Actual' not in self.df_cartera.columns:
            self.df_cartera['Valor_Actual'] = 0.0
        else:
            # Convertir a float si existe
            self.df_cartera['Valor_Actual'] = self.df_cartera['Valor_Actual'].astype(float)
        
        if 'Beneficio' not in self.df_cartera.columns:
            self.df_cartera['Beneficio'] = 0.0
        else:
            self.df_cartera['Beneficio'] = self.df_cartera['Beneficio'].astype(float)
        
        if 'Plusv%' not in self.df_cartera.columns:
            self.df_cartera['Plusv%'] = 0.0
        else:
            self.df_cartera['Plusv%'] = self.df_cartera['Plusv%'].astype(float)
        
        if 'Fecha_Actual' not in self.df_cartera.columns:
            self.df_cartera['Fecha_Actual'] = pd.NaT
        else:
            # Convertir a datetime si existe
            self.df_cartera['Fecha_Actual'] = pd.to_datetime(
                self.df_cartera['Fecha_Actual'], 
                errors='coerce'
            )
        
        # ========================================================================
        # ACTUALIZAR VALORES
        # ========================================================================
        
        actualizados = 0
        errores = 0
        total = len(self.df_cartera)
        
        for idx in range(total):
            try:
                row = self.df_cartera.iloc[idx]
                rentabilidad = self.calculador.calcular_rentabilidad_activo(row)
                
                if rentabilidad:
                    # Usar .at[] con conversión explícita de tipos
                    self.df_cartera.at[idx, 'Valor_Actual'] = float(rentabilidad['valor_actual'])
                    self.df_cartera.at[idx, 'Beneficio'] = float(rentabilidad['ganancia'])
                    self.df_cartera.at[idx, 'Plusv%'] = float(rentabilidad['rentabilidad_porcentual'])
                    self.df_cartera.at[idx, 'Fecha_Actual'] = pd.Timestamp(rentabilidad['fecha_ultimo_precio'])
                    
                    actualizados += 1
                    if (idx + 1) % 5 == 0:
                        print(f"     Procesados: {idx + 1}/{total}")
                else:
                    errores += 1
            except Exception as e:
                errores += 1
                print(f"     ❌ Error en activo {idx}: {str(e)[:50]}")
        
        print(f"   ✅ Actualizados: {actualizados} activos")
        if errores > 0:
            print(f"   ⚠️  Con errores: {errores} activos")
        
        # ========================================================================
        # ACTUALIZAR PESOS
        # ========================================================================
        
        if actualizados > 0:
            print("\n📊 Actualizando pesos...")
            try:
                self.actualizar_pesos_cartera(guardar=False)
            except Exception as e:
                print(f"   ⚠️  Error actualizando pesos: {e}")
        
        # ========================================================================
        # GUARDAR
        # ========================================================================
        
        if guardar and actualizados > 0:
            if hasattr(self, 'archivo_cartera_actual') and self.archivo_cartera_actual:
                self.guardar_cartera()
                print(f"   💾 Cartera guardada correctamente")
        
        return actualizados > 0


    # ============================================================================
    # ALTERNATIVA: Si el problema persiste, usa este enfoque más robusto
    # ============================================================================
    
    def actualizar_valores_cartera_ROBUSTO(self, guardar=True):
        """
        Versión más robusta que recrea las columnas con tipos correctos.
        """
        if self.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            return False
        
        print("\n🔄 Actualizando valores de la cartera...")
        
        # ========================================================================
        # ELIMINAR COLUMNAS ANTIGUAS Y RECREAR CON TIPOS CORRECTOS
        # ========================================================================
        
        columnas_actualizar = ['Valor_Actual', 'Beneficio', 'Plusv%', 'Fecha_Actual']
        
        # Eliminar columnas si existen
        for col in columnas_actualizar:
            if col in self.df_cartera.columns:
                self.df_cartera.drop(columns=[col], inplace=True)
        
        # Crear nuevas columnas con tipos correctos
        self.df_cartera['Valor_Actual'] = pd.Series(dtype='float64')
        self.df_cartera['Beneficio'] = pd.Series(dtype='float64')
        self.df_cartera['Plusv%'] = pd.Series(dtype='float64')
        self.df_cartera['Fecha_Actual'] = pd.Series(dtype='datetime64[ns]')
        
        # ========================================================================
        # ACTUALIZAR VALORES
        # ========================================================================
        
        actualizados = 0
        errores = 0
        total = len(self.df_cartera)
        
        for idx in range(total):
            try:
                row = self.df_cartera.iloc[idx]
                rentabilidad = self.calculador.calcular_rentabilidad_activo(row)
                
                if rentabilidad:
                    self.df_cartera.at[idx, 'Valor_Actual'] = float(rentabilidad['valor_actual'])
                    self.df_cartera.at[idx, 'Beneficio'] = float(rentabilidad['ganancia'])
                    self.df_cartera.at[idx, 'Plusv%'] = float(rentabilidad['rentabilidad_porcentual'])
                    self.df_cartera.at[idx, 'Fecha_Actual'] = pd.Timestamp(rentabilidad['fecha_ultimo_precio'])
                    
                    actualizados += 1
                    if (idx + 1) % 5 == 0:
                        print(f"     Procesados: {idx + 1}/{total}")
                else:
                    errores += 1
            except Exception as e:
                errores += 1
                print(f"     ❌ Error en activo {idx}: {str(e)[:50]}")
        
        print(f"   ✅ Actualizados: {actualizados} activos")
        if errores > 0:
            print(f"   ⚠️  Con errores: {errores} activos")
        
        # Actualizar pesos
        if actualizados > 0:
            print("\n📊 Actualizando pesos...")
            try:
                self.actualizar_pesos_cartera(guardar=False)
            except Exception as e:
                print(f"   ⚠️  Error actualizando pesos: {e}")
        
        # Guardar
        if guardar and actualizados > 0:
            if hasattr(self, 'archivo_cartera_actual') and self.archivo_cartera_actual:
                self.guardar_cartera()
                print(f"   💾 Cartera guardada correctamente")
        
        return actualizados > 0


    # ============================================================================
    # MÉTODO 1: Calcular pesos actuales (GestorCartera - Celda 4)
    # ============================================================================
    
    def actualizar_pesos_cartera(self, guardar=True):
        """
        Calcula y actualiza los pesos de cada activo en la cartera.
        
        - Peso_Actual: Porcentaje actual basado en Cantidad_Invertida
        - Peso_Objetivo: Se mantiene si existe, o se calcula proporcionalmente
        
        Args:
            guardar: Si True, guarda la cartera después de actualizar
            
        Returns:
            bool: True si se actualizó correctamente
        """
        if self.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            return False
        
        print("\n📊 Actualizando pesos de la cartera...")
        
        # Asegurar que existen las columnas
        if 'Peso_Objetivo' not in self.df_cartera.columns:
            self.df_cartera['Peso_Objetivo'] = 0.0
        if 'Peso_Actual' not in self.df_cartera.columns:
            self.df_cartera['Peso_Actual'] = 0.0
        
        # Calcular total invertido
        total_invertido = self.df_cartera['Cantidad_Invertida'].sum()
        
        if total_invertido <= 0:
            print("   ⚠️  Total invertido es 0, no se pueden calcular pesos")
            return False
        
        # Calcular Peso_Actual para cada activo
        for idx in self.df_cartera.index:
            cantidad_inv = self.df_cartera.loc[idx, 'Cantidad_Invertida']
            peso_actual = (cantidad_inv / total_invertido) * 100
            self.df_cartera.loc[idx, 'Peso_Actual'] = peso_actual
            
            # Si Peso_Objetivo es 0, establecerlo igual al Peso_Actual
            peso_objetivo = self.df_cartera.loc[idx, 'Peso_Objetivo']
            if pd.isna(peso_objetivo) or peso_objetivo == 0.0:
                self.df_cartera.loc[idx, 'Peso_Objetivo'] = peso_actual
        
        print(f"   ✅ Pesos calculados para {len(self.df_cartera)} activos")
        
        # Guardar
        if guardar:
            resultado = self.guardar_cartera()
            if resultado:
                print("   💾 Cartera guardada")
        
        return True
    
    
    # ============================================================================
    # MÉTODO 2: Optimizar cartera (GestorCartera - Celda 4)
    # ============================================================================
    
    def optimizar_cartera_simple(self, estrategia='equitativo'):
        """
        Optimiza la distribución de pesos de la cartera.
        
        Estrategias disponibles:
        - 'equitativo': Todos los activos con el mismo peso
        - 'por_categoria': Distribuir equitativamente dentro de cada categoría
        - 'riesgo_bajo': Mayor peso a activos de menor volatilidad
        
        Args:
            estrategia: Tipo de optimización
            
        Returns:
            bool: True si se optimizó correctamente
        """
        if self.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            return False
        
        print(f"\n🎯 Optimizando cartera: estrategia '{estrategia}'")
        
        num_activos = len(self.df_cartera)
        
        if estrategia == 'equitativo':
            # Todos los activos con el mismo peso
            peso_por_activo = 100.0 / num_activos
            self.df_cartera['Peso_Objetivo'] = peso_por_activo
            print(f"   ✅ Peso equitativo: {peso_por_activo:.2f}% por activo")
        
        elif estrategia == 'por_categoria':
            # Distribuir equitativamente dentro de cada categoría
            categorias = self.df_cartera['Categoria'].unique()
            peso_por_categoria = 100.0 / len(categorias)
            
            for categoria in categorias:
                activos_cat = self.df_cartera[self.df_cartera['Categoria'] == categoria]
                num_activos_cat = len(activos_cat)
                peso_por_activo_cat = peso_por_categoria / num_activos_cat
                
                self.df_cartera.loc[
                    self.df_cartera['Categoria'] == categoria, 
                    'Peso_Objetivo'
                ] = peso_por_activo_cat
                
                print(f"   📂 {categoria}: {peso_por_categoria:.2f}% total "
                      f"({peso_por_activo_cat:.2f}% × {num_activos_cat} activos)")
        
        elif estrategia == 'mantener_actual':
            # Mantener la distribución actual
            self.df_cartera['Peso_Objetivo'] = self.df_cartera['Peso_Actual']
            print(f"   ✅ Pesos objetivo = Pesos actuales")
        
        else:
            print(f"   ⚠️  Estrategia '{estrategia}' no reconocida")
            return False
        
        # Verificar que suma 100%
        total_peso = self.df_cartera['Peso_Objetivo'].sum()
        print(f"   📊 Total Peso_Objetivo: {total_peso:.2f}%")
        
        # Guardar
        self.guardar_cartera()
        print("   💾 Cartera guardada")
        
        return True


        
    # ============================================================================
    # MÉTODO PARA LIMPIAR COLUMNAS DUPLICADAS (si ya existen)
    # ============================================================================
    
    def limpiar_columnas_duplicadas(self):
        """
        Limpia TODAS las columnas duplicadas de forma robusta.
        
        VERSIÓN MEJORADA que:
        - Detecta cualquier columna con .1, .2, .3, etc.
        - Mantiene solo la columna base (sin sufijo)
        - Si no existe la base, renombra la última
        """
        if self.df_cartera.empty:
            return 0
        
        print("\n🧹 Limpiando columnas duplicadas (versión robusta)...")
        
        columnas_eliminadas = 0
        columnas_renombradas = 0
        
        # Obtener todas las columnas
        todas_columnas = list(self.df_cartera.columns)
        print(f"   📊 Columnas totales: {len(todas_columnas)}")
        
        # Detectar columnas con sufijos (.1, .2, .3, etc.)
        import re
        columnas_con_sufijo = {}
        
        for col in todas_columnas:
            # Buscar patrón: nombre.1, nombre.2, etc.
            match = re.match(r'^(.+?)\.(\d+)$', col)
            if match:
                base = match.group(1)
                numero = int(match.group(2))
                
                if base not in columnas_con_sufijo:
                    columnas_con_sufijo[base] = []
                columnas_con_sufijo[base].append((col, numero))
        
        if not columnas_con_sufijo:
            print(f"   ✅ No se encontraron columnas con sufijos numéricos")
            return 0
        
        print(f"   ⚠️  Encontradas {len(columnas_con_sufijo)} columnas base con duplicados:")
        
        # Procesar cada grupo de columnas duplicadas
        for base, duplicados in columnas_con_sufijo.items():
            # Ordenar por número de sufijo
            duplicados.sort(key=lambda x: x[1])
            nombres_dup = [d[0] for d in duplicados]
            
            print(f"\n   📋 Base: '{base}'")
            print(f"      Duplicados: {nombres_dup}")
            
            # Verificar si existe la columna base (sin sufijo)
            if base in self.df_cartera.columns:
                # Existe la base, eliminar todos los duplicados
                print(f"      ✅ Base '{base}' existe, eliminando duplicados...")
                for col_dup, _ in duplicados:
                    self.df_cartera.drop(columns=[col_dup], inplace=True)
                    columnas_eliminadas += 1
                    print(f"         🗑️  Eliminado: {col_dup}")
            else:
                # No existe la base, renombrar el último duplicado
                ultimo = duplicados[-1][0]
                print(f"      ⚠️  Base '{base}' NO existe")
                print(f"      ➡️  Renombrando '{ultimo}' → '{base}'")
                self.df_cartera.rename(columns={ultimo: base}, inplace=True)
                columnas_renombradas += 1
                
                # Eliminar los demás duplicados
                for col_dup, _ in duplicados[:-1]:
                    self.df_cartera.drop(columns=[col_dup], inplace=True)
                    columnas_eliminadas += 1
                    print(f"         🗑️  Eliminado: {col_dup}")
        
        print(f"\n   ✅ Columnas eliminadas: {columnas_eliminadas}")
        print(f"   ✅ Columnas renombradas: {columnas_renombradas}")
        print(f"   📊 Columnas finales: {len(self.df_cartera.columns)}")
        
        # Guardar después de limpiar
        if columnas_eliminadas + columnas_renombradas > 0:
            self.guardar_cartera()
            print(f"   💾 Cartera guardada con columnas limpias")
        
        return columnas_eliminadas + columnas_renombradas


    def agregar_activo(self, ticker, nombre, categoria, cantidad_invertida, 
                       fecha_inversion, archivo_csv):
        """
        Agrega un activo a la cartera.
        
        Args:
            ticker: ISIN o ticker del activo
            nombre: Nombre del activo
            categoria: Categoría del activo
            cantidad_invertida: Cantidad en euros
            fecha_inversion: Fecha de inversión (pd.Timestamp o string)
            archivo_csv: Nombre del archivo de datos históricos
            
        Returns:
            bool: True si se agregó correctamente
        """
        try:
            # Convertir fecha si es string
            if isinstance(fecha_inversion, str):
                fecha_inversion = pd.to_datetime(fecha_inversion)
            
            # Verificar que no exista ya el activo
            if not self.df_cartera.empty and 'ISIN' in self.df_cartera.columns:
                if ticker in self.df_cartera['ISIN'].values:
                    print(f"\n⚠️  El activo {ticker} ya existe en la cartera")
                    return False
            
            # Crear nuevo registro
            nuevo_activo = {
                'Valor': nombre,
                'Categoria': categoria,
                'ISIN': ticker,
                'Fichero_CSV': archivo_csv,
                'Fecha_Inversion': fecha_inversion,
                'Cantidad_Invertida': cantidad_invertida,
                'Precio_Compra': 0.0,  # Se calculará después
                'Operación': 'Suscripción',
                'Num_Acciones': 0.0,  # Se calculará después
                'Fecha_Actual': pd.NaT,
                'Precio_Actual': 0.0,
                'Valor_Actual': 0.0,
                'Beneficio': 0.0,
                'Plusv%': 0.0,
                'Peso_Objetivo': 0.0,
                'Peso_Actual': 0.0,
                'Observaciones': ''
            }
            
            # Añadir columnas adicionales si existen en df_cartera
            if not self.df_cartera.empty:
                for col in self.df_cartera.columns:
                    if col not in nuevo_activo:
                        nuevo_activo[col] = None
            
            # Crear DataFrame con el nuevo activo
            df_nuevo = pd.DataFrame([nuevo_activo])
            
            # Añadir a la cartera
            if self.df_cartera.empty:
                self.df_cartera = df_nuevo
            else:
                self.df_cartera = pd.concat([self.df_cartera, df_nuevo], 
                                            ignore_index=True)
            
            # Calcular precio de compra y participaciones
            try:
                df_hist = self.gestor_archivos.cargar_datos_historicos(
                    archivo_csv,
                    verbose=False
                )
                
                if df_hist is not None and not df_hist.empty:
                    # Asegurar DatetimeIndex
                    if not isinstance(df_hist.index, pd.DatetimeIndex):
                        if 'Date' in df_hist.columns:
                            df_hist = df_hist.set_index('Date')
                        df_hist.index = pd.to_datetime(df_hist.index)
                    
                    # Buscar precio en fecha de inversión
                    fecha_ts = pd.Timestamp(fecha_inversion)
                    df_desde_fecha = df_hist[df_hist.index >= fecha_ts]
                    
                    if not df_desde_fecha.empty and 'Close' in df_desde_fecha.columns:
                        precio_compra = df_desde_fecha['Close'].iloc[0]
                        participaciones = cantidad_invertida / precio_compra
                        
                        # Actualizar en el último registro (el que acabamos de añadir)
                        ultimo_idx = len(self.df_cartera) - 1
                        self.df_cartera.at[ultimo_idx, 'Precio_Compra'] = precio_compra
                        self.df_cartera.at[ultimo_idx, 'Num_Acciones'] = participaciones
            except Exception as e:
                print(f"   ⚠️  No se pudo calcular precio de compra: {e}")
            
            # Guardar
            resultado = self.guardar_cartera()
            
            return resultado
            
        except Exception as e:
            print(f"\n❌ Error al agregar activo: {e}")
            import traceback
            traceback.print_exc()
            return False

    

# ============================================================================
# CLASE 9: INTERFAZ DE USUARIO
# ============================================================================

class InterfazUsuario:
    """
    Gestiona toda la interacción con el usuario: menús, inputs, visualización.
    """
    
    def __init__(self, gestor_cartera):
        """
        Args:
            gestor_cartera: Instancia de GestorCartera
        """
        self.gestor = gestor_cartera
        self.validador = ValidadorEntradas()
        self.normalizador = NormalizadorFechas()
    
    def mostrar_ayuda_menu(self):
        """Muestra ayuda detallada del menú."""
        ayuda = """
╔══════════════════════════════════════════════════════════════════════════════╗
║                     📖 AYUDA - DESCRIPCIÓN DE OPCIONES                       ║
╚══════════════════════════════════════════════════════════════════════════════╝

📁 GESTIÓN DE CARTERA
─────────────────────────────────────────────────────────────────────────────
 1. Ver cartera actual
    → Muestra todos los activos con cantidades, precios y valores

 2. Agregar activo
    → Añade un nuevo activo a la cartera

 3. Modificar cantidad
    → Cambia la cantidad invertida en un activo existente

 4. Eliminar activo
    → Retira un activo de la cartera (se guarda en historial)

 5. Ver historial de bajas
    → Lista de activos eliminados con sus datos históricos

📊 ANÁLISIS Y REPORTES
─────────────────────────────────────────────────────────────────────────────
 6. Calcular rentabilidades
    → Calcula rendimiento de toda la cartera

 7. Generar gráficos
    → Crea visualizaciones de distribución y rentabilidades

 8. Comparar con benchmark
    → Compara tu cartera vs índice de referencia

 9. Análisis de riesgo
    → Evalúa volatilidad, diversificación y métricas de riesgo

10. Resumen ejecutivo
    → Genera reporte con métricas clave

11. Exportar análisis completo
    → Exporta todo a Excel con múltiples hojas

💾 SISTEMA
─────────────────────────────────────────────────────────────────────────────
12. Cambiar de cartera
    → Alterna entre diferentes carteras guardadas

13. Guardar y salir
    → Guarda cambios y cierra el programa

 0. Salir sin guardar
    → Sale sin guardar cambios recientes

═══════════════════════════════════════════════════════════════════════════════

💡 Presiona ENTER para volver al menú...
═══════════════════════════════════════════════════════════════════════════════
"""
        print(ayuda)
        input()
    
    def menu_principal(self):
        """
        Menú principal de la aplicación.
        
        Returns:
            str: Estado de salida ('guardado', 'sin_guardar', 'interrumpido')
        """
        print("\n💡 CONSEJO: Escribe H para ver la ayuda completa")
        input("Presiona ENTER para comenzar...")
        
        while True:
            nombre = self.gestor.nombre_cartera_actual or "Ninguna"
            
            print("\n" + "="*70)
            print(f"🎯 GESTOR DE CARTERA - {nombre}")
            print("="*70)
            
            print("\n📁 GESTIÓN DE CARTERA")
            print("  1. 📈 Ver cartera actual")
            print("  2. ➕ Agregar activo")
            print("  3. ✏️  Modificar cantidad")
            print("  4. ➖ Eliminar activo")
            print("  5. 📋 Ver historial de bajas")
            
            print("\n📊 ANÁLISIS Y REPORTES")
            print("  6. 📊 Calcular rentabilidades")
            print("  7. 📈 Dashboard de gráficos")
            print("  8. 🎯 Comparar con benchmark")
            print("  9. ⚠️  Análisis de riesgo")
            print(" 10. 📋 Resumen ejecutivo")
            print(" 11. 💾 Exportar análisis completo")
            print(" 12. 🔬 Análisis técnico de activo")
            print(" 13. 🔄 Actualizar valores cartera")  # ← Corregir nombre
            print(" 14. 🎯 Optimizar distribución (Markowitz)")
            print(" 15. 📊 Análisis completo cartera")   # ← Mover aquí el de línea 14
            
            print("\n💾 SISTEMA")
            print(" 16. 🔄 Cambiar de cartera")
            print(" 17. 💾 Guardar y salir")
            print("  0. 🚪 Salir sin guardar")
            
            # ELIMINAR la opción 98 (temporal)
            # ELIMINAR la opción 15 de "Limpiar columnas duplicadas" (ya no es necesaria)
            
            print("\n💡 AYUDA")
            print("  H. ❓ Ver ayuda detallada")
            
            print("\n" + "="*70)
            
            try:
                opcion = input("\nSelecciona opción: ").strip().upper()
                
                if opcion == 'H':
                    self.mostrar_ayuda_menu()
                    continue
                
                if opcion == '1':
                    self._ver_cartera()
                elif opcion == '2':
                    self._agregar_activo_interactivo()
                elif opcion == '3':
                    self._modificar_cantidad_interactivo()
                elif opcion == '4':
                    self._eliminar_activo_interactivo()
                elif opcion == '5':
                    self._ver_historial_bajas()
                elif opcion == '6':
                    self.gestor.calcular_rentabilidades(verbose=True)
                    input("\nPresiona ENTER para continuar...")
                elif opcion == '7':
                    self.gestor.generar_dashboard()
                    input("\nPresiona ENTER para continuar...")
                elif opcion == '8':
                    self._comparar_benchmark_interactivo()
                elif opcion == '9':
                    self._analizar_riesgo_interactivo()
                elif opcion == '10':
                    self._resumen_ejecutivo()
                elif opcion == '11':
                    self.gestor.exportar_completo()
                    input("\nPresiona ENTER para continuar...")
                elif opcion == '12':
                    self._analisis_tecnico_activo_interactivo()
                elif opcion == '13':
                    self._actualizar_valores_interactivo()
                elif opcion == '14':
                    self._optimizar_cartera_interactivo()  # ← NUEVA LÍNEA
                elif opcion == '15':
                    self._ejecutar_analisis_completo_cartera()  # ← Mover aquí
                elif opcion == '16':
                    self._cambiar_cartera()
                elif opcion == '17':
                    return self._guardar_y_salir()
                elif opcion == '0':
                    if self.validador.solicitar_confirmacion("\n⚠️  ¿Salir sin guardar? (s/n): "):
                        return 'sin_guardar'

                else:
                    print("\n❌ Opción no válida. Presiona H para ayuda.")
                    input("Presiona ENTER...")
                    
            except KeyboardInterrupt:
                print("\n\n⚠️  Interrupción detectada")
                if self.validador.solicitar_confirmacion("¿Salir sin guardar? (s/n): "):
                    return 'interrumpido'
            except Exception as e:
                print(f"\n❌ Error: {e}")
                import traceback
                traceback.print_exc()
                input("Presiona ENTER...")
    
    def _ver_cartera(self):
        """Muestra la cartera actual."""
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            input("Presiona ENTER...")
            return
        
        print("\n" + "="*100)
        print("MI CARTERA DE INVERSIÓN")
        print("="*100)
        
        for categoria in self.gestor.df_cartera['Categoria'].unique():
            df_cat = self.gestor.df_cartera[
                self.gestor.df_cartera['Categoria'] == categoria
            ]
            total_cat = df_cat['Cantidad_Invertida'].sum()
            
            print(f"\n📊 {categoria.upper()} - Total: {total_cat:,.2f} €")
            print("-" * 100)
            
            for _, row in df_cat.iterrows():
                print(f"   • {row['Valor'][:45]:<45} | "
                      f"{row['Cantidad_Invertida']:>10,.2f} € | "
                      f"{row['Fecha_Inversion']}")
        
        total = self.gestor.df_cartera['Cantidad_Invertida'].sum()
        print("\n" + "="*100)
        print(f"💰 TOTAL: {total:,.2f} € | Posiciones: {len(self.gestor.df_cartera)}")
        print("="*100)
        
        input("\nPresiona ENTER...")
    
    def _agregar_activo_interactivo(self):
        """
        Proceso interactivo para agregar activo a la cartera.
        
        Características:
        - Busca automáticamente en asset_data
        - Muestra nombre del activo junto al archivo CSV
        - Usa nombre, ticker y categoría si los encuentra
        - Solo pide cantidad y fecha manualmente
        - Muestra proyección de rentabilidad
        """
        print("\n" + "="*100)
        print("➕ AGREGAR ACTIVO A LA CARTERA")
        print("="*100)
        
        # ========================================================================
        # PASO 1: Listar archivos CSV disponibles CON NOMBRES
        # ========================================================================
        print("\n📂 ARCHIVOS CSV DISPONIBLES:")
        print("-"*100)
        
        # Verificar que existe lista_df
        if not hasattr(self.gestor, 'lista_df'):
            print("\n❌ Error: No se ha cargado la lista de archivos CSV")
            print("   Verifica que se ejecutó FondosLoader correctamente")
            input("\nPresiona ENTER para continuar...")
            return
        
        archivos_disponibles = sorted(self.gestor.lista_df)
        
        if not archivos_disponibles:
            print("\n❌ No hay archivos CSV disponibles en el directorio de datos")
            print(f"   Directorio: {self.gestor.DIR_DATOS}")
            input("\nPresiona ENTER para continuar...")
            return
        
        # Crear mapeo de archivos a nombres de activos
        archivo_a_nombre = {}
        
        if hasattr(self.gestor, 'asset_data') and self.gestor.asset_data:
            for categoria, activos in self.gestor.asset_data.items():
                for nombre_activo, datos in activos.items():
                    ticker = datos.get('ticker', '')
                    isin = datos.get('ISIN', '')
                    
                    # Buscar archivo que coincida con ticker o ISIN
                    for archivo in archivos_disponibles:
                        archivo_limpio = archivo.replace('.csv', '').replace('vliq_', '')
                        
                        if (ticker and archivo_limpio == ticker) or \
                           (isin and archivo_limpio == isin) or \
                           (ticker and archivo_limpio in ticker) or \
                           (isin and archivo_limpio in isin):
                            archivo_a_nombre[archivo] = nombre_activo
                            break
        
        # Mostrar archivos CON nombres
        for idx, archivo in enumerate(archivos_disponibles):
            if archivo in archivo_a_nombre:
                nombre = archivo_a_nombre[archivo]
                # Truncar nombre si es muy largo
                if len(nombre) > 50:
                    nombre = nombre[:47] + "..."
                print(f"  [{idx:3d}] {archivo:<30} → {nombre}")
            else:
                print(f"  [{idx:3d}] {archivo:<30} → (No identificado)")
        
        print("-"*100)
        print(f"📊 Total de archivos: {len(archivos_disponibles)}")
        print(f"✅ Identificados: {len(archivo_a_nombre)}/{len(archivos_disponibles)}")
        
        # ========================================================================
        # PASO 2: Seleccionar archivo
        # ========================================================================
        indice_csv = self.validador.solicitar_entero(
            "\nÍndice del archivo CSV (-1 para cancelar): ",
            -1,
            len(archivos_disponibles) - 1
        )
        
        if indice_csv == -1:
            print("\n🚫 Operación cancelada")
            input("\nPresiona ENTER para continuar...")
            return
        
        archivo_seleccionado = archivos_disponibles[indice_csv]
        
        # ========================================================================
        # PASO 3: Cargar datos históricos del archivo
        # ========================================================================
        print(f"\n📥 Cargando datos de: {archivo_seleccionado}...")
        df_hist = self.gestor.gestor_archivos.cargar_datos_historicos(
            archivo_seleccionado,
            verbose=False
        )
        
        if df_hist is None or df_hist.empty:
            print(f"\n❌ Error al cargar el archivo {archivo_seleccionado}")
            print("   Verifica que el archivo existe y tiene el formato correcto")
            input("\nPresiona ENTER para continuar...")
            return
        
        # Asegurar que el índice es DatetimeIndex
        try:
            if not isinstance(df_hist.index, pd.DatetimeIndex):
                # Buscar columna de fecha
                if 'Date' in df_hist.columns:
                    df_hist = df_hist.set_index('Date')
                elif 'date' in df_hist.columns:
                    df_hist = df_hist.set_index('date')
                
                # Convertir índice a datetime
                df_hist.index = pd.to_datetime(df_hist.index)
            
            # Verificar que ahora sí es DatetimeIndex
            if not isinstance(df_hist.index, pd.DatetimeIndex):
                raise ValueError("No se pudo convertir el índice a fechas")
                
        except Exception as e:
            print(f"\n❌ Error al procesar fechas del archivo: {e}")
            print("   El archivo debe tener fechas válidas en el índice o columna 'Date'")
            input("\nPresiona ENTER para continuar...")
            return
        
        # ========================================================================
        # PASO 4: BUSCAR ACTIVO EN ASSET_DATA (AUTOMÁTICO)
        # ========================================================================
        print("\n🔍 Buscando activo en la base de datos...")
        
        # Extraer posible ticker del nombre de archivo
        ticker_archivo = archivo_seleccionado.replace('.csv', '').replace('vliq_', '')
        
        nombre_encontrado = None
        ticker_encontrado = None
        categoria_encontrada = None
        
        # Buscar en asset_data
        if hasattr(self.gestor, 'asset_data') and self.gestor.asset_data:
            for categoria, activos in self.gestor.asset_data.items():
                for nombre_activo, datos in activos.items():
                    # Buscar por ticker o ISIN
                    ticker_dato = datos.get('ticker', '')
                    isin_dato = datos.get('ISIN', '')
                    
                    if (ticker_dato == ticker_archivo or 
                        isin_dato == ticker_archivo or
                        ticker_archivo in ticker_dato or
                        ticker_archivo in isin_dato):
                        
                        nombre_encontrado = nombre_activo
                        ticker_encontrado = isin_dato if isin_dato else ticker_dato
                        categoria_encontrada = categoria
                        break
                if nombre_encontrado:
                    break
        
        # ========================================================================
        # PASO 5: Mostrar información del activo
        # ========================================================================
        print("\n" + "="*100)
        print("📊 INFORMACIÓN DEL ACTIVO")
        print("="*100)
        
        if nombre_encontrado:
            # USAR AUTOMÁTICAMENTE los datos encontrados
            print(f"\n✅ Activo identificado automáticamente en asset_data:")
            print(f"   📛 Nombre:     {nombre_encontrado}")
            print(f"   🏷️  Ticker:    {ticker_encontrado}")
            print(f"   📂 Categoría:  {categoria_encontrada}")
            
            nombre = nombre_encontrado
            ticker = ticker_encontrado
            categoria = categoria_encontrada
            
        else:
            # NO encontrado - Pedir datos manualmente
            print(f"\n⚠️  Activo NO encontrado en asset_data")
            print(f"   Deberás introducir los datos manualmente")
            
            # Pedir nombre
            nombre = self.validador.solicitar_texto(
                "\n   📛 Nombre del activo: ", 
                no_vacio=True
            )
            
            # Pedir ticker
            print(f"   💡 Sugerencia basada en archivo: {ticker_archivo}")
            ticker_input = self.validador.solicitar_texto(
                "   🏷️  Ticker/ISIN (ENTER para usar sugerencia): ", 
                no_vacio=False
            )
            ticker = ticker_input if ticker_input else ticker_archivo
            
            # Seleccionar categoría
            categorias_existentes = []
            if not self.gestor.df_cartera.empty and 'Categoria' in self.gestor.df_cartera.columns:
                categorias_existentes = list(self.gestor.df_cartera['Categoria'].unique())
            
            if categorias_existentes:
                print("\n   📂 Categorías existentes en la cartera:")
                for idx, cat in enumerate(categorias_existentes):
                    num_activos = len(self.gestor.df_cartera[
                        self.gestor.df_cartera['Categoria'] == cat
                    ])
                    print(f"      [{idx}] {cat} ({num_activos} activos)")
                print("      [N] Nueva categoría")
                
                opcion_cat = input("\n   Selecciona categoría: ").strip().upper()
                
                if opcion_cat == 'N':
                    categoria = self.validador.solicitar_texto(
                        "   Nueva categoría: ", 
                        no_vacio=True
                    )
                elif opcion_cat.isdigit() and 0 <= int(opcion_cat) < len(categorias_existentes):
                    categoria = categorias_existentes[int(opcion_cat)]
                else:
                    # Si escribe texto, usarlo como nueva categoría
                    categoria = opcion_cat if opcion_cat else self.validador.solicitar_texto(
                        "   Categoría: ", 
                        no_vacio=True
                    )
            else:
                print("   📝 No hay categorías existentes. Crea la primera.")
                categoria = self.validador.solicitar_texto(
                    "   📂 Categoría: ", 
                    no_vacio=True
                )
        
        # Mostrar información del archivo
        print(f"\n📄 Archivo:        {archivo_seleccionado}")
        
        try:
            fecha_inicio = df_hist.index.min()
            fecha_fin = df_hist.index.max()
            print(f"📅 Primer dato:    {fecha_inicio.date()}")
            print(f"📅 Último dato:    {fecha_fin.date()}")
            print(f"📈 Días de datos:  {len(df_hist)}")
            
            if 'Close' in df_hist.columns:
                precio_inicial = df_hist['Close'].iloc[0]
                precio_actual = df_hist['Close'].iloc[-1]
                cambio_pct = ((precio_actual - precio_inicial) / precio_inicial * 100)
                print(f"💰 Precio inicial: {precio_inicial:.4f}")
                print(f"💰 Precio actual:  {precio_actual:.4f}")
                print(f"📊 Cambio total:   {cambio_pct:+.2f}%")
        except Exception as e:
            print(f"⚠️  Error mostrando detalles: {e}")
        
        print("="*100)
        
        # ========================================================================
        # PASO 6: CANTIDAD INVERTIDA (siempre se pide)
        # ========================================================================
        print("\n💰 CANTIDAD A INVERTIR:")
        cantidad = self.validador.solicitar_decimal("   Cantidad (€): ", min_val=0.01)
        
        # ========================================================================
        # PASO 7: FECHA DE INVERSIÓN (siempre se pide)
        # ========================================================================
        print("\n📅 FECHA DE INVERSIÓN:")
        try:
            print(f"   Rango disponible: {df_hist.index.min().date()} a {df_hist.index.max().date()}")
        except:
            print(f"   Rango disponible: {df_hist.index.min()} a {df_hist.index.max()}")
        
        fecha_valida = False
        fecha = None
        
        while not fecha_valida:
            fecha = self.normalizador.solicitar_fecha_usuario()
            
            # Convertir fecha a timestamp para comparar
            fecha_ts = pd.Timestamp(fecha)
            
            # Validar que esté dentro del rango
            if fecha_ts < df_hist.index.min() or fecha_ts > df_hist.index.max():
                try:
                    print(f"   ⚠️  La fecha debe estar entre {df_hist.index.min().date()} "
                          f"y {df_hist.index.max().date()}")
                except:
                    print(f"   ⚠️  La fecha debe estar entre {df_hist.index.min()} "
                          f"y {df_hist.index.max()}")
                print("   Intenta de nuevo.")
            else:
                fecha_valida = True
        
        # ========================================================================
        # PASO 8: CONFIRMACIÓN CON PROYECCIÓN
        # ========================================================================
        print("\n" + "="*100)
        print("✅ CONFIRMAR DATOS DEL NUEVO ACTIVO")
        print("="*100)
        print(f"\n📄 Archivo CSV:       {archivo_seleccionado}")
        print(f"📛 Nombre:            {nombre}")
        print(f"🏷️  Ticker/ISIN:      {ticker}")
        print(f"📂 Categoría:         {categoria}")
        print(f"💰 Cantidad:          {cantidad:,.2f} €")
        
        try:
            print(f"📅 Fecha inversión:   {fecha.date()}")
        except:
            print(f"📅 Fecha inversión:   {fecha}")
        
        # Calcular proyección
        try:
            fecha_ts = pd.Timestamp(fecha)
            df_fecha = df_hist[df_hist.index >= fecha_ts]
            if not df_fecha.empty and 'Close' in df_fecha.columns:
                precio_en_fecha = df_fecha['Close'].iloc[0]
                participaciones = cantidad / precio_en_fecha
                
                print(f"\n💵 Precio en {fecha_ts.date()}:  {precio_en_fecha:.4f}")
                print(f"📊 Participaciones:      {participaciones:.4f}")
                
                # Valor actual y rentabilidad
                precio_actual = df_hist['Close'].iloc[-1]
                valor_actual = participaciones * precio_actual
                ganancia = valor_actual - cantidad
                rentabilidad = (ganancia / cantidad * 100) if cantidad > 0 else 0
                
                print(f"\n📈 PROYECCIÓN SI HUBIERAS INVERTIDO EN {fecha_ts.date()}:")
                print(f"   Valor actual:      {valor_actual:,.2f} €")
                print(f"   Ganancia:          {ganancia:+,.2f} €")
                print(f"   Rentabilidad:      {rentabilidad:+.2f}%")
                
                # Calcular CAGR si hay más de 1 año
                from datetime import datetime
                dias = (df_hist.index.max() - fecha_ts).days
                años = dias / 365.25
                
                if años >= 1:
                    cagr = ((valor_actual / cantidad) ** (1 / años) - 1) * 100
                    print(f"   CAGR:              {cagr:+.2f}%")
        except Exception as e:
            print(f"   ⚠️  No se pudo calcular proyección: {e}")
        
        print("="*100)
        
        # ========================================================================
        # PASO 9: AGREGAR ACTIVO A LA CARTERA
        # ========================================================================
        if self.validador.solicitar_confirmacion("\n¿Agregar este activo a la cartera? (s/n): "):
            try:
                resultado = self.gestor.agregar_activo(
                    ticker=ticker,
                    nombre=nombre,
                    categoria=categoria,
                    cantidad_invertida=cantidad,
                    fecha_inversion=fecha,
                    archivo_csv=archivo_seleccionado
                )
                
                if resultado:
                    print("\n" + "="*100)
                    print("✅ ACTIVO AGREGADO CORRECTAMENTE")
                    print("="*100)
                    print(f"\n🎉 {nombre} ha sido agregado a tu cartera")
                    print(f"📊 Total de activos en cartera: {len(self.gestor.df_cartera)}")
                    
                    # Mostrar resumen de la categoría
                    if 'Categoria' in self.gestor.df_cartera.columns:
                        df_cat = self.gestor.df_cartera[
                            self.gestor.df_cartera['Categoria'] == categoria
                        ]
                        total_categoria = df_cat['Cantidad_Invertida'].sum()
                        num_activos_cat = len(df_cat)
                        print(f"💰 Total invertido en {categoria}: {total_categoria:,.2f} € ({num_activos_cat} activos)")
                    
                    # Mostrar total de la cartera
                    total_cartera = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                    print(f"💼 Total en cartera: {total_cartera:,.2f} €")
                    
                    # Porcentaje de la inversión
                    porcentaje = (cantidad / total_cartera * 100) if total_cartera > 0 else 0
                    print(f"📊 Esta inversión representa el {porcentaje:.2f}% de tu cartera")
                else:
                    print("\n" + "="*100)
                    print("❌ ERROR AL AGREGAR ACTIVO")
                    print("="*100)
                    print("\n⚠️  No se pudo agregar el activo a la cartera")
                    print("   Posibles causas:")
                    print("   - El activo ya existe en la cartera")
                    print("   - Error en los datos introducidos")
                    print("   - Problema al guardar el archivo")
            
            except Exception as e:
                print("\n" + "="*100)
                print("❌ ERROR AL AGREGAR ACTIVO")
                print("="*100)
                print(f"\n⚠️  Error: {e}")
                import traceback
                traceback.print_exc()
        else:
            print("\n🚫 Operación cancelada - El activo NO fue agregado")
        
        input("\nPresiona ENTER para continuar...")
        
    def _modificar_cantidad_interactivo(self):
        """Modifica cantidad de un activo."""
        # Implementación similar a agregar
        print("\n⚠️  Función en desarrollo")
        input("Presiona ENTER...")
    
    def _eliminar_activo_interactivo(self):
        """Elimina un activo de forma interactiva."""
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            input("Presiona ENTER...")
            return
        
        print("\n➖ ELIMINAR ACTIVO")
        print("="*70)
        
        for idx, row in self.gestor.df_cartera.iterrows():
            print(f"[{idx}] {row['Valor'][:40]} | {row['Cantidad_Invertida']:,.2f} €")
        
        indice = self.validador.solicitar_entero(
            "\nÍndice del activo a eliminar (-1 para cancelar): ",
            -1, len(self.gestor.df_cartera)-1
        )
        
        if indice == -1:
            return
        
        if self.validador.solicitar_confirmacion("\n¿Confirmar eliminación? (s/n): "):
            motivo = self.validador.solicitar_texto(
                "Motivo (opcional): ",
                no_vacio=False
            ) or "No especificado"
            
            self.gestor.eliminar_activo(indice, motivo)
        
        input("\nPresiona ENTER...")
    
    def _ver_historial_bajas(self):
        """Muestra historial de bajas."""
        if self.gestor.df_bajas.empty:
            print("\n✅ No hay bajas registradas")
            input("Presiona ENTER...")
            return
        
        print("\n📋 HISTORIAL DE BAJAS")
        print("="*100)
        
        for _, row in self.gestor.df_bajas.iterrows():
            print(f"\n• {row['Valor']}")
            print(f"  Inversión: {row['Cantidad_Invertida']:,.2f} € | "
                  f"Fecha inversión: {row['Fecha_Inversion']}")
            print(f"  Baja: {row['Fecha_Baja']} | Motivo: {row['Motivo']}")
        
        print("="*100)
        input("\nPresiona ENTER...")
    

    def _comparar_benchmark_interactivo(self):
        """
        Compara la cartera con un benchmark seleccionado interactivamente.
        
        MEJORA: Permite elegir el benchmark de la lista de índices
        """
        print("\n" + "="*80)
        print("🎯 COMPARAR CARTERA CON BENCHMARK")
        print("="*80)
        
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            input("\nPresiona ENTER para continuar...")
            return
        
        # ========================================================================
        # PASO 1: Mostrar índices disponibles
        # ========================================================================
        print("\n📊 ÍNDICES DISPONIBLES:")
        print("-"*80)
        
        indices_disponibles = []
        
        if hasattr(self.gestor, 'asset_data') and 'Indices' in self.gestor.asset_data:
            indices_disponibles = list(self.gestor.asset_data['Indices'].items())
            
            for idx, (nombre, datos) in enumerate(indices_disponibles):
                ticker = datos.get('ticker', datos.get('ISIN', ''))
                print(f"  [{idx}] {nombre:<30} ({ticker})")
            
            # Opción para usar el predeterminado
            ticker_default = getattr(self.gestor, 'benchmark_ticker', '^STOXX50E')
            print(f"  [ ] Usar predeterminado: {ticker_default} (presiona ENTER)")
        else:
            print("\n⚠️  No hay índices disponibles en asset_data")
            ticker_default = getattr(self.gestor, 'benchmark_ticker', '^STOXX50E')
            print(f"💡 Usando benchmark predeterminado: {ticker_default}")
            indices_disponibles = []
        
        print("-"*80)
        
        # ========================================================================
        # PASO 2: Seleccionar benchmark
        # ========================================================================
        ticker_benchmark = None
        nombre_benchmark = None
        
        if indices_disponibles:
            seleccion = input("\nSelecciona índice (número o ENTER para predeterminado): ").strip()
            
            if seleccion and seleccion.isdigit():
                idx_bench = int(seleccion)
                if 0 <= idx_bench < len(indices_disponibles):
                    nombre_bench, datos_bench = indices_disponibles[idx_bench]
                    ticker_benchmark = datos_bench.get('ticker', datos_bench.get('ISIN'))
                    nombre_benchmark = nombre_bench
                    print(f"\n✅ Benchmark seleccionado: {nombre_benchmark} ({ticker_benchmark})")
                else:
                    print("\n⚠️  Índice no válido - Usando predeterminado")
                    ticker_benchmark = ticker_default
                    nombre_benchmark = "Benchmark predeterminado"
            else:
                ticker_benchmark = ticker_default
                nombre_benchmark = "Benchmark predeterminado"
                print(f"\n💡 Usando benchmark predeterminado: {ticker_benchmark}")
        else:
            ticker_benchmark = ticker_default
            nombre_benchmark = "Benchmark predeterminado"
        
        # ========================================================================
        # PASO 3: Calcular rentabilidades de la cartera
        # ========================================================================
        print("\n" + "="*80)
        print("📊 CALCULANDO RENTABILIDADES DE LA CARTERA")
        print("="*80)
        
        df_rent = self.gestor.calcular_rentabilidades(verbose=True)
        
        if df_rent is None:
            print("\n❌ No se pudieron calcular las rentabilidades")
            input("\nPresiona ENTER para continuar...")
            return
        
        # ========================================================================
        # PASO 4: Calcular rentabilidad del benchmark
        # ========================================================================
        print("\n" + "="*80)
        print(f"📊 CALCULANDO BENCHMARK: {nombre_benchmark}")
        print("="*80)
        
        # Obtener fecha de inicio (la más antigua de la cartera)
        fecha_inicio = self.gestor.df_cartera['Fecha_Inversion'].min()
        
        print(f"\n📊 Calculando benchmark {ticker_benchmark}...")
        print(f"📅 Período: {fecha_inicio.date()} → {pd.Timestamp.now().date()}")
        
        # Descargar datos del benchmark
        try:
            import yfinance as yf
            
            print("⬇️  Descargando datos...")
            df_benchmark = yf.download(
                ticker_benchmark,
                start=fecha_inicio,
                end=pd.Timestamp.now(),
                progress=False
            )
            
            if df_benchmark.empty:
                print(f"\n❌ No se pudieron obtener datos de {ticker_benchmark}")
                input("\nPresiona ENTER para continuar...")
                return
            
            # Normalizar columnas si es MultiIndex
            if isinstance(df_benchmark.columns, pd.MultiIndex):
                df_benchmark.columns = df_benchmark.columns.droplevel(1)
            
            # Calcular rentabilidad del benchmark
            precio_inicial = df_benchmark['Close'].iloc[0]
            precio_final = df_benchmark['Close'].iloc[-1]
            rentabilidad_benchmark = ((precio_final - precio_inicial) / precio_inicial) * 100
            
            print(f"✅ Rentabilidad {ticker_benchmark}: {rentabilidad_benchmark:+.2f}%")
            print(f"   Precio inicial: {precio_inicial:.2f}")
            print(f"   Precio final: {precio_final:.2f}")
            
        except Exception as e:
            print(f"\n❌ Error al descargar benchmark: {e}")
            input("\nPresiona ENTER para continuar...")
            return
        
        # ========================================================================
        # PASO 5: Comparar resultados
        # ========================================================================
        print("\n" + "="*80)
        print("📊 COMPARACIÓN CON BENCHMARK")
        print("="*80)
        
        # Calcular rentabilidad total de la cartera
        total_invertido = df_rent['Invertido'].sum()
        total_actual = df_rent['Valor_Actual'].sum()
        ganancia_total = total_actual - total_invertido
        rentabilidad_cartera = (ganancia_total / total_invertido * 100) if total_invertido > 0 else 0
        
        # Calcular alpha (diferencia de rentabilidad)
        alpha = rentabilidad_cartera - rentabilidad_benchmark
        
        print(f"\n💼 TU CARTERA:")
        print(f"   Invertido:      {total_invertido:>15,.2f} €")
        print(f"   Valor actual:   {total_actual:>15,.2f} €")
        print(f"   Ganancia:       {ganancia_total:>+15,.2f} €")
        print(f"   Rentabilidad:   {rentabilidad_cartera:>+15.2f} %")
        
        print(f"\n📊 {nombre_benchmark.upper()}:")
        print(f"   Rentabilidad:   {rentabilidad_benchmark:>+15.2f} %")
        
        print(f"\n🎯 COMPARACIÓN:")
        print(f"   Alpha (diferencia): {alpha:>+12.2f} %")
        
        # Diagnóstico
        if alpha > 5:
            print(f"\n✅ ¡EXCELENTE! Tu cartera supera al {nombre_benchmark} en {alpha:.2f}%")
        elif alpha > 0:
            print(f"\n👍 ¡BIEN! Tu cartera supera al {nombre_benchmark} en {alpha:.2f}%")
        elif alpha > -5:
            print(f"\n⚠️  Tu cartera está ligeramente por debajo del {nombre_benchmark} ({alpha:.2f}%)")
        else:
            print(f"\n🔴 Tu cartera está significativamente por debajo del {nombre_benchmark} ({alpha:.2f}%)")
        
        # Mostrar rendimiento por categoría
        print("\n" + "="*80)
        print("📊 RENDIMIENTO POR CATEGORÍA")
        print("="*80)
        
        for categoria in df_rent['Categoria'].unique():
            df_cat = df_rent[df_rent['Categoria'] == categoria]
            inv_cat = df_cat['Invertido'].sum()
            val_cat = df_cat['Valor_Actual'].sum()
            rent_cat = ((val_cat - inv_cat) / inv_cat * 100) if inv_cat > 0 else 0
            alpha_cat = rent_cat - rentabilidad_benchmark
            
            simbolo = "✅" if alpha_cat > 0 else "⚠️" if alpha_cat > -5 else "🔴"
            
            print(f"\n{simbolo} {categoria}:")
            print(f"   Invertido:     {inv_cat:>12,.2f} €")
            print(f"   Valor actual:  {val_cat:>12,.2f} €")
            print(f"   Rentabilidad:  {rent_cat:>+12.2f} %")
            print(f"   vs Benchmark:  {alpha_cat:>+12.2f} %")
        
        print("\n" + "="*80)
        
        input("\nPresiona ENTER para continuar...")
    
    def _analizar_riesgo_interactivo(self):
        """Análisis de riesgo interactivo."""
        analisis = self.gestor.analizar_riesgo_completo()
        
        if analisis is None:
            print("\n❌ No se pudo realizar el análisis")
            input("Presiona ENTER...")
            return
        
        print("\n⚠️  ANÁLISIS DE RIESGO")
        print("="*70)
        print(f"Concentración Top 5:     {analisis['concentracion']['peso_top5']:.2f}%")
        print(f"Volatilidad:             {analisis['volatilidad']['volatilidad']:.2f}%")
        print(f"Ratio Sharpe:            {analisis['sharpe']:.2f}")
        print(f"Posiciones negativas:    {analisis['posiciones_negativas']} "
              f"({analisis['pct_posiciones_negativas']:.1f}%)")
        print("="*70)
        
        input("\nPresiona ENTER...")
    
    def _resumen_ejecutivo(self):
        """Genera y muestra resumen ejecutivo."""
        df_rent = self.gestor.calcular_rentabilidades(verbose=False)
        
        if df_rent is None:
            print("\n❌ No se pudo generar el resumen")
            input("Presiona ENTER...")
            return
        
        self.gestor.generador_reportes.generar_resumen_ejecutivo_texto(
            self.gestor.df_cartera,
            df_rent,
            guardar=True,
            mostrar=True
        )
        
        input("\nPresiona ENTER...")
    
    def _cambiar_cartera(self):
        """Permite cambiar a otra cartera existente."""
        print("\n" + "="*70)
        print("🔄 CAMBIAR DE CARTERA")
        print("="*70)
        
        try:
            # Listar carteras disponibles
            carteras = self.gestor.gestor_archivos.listar_carteras_disponibles()
            
            if not carteras:
                print("\n⚠️  No hay carteras disponibles")
                print("   Usa la opción 2 para crear una cartera nueva")
                input("\n🔙 Presiona ENTER para volver al menú...")
                return
            
            print("\n📁 Carteras disponibles:")
            print("-"*70)
            
            for idx, cartera in enumerate(carteras, 1):
                # Si cartera es una tupla (nombre, ruta), usar el nombre
                if isinstance(cartera, tuple):
                    nombre = cartera[0]  # Primer elemento de la tupla
                else:
                    nombre = cartera  # Es un string
                
                # Quitar la extensión .xlsx si existe
                if nombre.endswith('.xlsx'):
                    nombre = nombre.replace('.xlsx', '')
                
                # Marcar la cartera actual
                if self.gestor.nombre_cartera_actual == nombre:
                    print(f"  [{idx}] {nombre} ⭐ (actual)")
                else:
                    print(f"  [{idx}] {nombre}")
            
            print(f"  [0] Cancelar")
            print("-"*70)
            
            # Seleccionar cartera
            seleccion = self.validador.solicitar_entero(
                "\nSeleccione cartera: ",
                min_val=0,
                max_val=len(carteras)
            )
            
            if seleccion == 0:
                print("\n🚫 Operación cancelada")
                input("\n🔙 Presiona ENTER para volver al menú...")
                return
            
            # Obtener nombre de cartera seleccionada
            cartera_seleccionada = carteras[seleccion - 1]
            
            # Manejar tupla o string
            if isinstance(cartera_seleccionada, tuple):
                nombre_cartera = cartera_seleccionada[0]
            else:
                nombre_cartera = cartera_seleccionada
            
            # Quitar extensión .xlsx
            if nombre_cartera.endswith('.xlsx'):
                nombre_cartera = nombre_cartera.replace('.xlsx', '')
            
            # Verificar si es la misma cartera
            if self.gestor.nombre_cartera_actual == nombre_cartera:
                print(f"\n⚠️  Ya estás usando la cartera '{nombre_cartera}'")
                input("\n🔙 Presiona ENTER para volver al menú...")
                return
            
            # Cargar nueva cartera
            print(f"\n📥 Cargando cartera '{nombre_cartera}'...")
            
            resultado = self.gestor.cargar_cartera(nombre_cartera)
            
            if resultado:
                print(f"\n✅ Cartera '{nombre_cartera}' cargada correctamente")
                print(f"   📊 Activos: {len(self.gestor.df_cartera)}")
            else:
                print(f"\n❌ Error al cargar la cartera '{nombre_cartera}'")
            
        except Exception as e:
            print(f"\n❌ Error al cambiar de cartera: {e}")
            print(f"🐞 Tipo de error: {type(e).__name__}")
            import traceback
            traceback.print_exc()
        
        input("\n🔙 Presiona ENTER para volver al menú...")
    
    def _analisis_tecnico_activo_interactivo(self):
        """
        Análisis técnico avanzado de un activo CON selección de benchmark.
        """
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            input("Presiona ENTER...")
            return
        
        print("\n🔬 ANÁLISIS TÉCNICO")
        print("="*70)
        
        # ========================================================================
        # PASO 1: Seleccionar activo de la cartera
        # ========================================================================
        # Mostrar activos
        for idx, row in self.gestor.df_cartera.iterrows():
            print(f"[{idx}] {row['Valor'][:50]}")
        
        indice = self.validador.solicitar_entero(
            "\nÍndice del activo (-1 cancelar): ",
            -1, len(self.gestor.df_cartera)-1
        )
        
        if indice == -1:
            return
        
        activo = self.gestor.df_cartera.loc[indice]
        
        # ========================================================================
        # PASO 2: SELECCIONAR BENCHMARK (NUEVO)
        # ========================================================================
        print("\n" + "="*70)
        print("📊 SELECCIÓN DE BENCHMARK")
        print("="*70)
        print("\nSelecciona un índice de referencia para comparar:")
        print("(Déjalo vacío para análisis sin benchmark)")
        print("-"*70)
        
        # Obtener índices de asset_data
        indices_disponibles = []
        if 'Indices' in self.gestor.asset_data:
            indices_disponibles = list(self.gestor.asset_data['Indices'].items())
        
        if indices_disponibles:
            print("\n📈 ÍNDICES DISPONIBLES:")
            for idx, (nombre, datos) in enumerate(indices_disponibles):
                ticker = datos.get('ticker', datos.get('ISIN', ''))
                print(f"  [{idx}] {nombre:<25} ({ticker})")
            
            print(f"  [ ] Sin benchmark (presiona ENTER)")
            print("-"*70)
            
            # Solicitar selección
            seleccion = input("\nSelecciona índice (número o ENTER para ninguno): ").strip()
            
            benchmark_ticker = None
            benchmark_nombre = None
            
            if seleccion and seleccion.isdigit():
                idx_bench = int(seleccion)
                if 0 <= idx_bench < len(indices_disponibles):
                    nombre_bench, datos_bench = indices_disponibles[idx_bench]
                    benchmark_ticker = datos_bench.get('ticker', datos_bench.get('ISIN'))
                    benchmark_nombre = nombre_bench
                    print(f"\n✅ Benchmark seleccionado: {benchmark_nombre} ({benchmark_ticker})")
                else:
                    print("\n⚠️  Índice no válido - Continuando sin benchmark")
            else:
                print("\n💡 Continuando sin benchmark")
        else:
            print("\n⚠️  No hay índices disponibles en asset_data")
            benchmark_ticker = None
            benchmark_nombre = None
        
        # ========================================================================
        # PASO 3: Cargar datos del activo
        # ========================================================================
        print("\n" + "="*70)
        print(f"📥 Cargando datos de: {activo['Valor'][:50]}...")
        
        df_hist = self.gestor.gestor_archivos.cargar_datos_historicos(
            activo['Fichero_CSV'],
            verbose=False
        )
        
        if df_hist is None:
            print("\n❌ No se pudieron cargar datos")
            input("Presiona ENTER...")
            return
        
        # ========================================================================
        # PASO 4: Cargar benchmark si se seleccionó
        # ========================================================================
        df_benchmark = None
        
        if benchmark_ticker:
            print(f"📥 Descargando benchmark: {benchmark_nombre}...")
            
            try:
                import yfinance as yf
                
                # Obtener rango de fechas del activo
                fecha_inicio = activo['Fecha_Inversion']
                fecha_fin = pd.Timestamp.now()
                
                df_benchmark = yf.download(
                    benchmark_ticker,
                    start=fecha_inicio,
                    end=fecha_fin,
                    progress=False
                )
                
                if df_benchmark.empty:
                    print(f"⚠️  No se pudo descargar {benchmark_nombre}")
                    df_benchmark = None
                else:
                    # Normalizar columnas si es MultiIndex
                    if isinstance(df_benchmark.columns, pd.MultiIndex):
                        df_benchmark.columns = df_benchmark.columns.droplevel(1)
                    
                    print(f"✅ Benchmark cargado: {len(df_benchmark)} días")
                    
            except Exception as e:
                print(f"⚠️  Error descargando benchmark: {e}")
                df_benchmark = None
        
        # ========================================================================
        # PASO 5: Crear analizador y ejecutar análisis
        # ========================================================================
        print("\n" + "="*70)
        print("🔬 Ejecutando análisis técnico...")
        print("="*70)
        
        # Crear analizador con o sin benchmark
        analizador = ValorAnalizador(
            df=df_hist.reset_index().set_index('Date'),
            df_benchmark=df_benchmark,  # Puede ser None
            nombre_fondo=activo['Valor'],
            nombre_benchmark=benchmark_nombre if benchmark_nombre else "Sin benchmark",
            fecha_inicio_inversion=activo['Fecha_Inversion'],
            graficar=True,
            tipo_grafico='plotly'
        )
        
        # Ejecutar análisis
        analizador.generar_analisis_completo(
            capital=activo['Cantidad_Invertida']
        )
        
        input("\nPresiona ENTER...")

    def _ejecutar_analisis_completo_cartera(self):
        """
        Ejecuta análisis técnico completo de toda la cartera.
        Usa la clase AnalizadorCarteraCompleto.
        """
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía. Agrega activos primero.")
            input("\nPresiona ENTER para continuar...")
            return
        
        try:
            print("\n" + "="*100)
            print("🎯 ANÁLISIS COMPLETO DE CARTERA")
            print("="*100)
            print(f"\n📊 Analizando cartera: {self.gestor.nombre_cartera_actual}")
            print(f"📈 Activos en cartera: {len(self.gestor.df_cartera)}")
            
            # Crear instancia del analizador
            analizador = AnalizadorCarteraCompleto(
                gestor_cartera=self.gestor,
                gestor_archivos=self.gestor.gestor_archivos,
                benchmark_ticker='^STOXX50E'  # Puedes hacerlo configurable
            )
            
            # Ejecutar análisis completo
            resultados = analizador.ejecutar_analisis_completo(verbose=True)
            
            # Opciones post-análisis
            print("\n" + "="*100)
            print("💾 OPCIONES")
            print("="*100)
            print("  1. 💾 Exportar resultados a Excel")
            print("  2. 📊 Ver gráficos adicionales")
            print("  0. ↩️  Volver al menú")
            
            opcion = input("\nSelecciona opción: ").strip()
            
            if opcion == '1':
                self._exportar_analisis_completo(resultados)
            elif opcion == '2':
                print("\n⏳ Función en desarrollo...")
            
        except Exception as e:
            print(f"\n❌ Error al ejecutar análisis: {e}")
            import traceback
            traceback.print_exc()
        
        input("\nPresiona ENTER para continuar...")
    
    def _exportar_analisis_completo(self, resultados):
        """
        Exporta resultados del análisis completo a Excel.
        
        Args:
            resultados: Diccionario con resultados del análisis
        """
        try:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            nombre_archivo = f"analisis_completo_{self.gestor.nombre_cartera_actual}_{timestamp}.xlsx"
            ruta = self.gestor.DIR_EXPORTS / nombre_archivo
            
            with pd.ExcelWriter(ruta, engine='openpyxl') as writer:
                # Hoja 1: Resumen de cartera
                df_resumen = pd.DataFrame([resultados['cartera']])
                df_resumen.to_excel(writer, sheet_name='Resumen_Cartera', index=False)
                
                # Hoja 2: Análisis por activo
                df_activos = pd.DataFrame(resultados['activos'])
                df_activos.to_excel(writer, sheet_name='Detalle_Activos', index=False)
                
                # Hoja 3: Diagnóstico
                if 'diagnostico' in resultados['cartera']:
                    df_diag = pd.DataFrame({
                        'Diagnóstico': resultados['cartera']['diagnostico']
                    })
                    df_diag.to_excel(writer, sheet_name='Diagnostico', index=False)
            
            print(f"\n✅ Análisis exportado: {nombre_archivo}")
            print(f"📁 Ubicación: {ruta}")
            
        except Exception as e:
            print(f"\n❌ Error al exportar: {e}")

    def _actualizar_valores_interactivo(self):
        """
        Actualiza los valores de la cartera de forma interactiva.
        
        UBICACIÓN: InterfazUsuario (Celda 7)
        AÑADIR DESPUÉS DE: _ejecutar_analisis_completo_cartera()
        """
        print("\n" + "="*80)
        print("🔄 ACTUALIZAR VALORES DE LA CARTERA")
        print("="*80)
        
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            input("\nPresiona ENTER para continuar...")
            return
        
        print(f"\n📊 Cartera actual: {self.gestor.nombre_cartera_actual}")
        print(f"📈 Activos: {len(self.gestor.df_cartera)}")
        
        if self.validador.solicitar_confirmacion("\n¿Actualizar valores? (s/n): "):
            # Llamar al método de GestorCartera
            resultado = self.gestor.actualizar_valores_cartera(guardar=True)
            
            if resultado:
                print("\n" + "="*80)
                print("✅ VALORES ACTUALIZADOS CORRECTAMENTE")
                print("="*80)
                
                # Mostrar resumen si las columnas existen
                try:
                    total_invertido = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                    
                    if 'Valor_actual' in self.gestor.df_cartera.columns:
                        total_actual = self.gestor.df_cartera['Valor_actual'].sum()
                        ganancia_total = total_actual - total_invertido
                        rentabilidad = (ganancia_total / total_invertido * 100) if total_invertido > 0 else 0
                        
                        print(f"\n💰 RESUMEN:")
                        print(f"   Invertido:     {total_invertido:>15,.2f} €")
                        print(f"   Valor actual:  {total_actual:>15,.2f} €")
                        print(f"   Ganancia:      {ganancia_total:>+15,.2f} €")
                        print(f"   Rentabilidad:  {rentabilidad:>+15.2f} %")
                    else:
                        print(f"\n💰 Total invertido: {total_invertido:,.2f} €")
                except Exception as e:
                    print(f"\n⚠️  No se pudo calcular resumen: {e}")
            else:
                print("\n❌ No se pudieron actualizar los valores")
        else:
            print("\n🚫 Actualización cancelada")
        
        input("\nPresiona ENTER para continuar...")
   
    def _optimizar_cartera_interactivo(self):
        """
        Proceso interactivo para optimizar la cartera usando Markowitz.
        """
       # Verificar que existe OptimizadorMarkowitz
        try:
            optimizador = OptimizadorMarkowitz(self.gestor, tasa_libre_riesgo=0.02)
        except NameError:
            print("\n❌ OptimizadorMarkowitz no está disponible")
            print("   Ejecuta la celda con la clase OptimizadorMarkowitz")
            input("\nPresiona ENTER...")
            return
       
        if self.gestor.df_cartera.empty:
            print("\n⚠️  La cartera está vacía")
            input("\nPresiona ENTER para continuar...")
            return
        
        print("\n" + "="*80)
        print("🎯 OPTIMIZACIÓN DE CARTERA - TEORÍA MODERNA DE PORTFOLIO")
        print("="*80)
        
        # Crear optimizador
        from optimizador_markowitz import OptimizadorMarkowitz
        optimizador = OptimizadorMarkowitz(self.gestor, tasa_libre_riesgo=0.02)
        
        # Preguntar días históricos
        print("\n📅 ¿Cuántos días de histórico usar?")
        print("   1. 252 días (1 año)")
        print("   2. 504 días (2 años)")
        print("   3. 756 días (3 años)")
        print("   4. Personalizado")
        
        opcion_dias = input("\nOpción (1-4): ").strip()
        
        if opcion_dias == '1':
            dias = 252
        elif opcion_dias == '2':
            dias = 504
        elif opcion_dias == '3':
            dias = 756
        elif opcion_dias == '4':
            try:
                dias = int(input("   Número de días: "))
                if dias < 30:
                    print("   ⚠️  Mínimo 30 días, usando 252")
                    dias = 252
            except:
                print("   ⚠️  Valor inválido, usando 252 días")
                dias = 252
        else:
            print("   ⚠️  Opción inválida, usando 252 días")
            dias = 252
        
        # Preparar datos
        if not optimizador.preparar_datos(dias_historico=dias, verbose=True):
            input("\n❌ No se pudieron preparar los datos. Presiona ENTER...")
            return
        
        # Menú de optimización
        while True:
            print("\n" + "="*80)
            print("OPCIONES DE OPTIMIZACIÓN")
            print("="*80)
            print("\n  1. 🎯 Maximizar Ratio Sharpe (mejor relación riesgo/rentabilidad)")
            print("  2. 📉 Minimizar Volatilidad (menor riesgo)")
            print("  3. 📈 Optimizar para retorno objetivo")
            print("  4. 🌐 Ver frontera eficiente")
            print("  5. 📊 Comparar actual vs optimizada")
            print("  0. ⬅️  Volver")
            
            opcion = input("\nSelecciona opción: ").strip()
            
            if opcion == '0':
                break
            
            elif opcion == '1':
                # Maximizar Sharpe
                resultado = optimizador.optimizar_sharpe(verbose=True)
                
                if resultado:
                    print("\n💡 ¿Qué deseas hacer?")
                    print("   1. Ver gráficos")
                    print("   2. Comparar con cartera actual")
                    print("   3. Exportar resultados")
                    print("   0. Continuar")
                    
                    sub_opcion = input("\nOpción: ").strip()
                    
                    if sub_opcion == '1':
                        # Calcular pesos actuales para el gráfico
                        total = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                        pesos_actuales = []
                        for activo in resultado['activos']:
                            row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
                            if not row.empty:
                                pesos_actuales.append(row['Cantidad_Invertida'].iloc[0] / total)
                            else:
                                pesos_actuales.append(0.0)
                        
                        optimizador.graficar_frontera_eficiente(
                            portfolio_sharpe=resultado,
                            portfolio_actual=np.array(pesos_actuales)
                        )
                        optimizador.graficar_composicion_comparada(resultado)
                    
                    elif sub_opcion == '2':
                        optimizador.comparar_con_actual(resultado, verbose=True)
                    
                    elif sub_opcion == '3':
                        self._exportar_optimizacion(resultado, optimizador)
                
                input("\nPresiona ENTER para continuar...")
            
            elif opcion == '2':
                # Minimizar volatilidad
                resultado = optimizador.optimizar_minima_volatilidad(verbose=True)
                
                if resultado:
                    print("\n💡 ¿Qué deseas hacer?")
                    print("   1. Ver gráficos")
                    print("   2. Comparar con cartera actual")
                    print("   3. Exportar resultados")
                    print("   0. Continuar")
                    
                    sub_opcion = input("\nOpción: ").strip()
                    
                    if sub_opcion == '1':
                        total = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                        pesos_actuales = []
                        for activo in resultado['activos']:
                            row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
                            if not row.empty:
                                pesos_actuales.append(row['Cantidad_Invertida'].iloc[0] / total)
                            else:
                                pesos_actuales.append(0.0)
                        
                        optimizador.graficar_composicion_comparada(resultado)
                    
                    elif sub_opcion == '2':
                        optimizador.comparar_con_actual(resultado, verbose=True)
                    
                    elif sub_opcion == '3':
                        self._exportar_optimizacion(resultado, optimizador)
                
                input("\nPresiona ENTER para continuar...")
            
            elif opcion == '3':
                # Retorno objetivo
                print("\n📈 Retorno objetivo anualizado")
                print(f"   Mínimo posible: {optimizador.rentabilidades_esperadas.min()*100:.2f}%")
                print(f"   Máximo posible: {optimizador.rentabilidades_esperadas.max()*100:.2f}%")
                
                try:
                    retorno = float(input("\n   Retorno deseado (%): ")) / 100
                    
                    resultado = optimizador.optimizar_retorno_objetivo(retorno, verbose=True)
                    
                    if resultado:
                        print("\n💡 ¿Qué deseas hacer?")
                        print("   1. Ver gráficos")
                        print("   2. Comparar con cartera actual")
                        print("   3. Exportar resultados")
                        print("   0. Continuar")
                        
                        sub_opcion = input("\nOpción: ").strip()
                        
                        if sub_opcion == '1':
                            total = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                            pesos_actuales = []
                            for activo in resultado['activos']:
                                row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
                                if not row.empty:
                                    pesos_actuales.append(row['Cantidad_Invertida'].iloc[0] / total)
                                else:
                                    pesos_actuales.append(0.0)
                            
                            optimizador.graficar_composicion_comparada(resultado)
                        
                        elif sub_opcion == '2':
                            optimizador.comparar_con_actual(resultado, verbose=True)
                        
                        elif sub_opcion == '3':
                            self._exportar_optimizacion(resultado, optimizador)
                
                except ValueError:
                    print("\n❌ Valor inválido")
                
                input("\nPresiona ENTER para continuar...")
            
            elif opcion == '4':
                # Frontera eficiente
                print("\n📊 Calculando frontera eficiente...")
                
                # Optimizar Sharpe para el gráfico
                resultado_sharpe = optimizador.optimizar_sharpe(verbose=False)
                
                # Calcular pesos actuales
                total = self.gestor.df_cartera['Cantidad_Invertida'].sum()
                pesos_actuales = []
                if resultado_sharpe:
                    for activo in resultado_sharpe['activos']:
                        row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
                        if not row.empty:
                            pesos_actuales.append(row['Cantidad_Invertida'].iloc[0] / total)
                        else:
                            pesos_actuales.append(0.0)
                
                optimizador.graficar_frontera_eficiente(
                    portfolio_sharpe=resultado_sharpe,
                    portfolio_actual=np.array(pesos_actuales) if pesos_actuales else None
                )
                
                input("\nPresiona ENTER para continuar...")
            
            elif opcion == '5':
                # Comparar actual vs optimizada
                print("\n🎯 ¿Qué optimización deseas comparar?")
                print("   1. Máximo Sharpe")
                print("   2. Mínima volatilidad")
                
                tipo = input("\nOpción: ").strip()
                
                if tipo == '1':
                    resultado = optimizador.optimizar_sharpe(verbose=False)
                elif tipo == '2':
                    resultado = optimizador.optimizar_minima_volatilidad(verbose=False)
                else:
                    print("\n❌ Opción inválida")
                    input("\nPresiona ENTER para continuar...")
                    continue
                
                if resultado:
                    comparacion = optimizador.comparar_con_actual(resultado, verbose=True)
                    
                    if comparacion:
                        print("\n💡 ¿Generar gráficos de comparación?")
                        if self.validador.solicitar_confirmacion("   (s/n): "):
                            optimizador.graficar_composicion_comparada(resultado)
                
                input("\nPresiona ENTER para continuar...")
            
            else:
                print("\n❌ Opción no válida")
                input("\nPresiona ENTER...")
    
    
    def _exportar_optimizacion(self, resultado, optimizador):
        """
        Exporta los resultados de la optimización a Excel.
        
        Args:
            resultado: Diccionario con los resultados de la optimización
            optimizador: Instancia de OptimizadorMarkowitz
        """
        try:
            import openpyxl
            from openpyxl.styles import Font, PatternFill, Alignment
            
            # Nombre de archivo
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            nombre_archivo = f"optimizacion_{resultado['tipo']}_{timestamp}.xlsx"
            ruta_completa = self.gestor.DIR_EXPORTS / nombre_archivo
            
            # Crear DataFrame con los resultados
            df_pesos = pd.DataFrame({
                'Activo': resultado['activos'],
                'Peso_Optimizado_%': resultado['pesos'] * 100
            })
            
            # Calcular pesos actuales
            total_invertido = self.gestor.df_cartera['Cantidad_Invertida'].sum()
            pesos_actuales = []
            valores_actuales = []
            
            for activo in resultado['activos']:
                row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
                if not row.empty:
                    valor = row['Cantidad_Invertida'].iloc[0]
                    peso = (valor / total_invertido) * 100
                    pesos_actuales.append(peso)
                    valores_actuales.append(valor)
                else:
                    pesos_actuales.append(0.0)
                    valores_actuales.append(0.0)
            
            df_pesos['Peso_Actual_%'] = pesos_actuales
            df_pesos['Valor_Actual_€'] = valores_actuales
            df_pesos['Diferencia_%'] = df_pesos['Peso_Optimizado_%'] - df_pesos['Peso_Actual_%']
            df_pesos['Valor_Optimizado_€'] = (df_pesos['Peso_Optimizado_%'] / 100) * total_invertido
            df_pesos['Ajuste_€'] = df_pesos['Valor_Optimizado_€'] - df_pesos['Valor_Actual_€']
            
            # Ordenar por peso optimizado descendente
            df_pesos = df_pesos.sort_values('Peso_Optimizado_%', ascending=False)
            
            # Crear resumen
            rent_actual = optimizador.calcular_rentabilidad_portfolio(np.array(pesos_actuales) / 100)
            vol_actual = optimizador.calcular_volatilidad_portfolio(np.array(pesos_actuales) / 100)
            sharpe_actual = optimizador.calcular_sharpe_ratio(np.array(pesos_actuales) / 100)
            
            df_resumen = pd.DataFrame({
                'Métrica': [
                    'Tipo de Optimización',
                    'Rentabilidad Esperada Actual (%)',
                    'Rentabilidad Esperada Optimizada (%)',
                    'Mejora Rentabilidad (pp)',
                    'Volatilidad Actual (%)',
                    'Volatilidad Optimizada (%)',
                    'Reducción Volatilidad (pp)',
                    'Ratio Sharpe Actual',
                    'Ratio Sharpe Optimizado',
                    'Mejora Sharpe',
                    'Total Invertido (€)',
                    'Periodo Análisis (días)',
                    'Fecha Análisis'
                ],
                'Valor': [
                    resultado['tipo'].replace('_', ' ').title(),
                    f"{rent_actual*100:.2f}",
                    f"{resultado['rentabilidad']*100:.2f}",
                    f"{(resultado['rentabilidad']-rent_actual)*100:+.2f}",
                    f"{vol_actual*100:.2f}",
                    f"{resultado['volatilidad']*100:.2f}",
                    f"{(resultado['volatilidad']-vol_actual)*100:+.2f}",
                    f"{sharpe_actual:.3f}",
                    f"{resultado['sharpe']:.3f}",
                    f"{resultado['sharpe']-sharpe_actual:+.3f}",
                    f"{total_invertido:,.2f}",
                    f"{optimizador.periodo_analisis_dias}",
                    datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                ]
            })
            
            # Guardar en Excel
            with pd.ExcelWriter(ruta_completa, engine='openpyxl') as writer:
                df_resumen.to_excel(writer, sheet_name='Resumen', index=False)
                df_pesos.to_excel(writer, sheet_name='Distribución', index=False)
            
            print(f"\n✅ Optimización exportada:")
            print(f"   📁 {nombre_archivo}")
            print(f"   📂 {self.gestor.DIR_EXPORTS}")
            
        except Exception as e:
            print(f"\n❌ Error al exportar: {e}")
    
    
    # NOTA: Estos dos métodos deben añadirse a la clase InterfazUsuario
    # en el notebook principal
    
    def _estandarizar_columnas_temporal(self):
        """Estandariza nombres de columnas y elimina duplicados."""
        print("📊 Columnas actuales:")
        for i, col in enumerate(self.gestor.df_cartera.columns):
            print(f"   [{i:2d}] {col}")
        
        print("\n🔧 Estandarizando nombres...")
        
        eliminar = []
        
        # Valor_Actual vs Valor_actual
        if 'Valor_Actual' in self.gestor.df_cartera.columns and 'Valor_actual' in self.gestor.df_cartera.columns:
            print("   ⚠️  Encontradas: 'Valor_Actual' y 'Valor_actual'")
            if self.gestor.df_cartera['Valor_actual'].notna().any():
                print("      📝 Copiando valores de 'Valor_actual' a 'Valor_Actual'")
                self.gestor.df_cartera['Valor_Actual'] = self.gestor.df_cartera['Valor_actual']
            eliminar.append('Valor_actual')
        
        # Fecha_Actual vs Fecha_actual
        if 'Fecha_Actual' in self.gestor.df_cartera.columns and 'Fecha_actual' in self.gestor.df_cartera.columns:
            print("   ⚠️  Encontradas: 'Fecha_Actual' y 'Fecha_actual'")
            if self.gestor.df_cartera['Fecha_actual'].notna().any():
                print("      📝 Copiando valores de 'Fecha_actual' a 'Fecha_Actual'")
                self.gestor.df_cartera['Fecha_Actual'] = self.gestor.df_cartera['Fecha_actual']
            eliminar.append('Fecha_actual')
        
        # Plusv% vs Plusv.%
        if 'Plusv%' in self.gestor.df_cartera.columns and 'Plusv.%' in self.gestor.df_cartera.columns:
            print("   ⚠️  Encontradas: 'Plusv%' y 'Plusv.%'")
            if self.gestor.df_cartera['Plusv.%'].notna().any():
                print("      📝 Copiando valores de 'Plusv.%' a 'Plusv%'")
                self.gestor.df_cartera['Plusv%'] = self.gestor.df_cartera['Plusv.%']
            eliminar.append('Plusv.%')
        
        # Eliminar
        if eliminar:
            print(f"\n🗑️  Eliminando {len(eliminar)} columnas duplicadas:")
            for col in eliminar:
                print(f"      - {col}")
            self.gestor.df_cartera.drop(columns=eliminar, inplace=True)
            print(f"✅ Eliminadas {len(eliminar)} columnas")
        else:
            print("\n✅ No hay columnas duplicadas")
        
        print("\n📊 Columnas finales:")
        for i, col in enumerate(self.gestor.df_cartera.columns):
            print(f"   [{i:2d}] {col}")
        
        self.gestor.guardar_cartera()
        print("\n💾 Cartera guardada")
        input("\nPresiona ENTER...")
    
    def _guardar_y_salir(self):
        """Guarda la cartera actual y sale del programa"""
        print("\n" + "="*70)
        print("💾 GUARDAR Y SALIR")
        print("="*70)
        
        try:
            # Llama al método guardar_cartera de GestorCartera
            self.gestor.guardar_cartera()
            print("\n✅ Cartera guardada exitosamente")
            print("👋 ¡Hasta pronto!")
            print("="*70)
            return True  # Señal para salir del bucle del menú
            
        except Exception as e:
            print(f"\n❌ Error al guardar: {str(e)}")
            confirmar = ValidadorEntradas.solicitar_confirmacion(
                "¿Deseas salir de todas formas?"
            )
            return confirmar  # Sale si confirma, vuelve al menú si no      
```

# OPTIMIZADOR DE MARKOWITZ

```python
"""
MÓDULO ADICIONAL: OPTIMIZADOR DE MARKOWITZ
Ejecutar después de las Celdas 1-5 del notebook

Contiene:
- OptimizadorMarkowitz: Optimización de carteras usando teoría moderna de portfolio
- Cálculo de frontera eficiente
- Optimización de Sharpe ratio
- Optimización de mínima volatilidad
- Visualizaciones interactivas con Plotly
"""

import pandas as pd
import numpy as np
from scipy.optimize import minimize
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from datetime import datetime, timedelta


# ============================================================================
# CLASE: OPTIMIZADOR DE MARKOWITZ
# ============================================================================

class OptimizadorMarkowitz:
    """
    Optimizador de carteras basado en la teoría moderna de portfolio de Markowitz.
    
    Hereda funcionalidades de GestorCartera para acceder a:
    - Datos históricos de activos
    - Cálculo de rentabilidades
    - Gestión de archivos
    
    Funcionalidades:
    - Calcular matriz de covarianza
    - Generar frontera eficiente
    - Optimizar para máximo Sharpe ratio
    - Optimizar para mínima volatilidad
    - Optimizar para retorno objetivo
    - Visualizar resultados
    """
    
    def __init__(self, gestor_cartera, tasa_libre_riesgo=0.02):
        """
        Inicializa el optimizador.
        
        Args:
            gestor_cartera: Instancia de GestorCartera
            tasa_libre_riesgo: Tasa anual libre de riesgo (default 2%)
        """
        self.gestor = gestor_cartera
        self.tasa_libre_riesgo = tasa_libre_riesgo
        
        # Variables que se calculan
        self.df_rentabilidades = None
        self.matriz_cov = None
        self.rentabilidades_esperadas = None
        self.activos = []
        self.periodo_analisis_dias = 252  # 1 año por defecto
        
    
    # ========================================================================
    # PREPARACIÓN DE DATOS
    # ========================================================================
    
    def preparar_datos(self, dias_historico=252, verbose=True):
        """
        Prepara los datos históricos necesarios para la optimización.
        
        Args:
            dias_historico: Días de histórico a usar (default 252 = 1 año)
            verbose: Si True, muestra progreso
            
        Returns:
            bool: True si se prepararon correctamente los datos
        """
        if self.gestor.df_cartera.empty:
            if verbose:
                print("\n⚠️  La cartera está vacía")
            return False
        
        if verbose:
            print("\n" + "="*80)
            print("📊 PREPARANDO DATOS PARA OPTIMIZACIÓN")
            print("="*80)
            print(f"\n🔄 Cargando datos históricos ({dias_historico} días)...")
        
        self.periodo_analisis_dias = dias_historico
        fecha_inicio = datetime.now() - timedelta(days=dias_historico)
        
        # DataFrame para almacenar rentabilidades diarias
        df_precios = pd.DataFrame()
        activos_validos = []
        
        for idx, row in self.gestor.df_cartera.iterrows():
            archivo_csv = row['Fichero_CSV']
            nombre = row['Valor']
            
            if verbose:
                print(f"   📈 {nombre[:50]}...")
            
            try:
                # Cargar datos históricos
                df_hist = self.gestor.gestor_archivos.cargar_datos_historicos(
                    archivo_csv,
                    verbose=False
                )
                
                if df_hist is None or df_hist.empty:
                    if verbose:
                        print(f"      ❌ Sin datos históricos")
                    continue
                
                # Asegurar que tiene índice de fecha
                if not isinstance(df_hist.index, pd.DatetimeIndex):
                    if 'Date' in df_hist.columns:
                        df_hist = df_hist.set_index('Date')
                    df_hist.index = pd.to_datetime(df_hist.index)
                
                # Filtrar por periodo
                df_hist = df_hist[df_hist.index >= fecha_inicio]
                
                if df_hist.empty or 'Close' not in df_hist.columns:
                    if verbose:
                        print(f"      ⚠️  Sin datos en el periodo")
                    continue
                
                # Añadir columna de precios
                df_precios[nombre] = df_hist['Close']
                activos_validos.append(nombre)
                
                if verbose:
                    print(f"      ✅ {len(df_hist)} datos cargados")
                
            except Exception as e:
                if verbose:
                    print(f"      ❌ Error: {e}")
                continue
        
        if len(activos_validos) < 2:
            if verbose:
                print(f"\n❌ Se necesitan al menos 2 activos con datos")
                print(f"   Solo se encontraron {len(activos_validos)} activos válidos")
            return False
        
        # Eliminar filas con NaN
        df_precios = df_precios.dropna()
        
        if df_precios.empty:
            if verbose:
                print("\n❌ No hay datos comunes entre los activos")
            return False
        
        # Calcular rentabilidades diarias
        self.df_rentabilidades = df_precios.pct_change().dropna()
        
        # Calcular rentabilidades esperadas (anualizadas)
        self.rentabilidades_esperadas = self.df_rentabilidades.mean() * 252
        
        # Calcular matriz de covarianza (anualizada)
        self.matriz_cov = self.df_rentabilidades.cov() * 252
        
        # Guardar lista de activos
        self.activos = list(self.df_rentabilidades.columns)
        
        if verbose:
            print(f"\n✅ Datos preparados correctamente")
            print(f"   📊 Activos válidos: {len(self.activos)}")
            print(f"   📅 Días de datos: {len(self.df_rentabilidades)}")
            print(f"   📈 Rentabilidad media: {self.rentabilidades_esperadas.mean()*100:.2f}%")
            print(f"   📉 Volatilidad media: {np.sqrt(np.diag(self.matriz_cov)).mean()*100:.2f}%")
        
        return True
    
    
    # ========================================================================
    # FUNCIONES DE CÁLCULO DE PORTFOLIO
    # ========================================================================
    
    def calcular_rentabilidad_portfolio(self, pesos):
        """
        Calcula la rentabilidad esperada del portfolio.
        
        Args:
            pesos: Array numpy con los pesos de cada activo
            
        Returns:
            float: Rentabilidad esperada anualizada
        """
        return np.sum(self.rentabilidades_esperadas * pesos)
    
    
    def calcular_volatilidad_portfolio(self, pesos):
        """
        Calcula la volatilidad del portfolio.
        
        Args:
            pesos: Array numpy con los pesos de cada activo
            
        Returns:
            float: Volatilidad anualizada (desviación estándar)
        """
        return np.sqrt(np.dot(pesos.T, np.dot(self.matriz_cov, pesos)))
    
    
    def calcular_sharpe_ratio(self, pesos):
        """
        Calcula el ratio de Sharpe del portfolio.
        
        Args:
            pesos: Array numpy con los pesos de cada activo
            
        Returns:
            float: Ratio de Sharpe
        """
        rentabilidad = self.calcular_rentabilidad_portfolio(pesos)
        volatilidad = self.calcular_volatilidad_portfolio(pesos)
        
        if volatilidad == 0:
            return 0
        
        return (rentabilidad - self.tasa_libre_riesgo) / volatilidad
    
    
    # ========================================================================
    # OPTIMIZACIÓN
    # ========================================================================
    
    def optimizar_sharpe(self, verbose=True):
        """
        Encuentra el portfolio con máximo ratio de Sharpe.
        
        Returns:
            dict: Resultado de la optimización con pesos, rentabilidad, volatilidad y Sharpe
        """
        if self.df_rentabilidades is None:
            if verbose:
                print("\n❌ Primero debes preparar los datos con preparar_datos()")
            return None
        
        if verbose:
            print("\n" + "="*80)
            print("🎯 OPTIMIZACIÓN: MÁXIMO SHARPE RATIO")
            print("="*80)
        
        n_activos = len(self.activos)
        
        # Función objetivo: Negativo del Sharpe (porque minimizamos)
        def objetivo(pesos):
            return -self.calcular_sharpe_ratio(pesos)
        
        # Restricciones
        restricciones = (
            {'type': 'eq', 'fun': lambda x: np.sum(x) - 1}  # La suma debe ser 1
        )
        
        # Límites: cada peso entre 0 y 1
        limites = tuple((0, 1) for _ in range(n_activos))
        
        # Punto inicial: pesos iguales
        pesos_iniciales = np.array([1/n_activos] * n_activos)
        
        # Optimizar
        resultado = minimize(
            objetivo,
            pesos_iniciales,
            method='SLSQP',
            bounds=limites,
            constraints=restricciones,
            options={'maxiter': 1000}
        )
        
        if not resultado.success:
            if verbose:
                print(f"\n⚠️  Optimización no convergió: {resultado.message}")
            return None
        
        pesos_optimos = resultado.x
        rentabilidad = self.calcular_rentabilidad_portfolio(pesos_optimos)
        volatilidad = self.calcular_volatilidad_portfolio(pesos_optimos)
        sharpe = self.calcular_sharpe_ratio(pesos_optimos)
        
        if verbose:
            print(f"\n✅ Optimización exitosa")
            print(f"   📈 Rentabilidad esperada: {rentabilidad*100:.2f}%")
            print(f"   📉 Volatilidad: {volatilidad*100:.2f}%")
            print(f"   ⚡ Ratio Sharpe: {sharpe:.3f}")
            print(f"\n📊 DISTRIBUCIÓN ÓPTIMA:")
            print("-"*80)
            
            # Ordenar por peso descendente
            df_pesos = pd.DataFrame({
                'Activo': self.activos,
                'Peso_%': pesos_optimos * 100
            }).sort_values('Peso_%', ascending=False)
            
            for _, row in df_pesos.iterrows():
                if row['Peso_%'] >= 0.5:  # Solo mostrar pesos >= 0.5%
                    print(f"   {row['Activo'][:50]:<50} {row['Peso_%']:>6.2f}%")
        
        return {
            'pesos': pesos_optimos,
            'activos': self.activos,
            'rentabilidad': rentabilidad,
            'volatilidad': volatilidad,
            'sharpe': sharpe,
            'tipo': 'max_sharpe'
        }
    
    
    def optimizar_minima_volatilidad(self, verbose=True):
        """
        Encuentra el portfolio con mínima volatilidad.
        
        Returns:
            dict: Resultado de la optimización
        """
        if self.df_rentabilidades is None:
            if verbose:
                print("\n❌ Primero debes preparar los datos con preparar_datos()")
            return None
        
        if verbose:
            print("\n" + "="*80)
            print("🎯 OPTIMIZACIÓN: MÍNIMA VOLATILIDAD")
            print("="*80)
        
        n_activos = len(self.activos)
        
        # Función objetivo: volatilidad
        def objetivo(pesos):
            return self.calcular_volatilidad_portfolio(pesos)
        
        # Restricciones
        restricciones = (
            {'type': 'eq', 'fun': lambda x: np.sum(x) - 1}
        )
        
        # Límites
        limites = tuple((0, 1) for _ in range(n_activos))
        
        # Punto inicial
        pesos_iniciales = np.array([1/n_activos] * n_activos)
        
        # Optimizar
        resultado = minimize(
            objetivo,
            pesos_iniciales,
            method='SLSQP',
            bounds=limites,
            constraints=restricciones,
            options={'maxiter': 1000}
        )
        
        if not resultado.success:
            if verbose:
                print(f"\n⚠️  Optimización no convergió: {resultado.message}")
            return None
        
        pesos_optimos = resultado.x
        rentabilidad = self.calcular_rentabilidad_portfolio(pesos_optimos)
        volatilidad = self.calcular_volatilidad_portfolio(pesos_optimos)
        sharpe = self.calcular_sharpe_ratio(pesos_optimos)
        
        if verbose:
            print(f"\n✅ Optimización exitosa")
            print(f"   📈 Rentabilidad esperada: {rentabilidad*100:.2f}%")
            print(f"   📉 Volatilidad: {volatilidad*100:.2f}%")
            print(f"   ⚡ Ratio Sharpe: {sharpe:.3f}")
            print(f"\n📊 DISTRIBUCIÓN ÓPTIMA:")
            print("-"*80)
            
            df_pesos = pd.DataFrame({
                'Activo': self.activos,
                'Peso_%': pesos_optimos * 100
            }).sort_values('Peso_%', ascending=False)
            
            for _, row in df_pesos.iterrows():
                if row['Peso_%'] >= 0.5:
                    print(f"   {row['Activo'][:50]:<50} {row['Peso_%']:>6.2f}%")
        
        return {
            'pesos': pesos_optimos,
            'activos': self.activos,
            'rentabilidad': rentabilidad,
            'volatilidad': volatilidad,
            'sharpe': sharpe,
            'tipo': 'min_volatilidad'
        }
    
    
    def optimizar_retorno_objetivo(self, retorno_objetivo, verbose=True):
        """
        Encuentra el portfolio con mínima volatilidad para un retorno objetivo.
        
        Args:
            retorno_objetivo: Retorno objetivo anualizado (ej: 0.08 para 8%)
            
        Returns:
            dict: Resultado de la optimización
        """
        if self.df_rentabilidades is None:
            if verbose:
                print("\n❌ Primero debes preparar los datos con preparar_datos()")
            return None
        
        if verbose:
            print("\n" + "="*80)
            print(f"🎯 OPTIMIZACIÓN: RETORNO OBJETIVO {retorno_objetivo*100:.2f}%")
            print("="*80)
        
        n_activos = len(self.activos)
        
        # Función objetivo: volatilidad
        def objetivo(pesos):
            return self.calcular_volatilidad_portfolio(pesos)
        
        # Restricciones
        restricciones = (
            {'type': 'eq', 'fun': lambda x: np.sum(x) - 1},  # Suma = 1
            {'type': 'eq', 'fun': lambda x: self.calcular_rentabilidad_portfolio(x) - retorno_objetivo}  # Retorno objetivo
        )
        
        # Límites
        limites = tuple((0, 1) for _ in range(n_activos))
        
        # Punto inicial
        pesos_iniciales = np.array([1/n_activos] * n_activos)
        
        # Optimizar
        resultado = minimize(
            objetivo,
            pesos_iniciales,
            method='SLSQP',
            bounds=limites,
            constraints=restricciones,
            options={'maxiter': 1000}
        )
        
        if not resultado.success:
            if verbose:
                print(f"\n⚠️  No se pudo alcanzar el retorno objetivo")
                print(f"   Retorno máximo posible: {self.rentabilidades_esperadas.max()*100:.2f}%")
            return None
        
        pesos_optimos = resultado.x
        rentabilidad = self.calcular_rentabilidad_portfolio(pesos_optimos)
        volatilidad = self.calcular_volatilidad_portfolio(pesos_optimos)
        sharpe = self.calcular_sharpe_ratio(pesos_optimos)
        
        if verbose:
            print(f"\n✅ Optimización exitosa")
            print(f"   📈 Rentabilidad esperada: {rentabilidad*100:.2f}%")
            print(f"   📉 Volatilidad: {volatilidad*100:.2f}%")
            print(f"   ⚡ Ratio Sharpe: {sharpe:.3f}")
            print(f"\n📊 DISTRIBUCIÓN ÓPTIMA:")
            print("-"*80)
            
            df_pesos = pd.DataFrame({
                'Activo': self.activos,
                'Peso_%': pesos_optimos * 100
            }).sort_values('Peso_%', ascending=False)
            
            for _, row in df_pesos.iterrows():
                if row['Peso_%'] >= 0.5:
                    print(f"   {row['Activo'][:50]:<50} {row['Peso_%']:>6.2f}%")
        
        return {
            'pesos': pesos_optimos,
            'activos': self.activos,
            'rentabilidad': rentabilidad,
            'volatilidad': volatilidad,
            'sharpe': sharpe,
            'tipo': 'retorno_objetivo',
            'retorno_objetivo': retorno_objetivo
        }
    
    
    # ========================================================================
    # FRONTERA EFICIENTE
    # ========================================================================
    
    def calcular_frontera_eficiente(self, num_puntos=100, verbose=True):
        """
        Calcula la frontera eficiente.
        
        Args:
            num_puntos: Número de puntos a calcular en la frontera
            
        Returns:
            DataFrame con rentabilidades, volatilidades y Sharpe ratios
        """
        if self.df_rentabilidades is None:
            if verbose:
                print("\n❌ Primero debes preparar los datos con preparar_datos()")
            return None
        
        if verbose:
            print("\n" + "="*80)
            print(f"📈 CALCULANDO FRONTERA EFICIENTE ({num_puntos} puntos)")
            print("="*80)
        
        # Rango de retornos
        ret_min = self.rentabilidades_esperadas.min()
        ret_max = self.rentabilidades_esperadas.max()
        retornos_objetivo = np.linspace(ret_min, ret_max, num_puntos)
        
        resultados = []
        
        for i, ret_obj in enumerate(retornos_objetivo):
            if verbose and (i+1) % 20 == 0:
                print(f"   Progreso: {i+1}/{num_puntos} puntos...")
            
            resultado = self.optimizar_retorno_objetivo(ret_obj, verbose=False)
            
            if resultado:
                resultados.append({
                    'Rentabilidad': resultado['rentabilidad'],
                    'Volatilidad': resultado['volatilidad'],
                    'Sharpe': resultado['sharpe']
                })
        
        if not resultados:
            if verbose:
                print("\n❌ No se pudo calcular la frontera eficiente")
            return None
        
        df_frontera = pd.DataFrame(resultados)
        
        if verbose:
            print(f"\n✅ Frontera eficiente calculada")
            print(f"   Puntos válidos: {len(df_frontera)}")
        
        return df_frontera
    
    
    # ========================================================================
    # COMPARACIÓN CON CARTERA ACTUAL
    # ========================================================================
    
    def comparar_con_actual(self, portfolio_optimizado, verbose=True):
        """
        Compara el portfolio optimizado con la cartera actual.
        
        Args:
            portfolio_optimizado: Resultado de una optimización
            
        Returns:
            dict con la comparación
        """
        if portfolio_optimizado is None:
            return None
        
        # Calcular pesos actuales
        total_invertido = self.gestor.df_cartera['Cantidad_Invertida'].sum()
        pesos_actuales = []
        
        for activo in portfolio_optimizado['activos']:
            # Buscar el activo en la cartera
            row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
            if not row.empty:
                peso = row['Cantidad_Invertida'].iloc[0] / total_invertido
                pesos_actuales.append(peso)
            else:
                pesos_actuales.append(0.0)
        
        pesos_actuales = np.array(pesos_actuales)
        
        # Calcular métricas de cartera actual
        rent_actual = self.calcular_rentabilidad_portfolio(pesos_actuales)
        vol_actual = self.calcular_volatilidad_portfolio(pesos_actuales)
        sharpe_actual = self.calcular_sharpe_ratio(pesos_actuales)
        
        if verbose:
            print("\n" + "="*80)
            print("📊 COMPARACIÓN: CARTERA ACTUAL vs CARTERA OPTIMIZADA")
            print("="*80)
            
            print(f"\n{'MÉTRICA':<30} {'ACTUAL':>15} {'OPTIMIZADA':>15} {'DIFERENCIA':>15}")
            print("-"*80)
            print(f"{'Rentabilidad esperada':<30} {rent_actual*100:>14.2f}% {portfolio_optimizado['rentabilidad']*100:>14.2f}% {(portfolio_optimizado['rentabilidad']-rent_actual)*100:>+14.2f}%")
            print(f"{'Volatilidad':<30} {vol_actual*100:>14.2f}% {portfolio_optimizado['volatilidad']*100:>14.2f}% {(portfolio_optimizado['volatilidad']-vol_actual)*100:>+14.2f}%")
            print(f"{'Ratio Sharpe':<30} {sharpe_actual:>14.3f} {portfolio_optimizado['sharpe']:>14.3f} {portfolio_optimizado['sharpe']-sharpe_actual:>+14.3f}")
            
            print("\n📊 DIFERENCIAS EN PESOS:")
            print("-"*80)
            
            df_comparacion = pd.DataFrame({
                'Activo': portfolio_optimizado['activos'],
                'Actual_%': pesos_actuales * 100,
                'Optimizado_%': portfolio_optimizado['pesos'] * 100,
                'Diferencia_%': (portfolio_optimizado['pesos'] - pesos_actuales) * 100
            }).sort_values('Diferencia_%', key=abs, ascending=False)
            
            for _, row in df_comparacion.iterrows():
                if abs(row['Diferencia_%']) >= 0.5:  # Solo cambios >= 0.5%
                    print(f"   {row['Activo'][:40]:<40} {row['Actual_%']:>7.2f}% → {row['Optimizado_%']:>7.2f}% ({row['Diferencia_%']:>+7.2f}%)")
        
        return {
            'pesos_actuales': pesos_actuales,
            'rentabilidad_actual': rent_actual,
            'volatilidad_actual': vol_actual,
            'sharpe_actual': sharpe_actual,
            'pesos_optimizados': portfolio_optimizado['pesos'],
            'rentabilidad_optimizada': portfolio_optimizado['rentabilidad'],
            'volatilidad_optimizada': portfolio_optimizado['volatilidad'],
            'sharpe_optimizado': portfolio_optimizado['sharpe'],
            'activos': portfolio_optimizado['activos']
        }
    
    
    # ========================================================================
    # VISUALIZACIONES
    # ========================================================================
    
    def graficar_frontera_eficiente(self, portfolio_sharpe=None, portfolio_actual=None):
        """
        Crea un gráfico interactivo de la frontera eficiente.
        
        Args:
            portfolio_sharpe: Resultado de optimizar_sharpe()
            portfolio_actual: Pesos de la cartera actual
        """
        # Calcular frontera
        df_frontera = self.calcular_frontera_eficiente(num_puntos=50, verbose=False)
        
        if df_frontera is None:
            print("\n❌ No se pudo calcular la frontera eficiente")
            return
        
        # Crear figura
        fig = go.Figure()
        
        # Frontera eficiente
        fig.add_trace(go.Scatter(
            x=df_frontera['Volatilidad'] * 100,
            y=df_frontera['Rentabilidad'] * 100,
            mode='lines',
            name='Frontera Eficiente',
            line=dict(color='blue', width=3),
            hovertemplate='<b>Volatilidad:</b> %{x:.2f}%<br>' +
                         '<b>Rentabilidad:</b> %{y:.2f}%<br>' +
                         '<extra></extra>'
        ))
        
        # Portfolio de máximo Sharpe
        if portfolio_sharpe:
            fig.add_trace(go.Scatter(
                x=[portfolio_sharpe['volatilidad'] * 100],
                y=[portfolio_sharpe['rentabilidad'] * 100],
                mode='markers',
                name='Máximo Sharpe',
                marker=dict(color='green', size=15, symbol='star'),
                hovertemplate='<b>Máximo Sharpe</b><br>' +
                             '<b>Volatilidad:</b> %{x:.2f}%<br>' +
                             '<b>Rentabilidad:</b> %{y:.2f}%<br>' +
                             f'<b>Sharpe:</b> {portfolio_sharpe["sharpe"]:.3f}<br>' +
                             '<extra></extra>'
            ))
        
        # Cartera actual
        if portfolio_actual is not None:
            pesos_actuales = portfolio_actual
            rent_actual = self.calcular_rentabilidad_portfolio(pesos_actuales)
            vol_actual = self.calcular_volatilidad_portfolio(pesos_actuales)
            sharpe_actual = self.calcular_sharpe_ratio(pesos_actuales)
            
            fig.add_trace(go.Scatter(
                x=[vol_actual * 100],
                y=[rent_actual * 100],
                mode='markers',
                name='Cartera Actual',
                marker=dict(color='red', size=15, symbol='diamond'),
                hovertemplate='<b>Cartera Actual</b><br>' +
                             '<b>Volatilidad:</b> %{x:.2f}%<br>' +
                             '<b>Rentabilidad:</b> %{y:.2f}%<br>' +
                             f'<b>Sharpe:</b> {sharpe_actual:.3f}<br>' +
                             '<extra></extra>'
            ))
        
        # Layout
        fig.update_layout(
            title='Frontera Eficiente de Markowitz',
            xaxis_title='Volatilidad (Riesgo) %',
            yaxis_title='Rentabilidad Esperada %',
            hovermode='closest',
            template='plotly_white',
            width=1000,
            height=600,
            showlegend=True
        )
        
        fig.show()
        print("\n✅ Gráfico generado")
    
    
    def graficar_composicion_comparada(self, portfolio_optimizado):
        """
        Compara la composición actual vs optimizada.
        
        Args:
            portfolio_optimizado: Resultado de una optimización
        """
        # Calcular pesos actuales
        total_invertido = self.gestor.df_cartera['Cantidad_Invertida'].sum()
        pesos_actuales = []
        
        for activo in portfolio_optimizado['activos']:
            row = self.gestor.df_cartera[self.gestor.df_cartera['Valor'] == activo]
            if not row.empty:
                peso = row['Cantidad_Invertida'].iloc[0] / total_invertido
                pesos_actuales.append(peso)
            else:
                pesos_actuales.append(0.0)
        
        # Crear figura con subplots
        fig = make_subplots(
            rows=1, cols=2,
            specs=[[{'type': 'pie'}, {'type': 'pie'}]],
            subplot_titles=('Cartera Actual', 'Cartera Optimizada')
        )
        
        # Pie chart cartera actual
        fig.add_trace(
            go.Pie(
                labels=portfolio_optimizado['activos'],
                values=[p*100 for p in pesos_actuales],
                name="Actual",
                hovertemplate='<b>%{label}</b><br>%{value:.2f}%<extra></extra>'
            ),
            row=1, col=1
        )
        
        # Pie chart cartera optimizada
        fig.add_trace(
            go.Pie(
                labels=portfolio_optimizado['activos'],
                values=[p*100 for p in portfolio_optimizado['pesos']],
                name="Optimizada",
                hovertemplate='<b>%{label}</b><br>%{value:.2f}%<extra></extra>'
            ),
            row=1, col=2
        )
        
        fig.update_layout(
            title_text=f"Comparación de Composición - {portfolio_optimizado['tipo'].replace('_', ' ').title()}",
            showlegend=False,
            width=1200,
            height=500
        )
        
        fig.show()
        print("\n✅ Gráfico de comparación generado")
```

```python
"""
MÓDULO 6: ANÁLISIS TÉCNICO AVANZADO
Ejecutar en Celda 6 de Jupyter Lab (después de Celda 5, antes de ejecutar app)

Contiene:
- IndicadoresCalculadora: Cálculo de indicadores técnicos (RSI, MACD, etc.)
- Graficadora: Visualización con Matplotlib y Plotly
- ValorAnalizador: Análisis completo de un activo con métricas avanzadas
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import timedelta
from dataclasses import dataclass
import statsmodels.api as sm

# Imports opcionales para Plotly (si está instalado)
try:
    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    PLOTLY_AVAILABLE = True
except ImportError:
    PLOTLY_AVAILABLE = False
    print("⚠️  Plotly no disponible. Solo se usarán gráficos Matplotlib.")


# ============================================================================
# CLASE 10: CALCULADORA DE INDICADORES TÉCNICOS
# ============================================================================

@dataclass
class IndicadoresCalculadora:
    """
    Calcula indicadores técnicos: SMA, EMA, Bollinger, RSI, MACD, Drawdown.
    """
    df: pd.DataFrame
    
    def calcular_indicadores(self) -> pd.DataFrame:
        """
        Calcula una serie completa de indicadores técnicos.
        
        Returns:
            DataFrame con todos los indicadores calculados
        """
        df_ind = self.df.copy()
        
        # Asegurar que tenemos la columna Close
        if 'Close' not in df_ind.columns:
            raise ValueError("El DataFrame debe tener columna 'Close'")
        
        # Retornos (para métricas de riesgo)
        df_ind['Retornos'] = df_ind['Close'].pct_change()
        
        # 1. Medias Móviles
        df_ind['SMA20'] = df_ind['Close'].rolling(window=20).mean()
        df_ind['SMA50'] = df_ind['Close'].rolling(window=50).mean()
        df_ind['EMA20'] = df_ind['Close'].ewm(span=20, adjust=False).mean()
        
        # 2. Bandas de Bollinger (20 días, 2 desviaciones)
        std_dev = df_ind['Close'].rolling(window=20).std()
        df_ind['BB_Media'] = df_ind['SMA20']
        df_ind['BB_Superior'] = df_ind['SMA20'] + (std_dev * 2)
        df_ind['BB_Inferior'] = df_ind['SMA20'] - (std_dev * 2)
        
        # 3. RSI (Relative Strength Index 14 días)
        delta = df_ind['Close'].diff()
        gain = (delta.where(delta > 0, 0)).ewm(span=14, adjust=False).mean()
        loss = (-delta.where(delta < 0, 0)).ewm(span=14, adjust=False).mean()
        rs = gain / loss
        df_ind['RSI'] = 100 - (100 / (1 + rs))
        
        # 4. MACD (12, 26, 9)
        EMA_12 = df_ind['Close'].ewm(span=12, adjust=False).mean()
        EMA_26 = df_ind['Close'].ewm(span=26, adjust=False).mean()
        df_ind['MACD'] = EMA_12 - EMA_26
        df_ind['MACD_signal'] = df_ind['MACD'].ewm(span=9, adjust=False).mean()
        df_ind['MACD_hist'] = df_ind['MACD'] - df_ind['MACD_signal']
        
        # 5. Drawdown
        df_ind['C_max'] = df_ind['Close'].cummax()
        df_ind['Drawdown'] = (df_ind['Close'] / df_ind['C_max']) - 1
        
        return df_ind.dropna(subset=['Close'])
    
    def backtest_buy_and_hold(self, df_periodo: pd.DataFrame, capital: float):
        """
        Calcula rendimiento Buy & Hold y métricas para un período.
        
        Args:
            df_periodo: DataFrame con datos del período
            capital: Capital inicial invertido
            
        Returns:
            tuple: (DataFrame procesado, dict con métricas)
        """
        if df_periodo.empty:
            return pd.DataFrame(), {'Error': 'DataFrame del periodo vacío'}
        
        df_calc = df_periodo.copy()
        
        # Asegurar retornos
        if 'Retornos' not in df_calc.columns:
            df_calc['Retornos'] = df_calc['Close'].pct_change()
        
        # Precios
        precio_inicial = df_calc['Close'].iloc[0]
        precio_final = df_calc['Close'].iloc[-1]
        
        # Cálculo de valor
        acciones = capital / precio_inicial
        valor_final = acciones * precio_final
        ganancia = valor_final - capital
        
        # CAGR (Tasa de Crecimiento Anual Compuesto)
        dias = (df_calc.index[-1] - df_calc.index[0]).days
        años = dias / 365.25 if dias > 0 else 0
        cagr = (precio_final / precio_inicial) ** (1 / años) - 1 if años > 0 else 0
        
        # Retornos y Riesgo
        retornos_exceso = df_calc['Retornos'].fillna(0)
        std_dev_diaria = retornos_exceso.std()
        std_dev_anual = std_dev_diaria * np.sqrt(252)
        
        # Sharpe Ratio
        sharpe = (retornos_exceso.mean() * 252) / std_dev_anual if std_dev_anual > 0 else np.nan
        
        # Sortino Ratio
        downside = retornos_exceso[retornos_exceso < 0]
        downside_dev_diaria = downside.std() if len(downside) > 0 else 0
        downside_dev_anual = downside_dev_diaria * np.sqrt(252)
        sortino = (retornos_exceso.mean() * 252) / downside_dev_anual if downside_dev_anual > 0 else np.nan
        
        # Max Drawdown
        df_calc['C_max'] = df_calc['Close'].cummax()
        df_calc['Drawdown'] = (df_calc['Close'] / df_calc['C_max']) - 1
        max_drawdown = df_calc['Drawdown'].min()
        
        resumen = {
            'Fecha Inicio': df_calc.index[0].date(),
            'Fecha Fin': df_calc.index[-1].date(),
            'Capital Inicial': capital,
            'Valor Final': valor_final,
            'Ganancia': ganancia,
            'CAGR': cagr,
            'Volatilidad': std_dev_anual,
            'Sharpe': sharpe,
            'Sortino': sortino,
            'Max Drawdown': max_drawdown,
            'Años': años
        }
        
        return df_calc, resumen
    
    def calcular_metricas_avanzadas(self, df_benchmark: pd.DataFrame, 
                                    risk_free_rate: float = 0.02) -> dict:
        """
        Calcula Alpha, Beta, R², Tracking Error, Info Ratio.
        
        Args:
            df_benchmark: DataFrame con datos del benchmark
            risk_free_rate: Tasa libre de riesgo anual
            
        Returns:
            dict con métricas avanzadas
        """
        if df_benchmark is None or df_benchmark.empty:
            return {'Error': 'Benchmark no proporcionado'}
        
        # Últimos 3 años
        end_date = self.df.index.max()
        start_date_3yr = end_date - timedelta(days=3 * 365.25)
        
        df_fondo = self.df.loc[self.df.index >= start_date_3yr, ['Close']].copy()
        df_bench = df_benchmark.loc[df_benchmark.index >= start_date_3yr, ['Close']].copy()
        
        # Merge
        df_conjunto = pd.merge(
            df_fondo, df_bench, 
            left_index=True, right_index=True, 
            suffixes=('_fondo', '_bench')
        )
        
        if len(df_conjunto) < 100:
            return {'Error': f'Datos insuficientes ({len(df_conjunto)} días)'}
        
        # Retornos
        df_conjunto['R_fondo'] = df_conjunto['Close_fondo'].pct_change()
        df_conjunto['R_bench'] = df_conjunto['Close_bench'].pct_change()
        df_conjunto.dropna(subset=['R_fondo', 'R_bench'], inplace=True)
        
        # Exceso de retorno
        rfr_diario = (1 + risk_free_rate) ** (1/252) - 1
        df_conjunto['Exceso_fondo'] = df_conjunto['R_fondo'] - rfr_diario
        df_conjunto['Exceso_bench'] = df_conjunto['R_bench'] - rfr_diario
        
        # Regresión CAPM
        Y = df_conjunto['Exceso_fondo']
        X = sm.add_constant(df_conjunto['Exceso_bench'])
        
        model = sm.OLS(Y, X, missing='drop').fit()
        
        alpha_diario = model.params.const
        beta = model.params['Exceso_bench']
        r_squared = model.rsquared
        alpha_anual = alpha_diario * 252
        
        # Tracking Error
        diff_retornos = df_conjunto['R_fondo'] - df_conjunto['R_bench']
        tracking_error_anual = diff_retornos.std() * np.sqrt(252)
        exceso_retorno_anual = diff_retornos.mean() * 252
        info_ratio = exceso_retorno_anual / tracking_error_anual if tracking_error_anual != 0 else np.nan
        
        # Sharpe/Sortino 3yr
        std_dev_3yr = df_conjunto['R_fondo'].std() * np.sqrt(252)
        downside_3yr = df_conjunto['R_fondo'][df_conjunto['R_fondo'] < rfr_diario]
        downside_dev_3yr = downside_3yr.std() * np.sqrt(252) if len(downside_3yr) > 0 else 0
        retorno_anual_3yr = df_conjunto['R_fondo'].mean() * 252
        
        sharpe_3yr = (retorno_anual_3yr - risk_free_rate) / std_dev_3yr if std_dev_3yr > 0 else np.nan
        sortino_3yr = (retorno_anual_3yr - risk_free_rate) / downside_dev_3yr if downside_dev_3yr > 0 else np.nan
        
        return {
            'Alpha (Jensen) [Anual]': alpha_anual,
            'Beta': beta,
            'R-cuadrado (R²)': r_squared,
            'Ratio de Información': info_ratio,
            'Tracking Error [Anual]': tracking_error_anual,
            'Sharpe (3yr)': sharpe_3yr,
            'Sortino (3yr)': sortino_3yr,
        }


# ============================================================================
# CLASE 11: GRAFICADORA
# ============================================================================

class Graficadora:
    """Visualización con Matplotlib y Plotly."""
    
    @staticmethod
    def graficar_indicadores_mpl(df: pd.DataFrame, titulo: str = 'Análisis Técnico'):
        """
        Gráficos con Matplotlib.
        
        Args:
            df: DataFrame con indicadores calculados
            titulo: Título del gráfico
        """
        plt.style.use('seaborn-v0_8-darkgrid')
        fig, axes = plt.subplots(4, 1, figsize=(16, 18), sharex=True)
        
        # Panel 1: Precio y Bollinger
        axes[0].plot(df.index, df['Close'], label='Precio', linewidth=1.5, color='blue')
        if 'BB_Media' in df.columns:
            axes[0].plot(df.index, df['BB_Media'], label='BB Media', linestyle=':', color='black')
        if 'BB_Superior' in df.columns and 'BB_Inferior' in df.columns:
            axes[0].plot(df.index, df['BB_Superior'], label='BB Sup', linestyle='--', color='red')
            axes[0].plot(df.index, df['BB_Inferior'], label='BB Inf', linestyle='--', color='green')
            axes[0].fill_between(df.index, df['BB_Inferior'], df['BB_Superior'], alpha=0.1)
        axes[0].set_title('Precio y Bandas de Bollinger', fontsize=14, fontweight='bold')
        axes[0].legend()
        axes[0].grid(True, alpha=0.3)
        
        # Panel 2: MACD
        if 'MACD' in df.columns:
            axes[1].plot(df.index, df['MACD'], label='MACD', color='blue')
            axes[1].plot(df.index, df['MACD_signal'], label='Señal', linestyle='--', color='red')
            colors = ['green' if x >= 0 else 'red' for x in df.get('MACD_hist', [0])]
            axes[1].bar(df.index, df.get('MACD_hist', 0), alpha=0.5, color=colors, label='Histograma')
            axes[1].axhline(0, color='black', linewidth=0.8)
            axes[1].set_title('MACD', fontsize=14, fontweight='bold')
            axes[1].legend()
            axes[1].grid(True, alpha=0.3)
        
        # Panel 3: RSI
        if 'RSI' in df.columns:
            axes[2].plot(df.index, df['RSI'], label='RSI', color='purple', linewidth=1.5)
            axes[2].axhline(70, color='red', linestyle='--', label='Sobrecompra', alpha=0.7)
            axes[2].axhline(30, color='green', linestyle='--', label='Sobreventa', alpha=0.7)
            axes[2].fill_between(df.index, 70, 100, alpha=0.1, color='red')
            axes[2].fill_between(df.index, 0, 30, alpha=0.1, color='green')
            axes[2].set_title('RSI', fontsize=14, fontweight='bold')
            axes[2].set_ylim(0, 100)
            axes[2].legend()
            axes[2].grid(True, alpha=0.3)
        
        # Panel 4: Drawdown
        if 'Drawdown' in df.columns:
            dd = df['Drawdown'] * 100
            axes[3].fill_between(dd.index, dd, 0, where=dd < 0, color='red', alpha=0.4)
            axes[3].plot(dd.index, dd, color='darkred', linewidth=1.5)
            axes[3].axhline(0, color='black', linewidth=0.8)
            axes[3].set_title('Drawdown (%)', fontsize=14, fontweight='bold')
            axes[3].set_ylabel('%')
            axes[3].grid(True, alpha=0.3)
        
        plt.suptitle(titulo, fontsize=18, fontweight='bold')
        plt.tight_layout()
        plt.show()
    
    @staticmethod
    def graficar_indicadores_plotly(df: pd.DataFrame, titulo: str = 'Análisis Técnico'):
        """
        Gráficos interactivos con Plotly.
        
        Args:
            df: DataFrame con indicadores
            titulo: Título del gráfico
        """
        if not PLOTLY_AVAILABLE:
            print("⚠️  Plotly no disponible. Usando Matplotlib...")
            Graficadora.graficar_indicadores_mpl(df, titulo)
            return
        
        fig = make_subplots(
            rows=4, cols=1, shared_xaxes=True, vertical_spacing=0.05,
            row_titles=['Precio y Bollinger', 'MACD', 'RSI', 'Drawdown'],
            row_heights=[0.5, 0.15, 0.15, 0.2]
        )
        
        # Precio
        fig.add_trace(
            go.Scatter(x=df.index, y=df['Close'], name='Precio', 
                      line=dict(color='blue', width=2)),
            row=1, col=1
        )
        
        if 'BB_Superior' in df.columns and 'BB_Inferior' in df.columns:
            fig.add_trace(
                go.Scatter(x=df.index, y=df['BB_Superior'], name='BB Sup',
                          line=dict(dash='dash', color='red')),
                row=1, col=1
            )
            fig.add_trace(
                go.Scatter(x=df.index, y=df['BB_Inferior'], name='BB Inf',
                          line=dict(dash='dash', color='green'),
                          fill='tonexty', fillcolor='rgba(0,100,80,0.1)'),
                row=1, col=1
            )
        
        # MACD
        if 'MACD' in df.columns:
            fig.add_trace(
                go.Scatter(x=df.index, y=df['MACD'], name='MACD',
                          line=dict(color='blue')),
                row=2, col=1
            )
            fig.add_trace(
                go.Scatter(x=df.index, y=df['MACD_signal'], name='Señal',
                          line=dict(dash='dot', color='red')),
                row=2, col=1
            )
            colors = ['green' if v >= 0 else 'red' for v in df['MACD_hist']]
            fig.add_trace(
                go.Bar(x=df.index, y=df['MACD_hist'], name='Hist',
                      marker_color=colors, opacity=0.5),
                row=2, col=1
            )
        
        # RSI
        if 'RSI' in df.columns:
            fig.add_trace(
                go.Scatter(x=df.index, y=df['RSI'], name='RSI',
                          line=dict(color='purple', width=2)),
                row=3, col=1
            )
            fig.add_hline(y=70, line_dash="dash", line_color="red", row=3, col=1)
            fig.add_hline(y=30, line_dash="dash", line_color="green", row=3, col=1)
            fig.update_yaxes(range=[0, 100], row=3, col=1)
        
        # Drawdown
        if 'Drawdown' in df.columns:
            dd = df['Drawdown'] * 100
            fig.add_trace(
                go.Scatter(x=dd.index, y=dd, fill='tozeroy', name='Drawdown',
                          fillcolor='rgba(255,0,0,0.3)',
                          line=dict(color='darkred')),
                row=4, col=1
            )
        
        fig.update_layout(title_text=titulo, height=900, showlegend=True)
        fig.show()


# ============================================================================
# CLASE 12: ANALIZADOR DE VALOR
# ============================================================================

@dataclass
class ValorAnalizador:
    """
    Análisis completo de un activo: indicadores, métricas, gráficos.
    """
    df: pd.DataFrame = None
    df_benchmark: pd.DataFrame = None
    risk_free_rate: float = 0.02
    nombre_fondo: str = "Fondo"
    nombre_benchmark: str = "Benchmark"
    fecha_inicio_inversion: pd.Timestamp = None
    fecha_fin_inversion: pd.Timestamp = None
    graficar: bool = False
    tipo_grafico: str = 'plotly'
    
    def __post_init__(self):
        """Validaciones iniciales."""
        if self.df is None or self.df.empty:
            raise ValueError("DataFrame vacío o None")
        
        self.fecha_min_data = self.df.index.min().date()
        self.fecha_max_data = self.df.index.max().date()
        self.fecha_inicio_3yr = self.df.index.max() - timedelta(days=3 * 365.25)
    
    def generar_analisis_completo(self, capital: float) -> None:
        """
        Ejecuta análisis completo del activo.
        
        Args:
            capital: Capital inicial a invertir
        """
        print("\n" + "="*80)
        print(f"🔬 ANÁLISIS TÉCNICO COMPLETO: {self.nombre_fondo.upper()}")
        print("="*80)
        
        # Calcular indicadores
        calculadora = IndicadoresCalculadora(self.df)
        df_ind = calculadora.calcular_indicadores()
        
        # Filtrar período de inversión
        if self.fecha_inicio_inversion:
            df_inversion = df_ind.loc[df_ind.index >= self.fecha_inicio_inversion].copy()
        else:
            df_inversion = df_ind.copy()
        
        if self.fecha_fin_inversion:
            df_inversion = df_inversion.loc[df_inversion.index <= self.fecha_fin_inversion]
        
        if df_inversion.empty:
            print("❌ Sin datos para el período especificado")
            return
        
        # Backtest Buy & Hold
        df_bh, resumen_bh = calculadora.backtest_buy_and_hold(df_inversion, capital)
        
        # Métricas avanzadas
        resumen_adv = self._ejecutar_analisis_avanzado(calculadora)
        
        # Imprimir resultados
        self._imprimir_resultados(resumen_bh, resumen_adv, df_ind)
        
        # Gráficos
        if self.graficar:
            if self.tipo_grafico == 'plotly':
                Graficadora.graficar_indicadores_plotly(df_ind, self.nombre_fondo)
            else:
                Graficadora.graficar_indicadores_mpl(df_ind, self.nombre_fondo)
    
    def _ejecutar_analisis_avanzado(self, calculadora):
        """Ejecuta análisis avanzado si hay benchmark."""
        if self.df_benchmark is None or self.df_benchmark.empty:
            return {'Error': 'Sin benchmark'}
        return calculadora.calcular_metricas_avanzadas(self.df_benchmark, self.risk_free_rate)
    
    def _imprimir_resultados(self, resumen_bh, resumen_adv, df_ind):
        """
        Imprime resultados formateados CON diagnóstico pedagógico.
        Maneja correctamente la presencia/ausencia de benchmark.
        """
        print(f"\n📅 PERÍODO:")
        print(f"   Inicio: {resumen_bh['Fecha Inicio']}")
        print(f"   Fin: {resumen_bh['Fecha Fin']}")
        print(f"   Duración: {resumen_bh['Años']:.2f} años")
        
        print(f"\n📊 RENTABILIDAD (BUY & HOLD):")
        print(f"   Capital Inicial:    {resumen_bh['Capital Inicial']:>12,.2f} €")
        print(f"   Valor Final:        {resumen_bh['Valor Final']:>12,.2f} €")
        print(f"   Ganancia:           {resumen_bh['Ganancia']:>+12,.2f} €")
        print(f"   CAGR:               {resumen_bh['CAGR']*100:>12.2f}%")
        
        print(f"\n⚠️  RIESGO:")
        print(f"   Volatilidad:        {resumen_bh['Volatilidad']*100:>12.2f}%")
        print(f"   Max Drawdown:       {resumen_bh['Max Drawdown']*100:>12.2f}%")
        print(f"   Sharpe Ratio:       {resumen_bh['Sharpe']:>12.2f}")
        print(f"   Sortino Ratio:      {resumen_bh['Sortino']:>12.2f}")
        
        # Indicadores actuales
        if not df_ind.empty:
            ultimo = df_ind.iloc[-1]
            print(f"\n📈 INDICADORES ACTUALES:")
            if 'RSI' in df_ind.columns:
                rsi_val = ultimo['RSI']
                rsi_estado = "Sobrecompra" if rsi_val > 70 else "Sobreventa" if rsi_val < 30 else "Neutral"
                print(f"   RSI:                {rsi_val:>12.2f} ({rsi_estado})")
            if 'MACD' in df_ind.columns:
                print(f"   MACD:               {ultimo['MACD']:>12.4f}")
                print(f"   MACD Señal:         {ultimo['MACD_signal']:>12.4f}")
        
        # ========================================================================
        # ANÁLISIS DE BENCHMARK (solo si está disponible)
        # ========================================================================
        alpha_calculado = None
        beta_calculado = None
        r2_calculado = None
        ir_calculado = None
        hay_benchmark = False
        
        # Verificar si hay benchmark disponible
        if self.df_benchmark is not None and not self.df_benchmark.empty:
            hay_benchmark = True
            
            try:
                print(f"\n🎯 VS BENCHMARK ({self.nombre_benchmark}):")
                
                # Si resumen_adv tiene los valores, usarlos
                if 'Error' not in resumen_adv:
                    alpha_calculado = resumen_adv.get('Alpha (Jensen) [Anual]')
                    beta_calculado = resumen_adv.get('Beta')
                    r2_calculado = resumen_adv.get('R-cuadrado (R²)')
                    ir_calculado = resumen_adv.get('Ratio de Información')
                    
                    if alpha_calculado is not None:
                        print(f"   Alpha (anual):      {alpha_calculado*100:>12.2f}%")
                    if beta_calculado is not None:
                        print(f"   Beta:               {beta_calculado:>12.2f}")
                    if r2_calculado is not None:
                        print(f"   R² (correlación):   {r2_calculado*100:>12.2f}%")
                    
                    tracking_error = resumen_adv.get('Tracking Error [Anual]')
                    if tracking_error is not None:
                        print(f"   Tracking Error:     {tracking_error*100:>12.2f}%")
                    if ir_calculado is not None:
                        print(f"   Info Ratio:         {ir_calculado:>12.2f}")
                    
                    sharpe_3yr = resumen_adv.get('Sharpe (3yr)')
                    if sharpe_3yr is not None:
                        print(f"   Sharpe (3yr):       {sharpe_3yr:>12.2f}")
                
                else:
                    # Calcular manualmente
                    df_merged = pd.DataFrame({
                        'activo': self.df['Close'],
                        'benchmark': self.df_benchmark['Close']
                    }).dropna()
                    
                    if not df_merged.empty and len(df_merged) >= 30:
                        ret_activo = df_merged['activo'].pct_change().dropna()
                        ret_benchmark = df_merged['benchmark'].pct_change().dropna()
                        
                        df_ret = pd.DataFrame({
                            'activo': ret_activo,
                            'benchmark': ret_benchmark
                        }).dropna()
                        
                        if not df_ret.empty and len(df_ret) >= 30:
                            from scipy.stats import linregress
                            
                            slope, intercept, r_value, p_value, std_err = linregress(
                                df_ret['benchmark'],
                                df_ret['activo']
                            )
                            
                            beta_calculado = slope
                            alpha_calculado = intercept * 252
                            r2_calculado = r_value ** 2
                            
                            excess_returns = df_ret['activo'] - df_ret['benchmark']
                            tracking_error = excess_returns.std() * np.sqrt(252)
                            mean_excess = excess_returns.mean() * 252
                            ir_calculado = mean_excess / tracking_error if tracking_error != 0 else np.nan
                            
                            print(f"   Alpha (anual):      {alpha_calculado*100:>12.2f}%")
                            print(f"   Beta:               {beta_calculado:>12.2f}")
                            print(f"   R² (correlación):   {r2_calculado*100:>12.2f}%")
                            print(f"   Tracking Error:     {tracking_error*100:>12.2f}%")
                            print(f"   Info Ratio:         {ir_calculado:>12.2f}")
                        else:
                            print("   ⚠️  Datos insuficientes para calcular Alpha/Beta")
                    else:
                        print("   ⚠️  No hay datos alineados con el benchmark")
                        
            except Exception as e:
                print(f"   ⚠️  Error calculando vs benchmark: {e}")
                hay_benchmark = False
        else:
            # Sin benchmark
            print(f"\n💡 ANÁLISIS SIN BENCHMARK")
            print("   No se seleccionó índice de referencia")
            print("   Alpha, Beta, R² no disponibles")
        
        # ========================================================================
        # DIAGNÓSTICO PEDAGÓGICO
        # ========================================================================
        print("\n" + "="*80)
        print("💡 DIAGNÓSTICO PEDAGÓGICO")
        print("="*80)
        
        diagnostico = []
        
        # 1. Rentabilidad
        cagr = resumen_bh['CAGR'] * 100
        if cagr > 15:
            diagnostico.append("✅ RENTABILIDAD EXCELENTE: CAGR >15% - Supera ampliamente la media")
        elif cagr > 10:
            diagnostico.append("👍 RENTABILIDAD MUY BUENA: CAGR 10-15% - Por encima de la media")
        elif cagr > 7:
            diagnostico.append("👌 RENTABILIDAD BUENA: CAGR 7-10% - En línea con mercados desarrollados")
        elif cagr > 5:
            diagnostico.append("⚠️  RENTABILIDAD MODERADA: CAGR 5-7% - Apenas supera inflación")
        elif cagr > 0:
            diagnostico.append("⚠️  RENTABILIDAD BAJA: CAGR 0-5% - Considerar alternativas")
        else:
            diagnostico.append("🔴 RENTABILIDAD NEGATIVA: Revisa tu estrategia")
        
        # 2. Volatilidad
        vol = resumen_bh['Volatilidad'] * 100
        if vol < 10:
            diagnostico.append("✅ RIESGO BAJO: Volatilidad <10% - Activo conservador")
        elif vol < 15:
            diagnostico.append("👍 RIESGO MODERADO-BAJO: Volatilidad 10-15% - Equilibrado")
        elif vol < 20:
            diagnostico.append("⚠️  RIESGO MODERADO: Volatilidad 15-20% - Horizonte 3-5 años")
        elif vol < 30:
            diagnostico.append("⚠️  RIESGO ALTO: Volatilidad 20-30% - Alta tolerancia necesaria")
        else:
            diagnostico.append("🔴 RIESGO MUY ALTO: Volatilidad >30% - Extremadamente volátil")
        
        # 3. Sharpe
        sharpe = resumen_bh['Sharpe']
        if sharpe > 2:
            diagnostico.append("✅ EFICIENCIA EXCEPCIONAL: Sharpe >2 - Excelente rentabilidad/riesgo")
        elif sharpe > 1.5:
            diagnostico.append("👍 EFICIENCIA EXCELENTE: Sharpe 1.5-2 - Muy buena relación")
        elif sharpe > 1:
            diagnostico.append("👌 EFICIENCIA BUENA: Sharpe 1-1.5 - Compensación adecuada")
        elif sharpe > 0.5:
            diagnostico.append("⚠️  EFICIENCIA MODERADA: Sharpe 0.5-1 - Riesgo justificado")
        elif sharpe > 0:
            diagnostico.append("⚠️  EFICIENCIA BAJA: Sharpe 0-0.5 - Riesgo elevado vs rentabilidad")
        else:
            diagnostico.append("🔴 EFICIENCIA NEGATIVA: Sharpe <0 - Pérdidas")
        
        # 4. Max Drawdown
        dd = abs(resumen_bh['Max Drawdown'] * 100)
        if dd < 10:
            diagnostico.append("✅ DRAWDOWN BAJO: Caída <10% - Muy estable")
        elif dd < 20:
            diagnostico.append("👍 DRAWDOWN MODERADO: Caída 10-20% - Correcciones normales")
        elif dd < 30:
            diagnostico.append("⚠️  DRAWDOWN CONSIDERABLE: Caída 20-30% - Requiere paciencia")
        elif dd < 50:
            diagnostico.append("⚠️  DRAWDOWN ALTO: Caída 30-50% - Estómago fuerte necesario")
        else:
            diagnostico.append("🔴 DRAWDOWN EXTREMO: Caída >50% - Pérdidas severas")
        
        # 5. RSI
        if not df_ind.empty and 'RSI' in df_ind.columns:
            rsi_val = ultimo['RSI']
            if rsi_val > 70:
                diagnostico.append("⚠️  RSI SOBRECOMPRA (>70): Posiblemente sobrevalorado - Cuidado")
            elif rsi_val < 30:
                diagnostico.append("💡 RSI SOBREVENTA (<30): Posiblemente infravalorado - Oportunidad")
            else:
                diagnostico.append("✅ RSI NEUTRAL (30-70): Rango normal, sin señales extremas")
        
        # 6. MACD
        if not df_ind.empty and 'MACD' in df_ind.columns:
            macd = ultimo['MACD']
            macd_signal = ultimo['MACD_signal']
            
            if macd > macd_signal:
                diagnostico.append("📈 MACD ALCISTA: Por encima de señal - Momentum positivo")
            else:
                diagnostico.append("📉 MACD BAJISTA: Por debajo de señal - Momentum negativo")
        
        # 7. Alpha/Beta (SOLO si hay benchmark)
        if hay_benchmark and alpha_calculado is not None and beta_calculado is not None:
            alpha_pct = alpha_calculado * 100 if isinstance(alpha_calculado, float) else alpha_calculado
            
            if alpha_pct > 2:
                diagnostico.append(f"✅ ALPHA POSITIVO (+{alpha_pct:.1f}%): Supera claramente al benchmark")
            elif alpha_pct > 0:
                diagnostico.append(f"👍 ALPHA POSITIVO (+{alpha_pct:.1f}%): Bate al benchmark")
            elif alpha_pct > -2:
                diagnostico.append(f"⚠️  ALPHA NEGATIVO ({alpha_pct:.1f}%): No supera al benchmark")
            else:
                diagnostico.append(f"🔴 ALPHA MUY NEGATIVO ({alpha_pct:.1f}%): Muy inferior al mercado")
            
            if beta_calculado < 0.8:
                diagnostico.append(f"✅ BETA DEFENSIVO ({beta_calculado:.2f}): Menos volátil que mercado")
            elif beta_calculado < 1.2:
                diagnostico.append(f"👌 BETA NEUTRAL ({beta_calculado:.2f}): Se mueve con el mercado")
            else:
                diagnostico.append(f"⚠️  BETA AGRESIVO ({beta_calculado:.2f}): Más volátil que mercado")
            
            if r2_calculado is not None:
                r2_pct = r2_calculado * 100 if isinstance(r2_calculado, float) else r2_calculado
                if r2_pct > 80:
                    diagnostico.append(f"📊 R² ALTO ({r2_pct:.1f}%): Muy correlacionado con benchmark")
                elif r2_pct > 60:
                    diagnostico.append(f"📊 R² MODERADO ({r2_pct:.1f}%): Correlación moderada")
                else:
                    diagnostico.append(f"📊 R² BAJO ({r2_pct:.1f}%): Baja correlación - Más independiente")
        elif not hay_benchmark:
            diagnostico.append("💡 Sin benchmark: Alpha, Beta, R² no calculados")
        
        # 8. Recomendación
        print("\n📋 EVALUACIÓN:")
        for d in diagnostico:
            print(f"   {d}")
        
        print("\n💼 RECOMENDACIÓN GENERAL:")
        
        puntos = 0
        if cagr > 10: puntos += 2
        elif cagr > 5: puntos += 1
        
        if sharpe > 1: puntos += 2
        elif sharpe > 0.5: puntos += 1
        
        if vol < 15: puntos += 1
        if dd < 20: puntos += 1
        
        if hay_benchmark and alpha_calculado is not None:
            alpha_val = alpha_calculado * 100 if isinstance(alpha_calculado, float) else alpha_calculado
            if alpha_val > 0: puntos += 1
        
        if puntos >= 6:
            print("   ✅ INVERSIÓN EXCELENTE: Mantén en cartera")
            print("   💡 Considera aumentar posición si tu perfil lo permite")
        elif puntos >= 4:
            print("   👍 INVERSIÓN SÓLIDA: Buen activo a largo plazo")
            print("   💡 Monitoriza regularmente sin urgencia")
        elif puntos >= 2:
            print("   ⚠️  INVERSIÓN ACEPTABLE: Cumple función con limitaciones")
            print("   💡 Evalúa alternativas en tu estrategia")
        else:
            print("   🔴 INVERSIÓN A REVISAR: Considera reemplazar")
            print("   💡 Busca mejor relación rentabilidad/riesgo")
        
        print("\n" + "="*80)

print("✅ Módulo 6 cargado: IndicadoresCalculadora, Graficadora, ValorAnalizador")
print("📊 Análisis técnico avanzado disponible")
if not PLOTLY_AVAILABLE:
    print("⚠️  Instala plotly para gráficos interactivos: pip install plotly")
```

```python

"""
EXTENSIÓN DEL MÓDULO 6: ANÁLISIS COMPLETO DE CARTERA
Añadir a la Celda 6 después de ValorAnalizador

Contiene:
- AnalizadorCarteraCompleto: Análisis integral de todos los activos
- Diagnóstico pedagógico y recomendaciones
- Métricas avanzadas de cartera
"""

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats


# ============================================================================
# CLASE 13: ANALIZADOR COMPLETO DE CARTERA
# ============================================================================

class AnalizadorCarteraCompleto:
    """
    Análisis profesional completo de toda la cartera:
    - Análisis individual de cada activo
    - Métricas de cartera (CAGR, Sharpe, Sortino, Alpha, Beta)
    - Matriz de correlaciones
    - Diagnóstico pedagógico con recomendaciones
    """
    
    def __init__(self, gestor_cartera, gestor_archivos, benchmark_ticker='^STOXX50E'):
        """
        Args:
            gestor_cartera: Instancia de GestorCartera
            gestor_archivos: Instancia de GestorArchivos
            benchmark_ticker: Ticker del benchmark para comparaciones
        """
        self.gestor = gestor_cartera
        self.archivos = gestor_archivos
        self.benchmark_ticker = benchmark_ticker
        self.resultados_activos = []
        self.metricas_cartera = {}
        
    def ejecutar_analisis_completo(self, verbose=True, generar_graficos=True):
        """
        Ejecuta análisis completo de la cartera.
        
        Args:
            verbose: Si True, muestra progreso detallado
            generar_graficos: Si True, genera gráficos de cada activo
        """
        # ... código existente ...
        
        # 1. Análisis individual de cada activo CON GRÁFICOS
        print("\n🔍 FASE 1: Análisis individual de activos")
        print("-"*100)
        self._analizar_activos_individuales(verbose, generar_graficos)
        
        # 2. Cargar datos del benchmark
        print("\n📊 FASE 2: Cargando benchmark para comparación")
        print("-"*100)
        df_benchmark = self._cargar_benchmark()
        
        # 3. Métricas de cartera agregadas
        print("\n📈 FASE 3: Cálculo de métricas de cartera")
        print("-"*100)
        self._calcular_metricas_cartera(df_benchmark)
        
        # 4. Matriz de correlaciones
        print("\n🔗 FASE 4: Análisis de correlaciones")
        print("-"*100)
        self._analizar_correlaciones()
        
        # 5. Diagnóstico y recomendaciones
        print("\n💡 FASE 5: Generando diagnóstico pedagógico")
        print("-"*100)
        self._generar_diagnostico_pedagogico()
        
        # 6. Resumen final
        self._mostrar_resumen_final()
        
        return {
            'activos': self.resultados_activos,
            'cartera': self.metricas_cartera
        }



    def _calcular_alpha_beta(self, df_activo, df_benchmark):
        """
        Calcula Alpha, Beta, Ratio de Información vs benchmark.
        
        Args:
            df_activo: DataFrame con datos del activo
            df_benchmark: DataFrame con datos del benchmark
            
        Returns:
            dict: {'alpha', 'beta', 'ratio_informacion', 'r_squared', 'tracking_error'}
        """
        try:
            # Alinear fechas
            df_merged = pd.DataFrame({
                'activo': df_activo['Close'],
                'benchmark': df_benchmark['Close']
            }).dropna()
            
            if df_merged.empty or len(df_merged) < 30:
                return {
                    'alpha': np.nan,
                    'beta': np.nan,
                    'ratio_informacion': np.nan,
                    'r_squared': np.nan,
                    'tracking_error': np.nan
                }
            
            # Calcular retornos
            ret_activo = df_merged['activo'].pct_change().dropna()
            ret_benchmark = df_merged['benchmark'].pct_change().dropna()
            
            df_ret = pd.DataFrame({
                'activo': ret_activo,
                'benchmark': ret_benchmark
            }).dropna()
            
            if df_ret.empty or len(df_ret) < 30:
                return {
                    'alpha': np.nan,
                    'beta': np.nan,
                    'ratio_informacion': np.nan,
                    'r_squared': np.nan,
                    'tracking_error': np.nan
                }
            
            # Regresión lineal
            from scipy.stats import linregress
            
            slope, intercept, r_value, p_value, std_err = linregress(
                df_ret['benchmark'],
                df_ret['activo']
            )
            
            beta = slope
            alpha_anual = intercept * 252
            r_squared = r_value ** 2
            
            # Ratio de Información
            excess_returns = df_ret['activo'] - df_ret['benchmark']
            tracking_error = excess_returns.std() * np.sqrt(252)
            mean_excess_return = excess_returns.mean() * 252
            ratio_informacion = mean_excess_return / tracking_error if tracking_error != 0 else np.nan
            
            return {
                'alpha': alpha_anual,
                'beta': beta,
                'ratio_informacion': ratio_informacion,
                'r_squared': r_squared,
                'tracking_error': tracking_error
            }
            
        except Exception as e:
            print(f"      ⚠️  Error calculando Alpha/Beta: {e}")
            return {
                'alpha': np.nan,
                'beta': np.nan,
                'ratio_informacion': np.nan,
                'r_squared': np.nan,
                'tracking_error': np.nan
            }
    
    def _analizar_activos_individuales(self, verbose, generar_graficos=True):
        """Analiza cada activo de la cartera individualmente."""
        self.resultados_activos = []
        
        # Cargar benchmark
        print("\n  📊 Cargando benchmark para cálculos de Alpha/Beta...")
        df_benchmark = self._cargar_benchmark()
        
        if df_benchmark is None:
            print("  ⚠️  Benchmark no disponible - Alpha/Beta no se calcularán")
        else:
            print(f"  ✅ Benchmark cargado: {len(df_benchmark)} días")
        
        # Analizar cada activo
        for idx, row in self.gestor.df_cartera.iterrows():
            nombre = row['Valor']
            archivo = row['Fichero_CSV']
            
            if verbose:
                print(f"\n  📊 Analizando [{idx+1}/{len(self.gestor.df_cartera)}]: {nombre[:60]}...")
            
            df_hist = self.archivos.cargar_datos_historicos(archivo, verbose=False)
            
            if df_hist is None or df_hist.empty:
                print(f"     ❌ Sin datos históricos")
                self.resultados_activos.append({
                    'nombre': nombre,
                    'estado': 'sin_datos',
                    'error': 'No se pudieron cargar datos históricos'
                })
                continue
            
            try:
                # Preparar DataFrame
                df_analisis = df_hist.reset_index()
                if 'Date' in df_analisis.columns:
                    df_analisis.set_index('Date', inplace=True)
                
                # Generar gráficos
                if generar_graficos:
                    print(f"     📈 Generando gráficos interactivos...")
                    try:
                        analizador_grafico = ValorAnalizador(
                            df=df_analisis,
                            nombre_fondo=nombre,
                            fecha_inicio_inversion=row['Fecha_Inversion'],
                            graficar=True,
                            tipo_grafico='plotly'
                        )
                        analizador_grafico.generar_analisis_completo(
                            capital=row['Cantidad_Invertida']
                        )
                        print(f"     ✅ Gráficos generados")
                    except Exception as e:
                        print(f"     ⚠️  Error generando gráficos: {e}")
                
                # Calcular indicadores
                calculadora = IndicadoresCalculadora(df_analisis)
                df_ind = calculadora.calcular_indicadores()
                
                fecha_inv = pd.to_datetime(row['Fecha_Inversion'])
                df_periodo = df_ind[df_ind.index >= fecha_inv].copy()
                
                if df_periodo.empty:
                    print(f"     ⚠️  Sin datos desde inversión")
                    continue
                
                df_bh, metricas = calculadora.backtest_buy_and_hold(
                    df_periodo, 
                    row['Cantidad_Invertida']
                )
                
                ultimo = df_ind.iloc[-1]
                
                # Calcular Alpha/Beta
                if df_benchmark is not None:
                    print(f"     📊 Calculando Alpha/Beta vs benchmark...")
                    df_bench_periodo = df_benchmark[df_benchmark.index >= fecha_inv]
                    alpha_beta = self._calcular_alpha_beta(df_periodo, df_bench_periodo)
                else:
                    alpha_beta = {
                        'alpha': np.nan,
                        'beta': np.nan,
                        'ratio_informacion': np.nan,
                        'r_squared': np.nan,
                        'tracking_error': np.nan
                    }
                
                # Construir resultado
                resultado = {
                    'nombre': nombre,
                    'categoria': row['Categoria'],
                    'estado': 'ok',
                    'invertido': row['Cantidad_Invertida'],
                    'valor_actual': metricas['Valor Final'],
                    'ganancia': metricas['Ganancia'],
                    'rentabilidad_pct': (metricas['Ganancia'] / row['Cantidad_Invertida'] * 100) if row['Cantidad_Invertida'] > 0 else 0,
                    'cagr': metricas['CAGR'],
                    'volatilidad': metricas['Volatilidad'],
                    'sharpe': metricas['Sharpe'],
                    'sortino': metricas['Sortino'],
                    'max_drawdown': metricas['Max Drawdown'],
                    'rsi_actual': ultimo.get('RSI', np.nan),
                    'macd_actual': ultimo.get('MACD', np.nan),
                    'macd_signal': ultimo.get('MACD_Signal', np.nan),
                    'macd_hist': ultimo.get('MACD_Hist', np.nan),
                    'bb_upper': ultimo.get('BB_Upper', np.nan),
                    'bb_middle': ultimo.get('BB_Middle', np.nan),
                    'bb_lower': ultimo.get('BB_Lower', np.nan),
                    'sma_20': ultimo.get('SMA_20', np.nan),
                    'sma_50': ultimo.get('SMA_50', np.nan),
                    'alpha': alpha_beta['alpha'],
                    'beta': alpha_beta['beta'],
                    'ratio_informacion': alpha_beta['ratio_informacion'],
                    'r_squared': alpha_beta['r_squared'],
                    'tracking_error': alpha_beta['tracking_error'],
                    'precio_actual': ultimo['Close'],
                    'fecha_inicio': metricas['Fecha Inicio'],
                    'fecha_fin': metricas['Fecha Fin'],
                    'años': metricas['Años']
                }
                
                self.resultados_activos.append(resultado)
                
                # Output mejorado
                if verbose:
                    print(f"     💰 Rent: {resultado['rentabilidad_pct']:+.2f}% | "
                          f"CAGR: {metricas['CAGR']*100:+.2f}% | "
                          f"Valor: {metricas['Valor Final']:,.2f} €")
                    
                    print(f"     ⚠️  Vol: {metricas['Volatilidad']*100:.2f}% | "
                          f"Sharpe: {metricas['Sharpe']:.2f} | "
                          f"Sortino: {metricas['Sortino']:.2f} | "
                          f"MaxDD: {metricas['Max Drawdown']*100:.2f}%")
                    
                    rsi = resultado['rsi_actual']
                    if not np.isnan(rsi):
                        señal_rsi = "🔴Sobrev" if rsi > 70 else "🟢Sobrevend" if rsi < 30 else "⚪Neutral"
                        print(f"     📊 RSI: {rsi:.1f} ({señal_rsi}) | "
                              f"MACD: {resultado['macd_actual']:.3f} | "
                              f"Señal: {resultado['macd_signal']:.3f}")
                    
                    if not np.isnan(resultado['alpha']):
                        print(f"     📈 Alpha: {resultado['alpha']*100:+.2f}% | "
                              f"Beta: {resultado['beta']:.2f} | "
                              f"R²: {resultado['r_squared']:.3f} | "
                              f"IR: {resultado['ratio_informacion']:.2f}")
                
            except Exception as e:
                print(f"     ❌ Error: {e}")
                import traceback
                traceback.print_exc()
                self.resultados_activos.append({
                    'nombre': nombre,
                    'estado': 'error',
                    'error': str(e)
                })
        
        # ========================================================================
        # RESUMEN FINAL (debe estar AQUÍ, dentro de _analizar_activos_individuales)
        # ========================================================================
        num_ok = len([r for r in self.resultados_activos if r['estado'] == 'ok'])
        num_total = len(self.gestor.df_cartera)
        print(f"\n  ✅ Activos analizados exitosamente: {num_ok}/{num_total}")
        
        if num_ok < num_total:
            num_error = len([r for r in self.resultados_activos if r['estado'] == 'error'])
            num_sin_datos = len([r for r in self.resultados_activos if r['estado'] == 'sin_datos'])
            print(f"     ⚠️  Errores: {num_error} | Sin datos: {num_sin_datos}")
    



    def _cargar_benchmark(self):
        """Carga datos del benchmark."""
        import yfinance as yf
        
        try:
            # Obtener fecha más antigua de la cartera
            fecha_inicio = self.gestor.df_cartera['Fecha_Inversion'].min()
            fecha_fin = pd.Timestamp.now()
            
            print(f"  📅 Descargando {self.benchmark_ticker} desde {fecha_inicio.date()}...")
            
            df_bench = yf.download(
                self.benchmark_ticker,
                start=fecha_inicio,
                end=fecha_fin,
                progress=False,
                auto_adjust=True
            )
            
            if df_bench.empty:
                print(f"  ❌ No se pudo descargar el benchmark")
                return None
            
            # Normalizar columnas si es MultiIndex
            if isinstance(df_bench.columns, pd.MultiIndex):
                df_bench.columns = df_bench.columns.droplevel(1)
            
            print(f"  ✅ Benchmark cargado: {len(df_bench)} días")
            return df_bench
            
        except Exception as e:
            print(f"  ❌ Error al cargar benchmark: {e}")
            return None
    
    def _calcular_metricas_cartera(self, df_benchmark):
        """Calcula métricas agregadas de la cartera."""
        activos_ok = [r for r in self.resultados_activos if r['estado'] == 'ok']
        
        if not activos_ok:
            print("  ⚠️  No hay datos suficientes para métricas de cartera")
            return
        
        # Capital y valor
        total_invertido = sum(r['invertido'] for r in activos_ok)
        total_actual = sum(r['valor_actual'] for r in activos_ok)
        ganancia_total = total_actual - total_invertido
        rentabilidad_total = (ganancia_total / total_invertido * 100) if total_invertido > 0 else 0
        
        # CAGR ponderado
        cagr_ponderado = sum(
            r['cagr'] * (r['invertido'] / total_invertido)
            for r in activos_ok
        )
        
        # Volatilidad ponderada
        vol_ponderada = np.sqrt(sum(
            (r['volatilidad'] ** 2) * ((r['invertido'] / total_invertido) ** 2)
            for r in activos_ok
        ))
        
        # Sharpe y Sortino ponderados
        sharpe_ponderado = sum(
            r['sharpe'] * (r['invertido'] / total_invertido)
            for r in activos_ok
            if not np.isnan(r['sharpe'])
        )
        
        sortino_ponderado = sum(
            r['sortino'] * (r['invertido'] / total_invertido)
            for r in activos_ok
            if not np.isnan(r['sortino'])
        )
        
        # Max Drawdown ponderado
        max_dd_ponderado = sum(
            r['max_drawdown'] * (r['invertido'] / total_invertido)
            for r in activos_ok
        )
        
        self.metricas_cartera = {
            'total_invertido': total_invertido,
            'total_actual': total_actual,
            'ganancia_total': ganancia_total,
            'rentabilidad_total': rentabilidad_total,
            'cagr_ponderado': cagr_ponderado,
            'volatilidad_ponderada': vol_ponderada,
            'sharpe_ponderado': sharpe_ponderado,
            'sortino_ponderado': sortino_ponderado,
            'max_drawdown_ponderado': max_dd_ponderado,
            'num_activos': len(activos_ok),
            'num_positivos': len([r for r in activos_ok if r['ganancia'] > 0]),
            'num_negativos': len([r for r in activos_ok if r['ganancia'] < 0])
        }
        
        print(f"  ✅ Métricas calculadas para {len(activos_ok)} activos")
        print(f"     💰 Capital: {total_invertido:,.2f} € → {total_actual:,.2f} €")
        print(f"     📈 Rent. Total: {rentabilidad_total:+.2f}%")
        print(f"     📊 CAGR: {cagr_ponderado*100:+.2f}%")
        print(f"     ⚠️  Sharpe: {sharpe_ponderado:.2f}")
    
    def _analizar_correlaciones(self):
        """Calcula y visualiza matriz de correlaciones."""
        activos_ok = [r for r in self.resultados_activos if r['estado'] == 'ok']
        
        if len(activos_ok) < 2:
            print("  ⚠️  Se necesitan al menos 2 activos para correlaciones")
            return
        
        # Construir matriz de retornos
        retornos_dict = {}
        
        for idx, row in self.gestor.df_cartera.iterrows():
            nombre = row['Valor']
            archivo = row['Fichero_CSV']
            
            df_hist = self.archivos.cargar_datos_historicos(archivo, verbose=False)
            
            if df_hist is None or df_hist.empty:
                continue
            
            df_analisis = df_hist.reset_index().set_index('Date')
            df_analisis['Retornos'] = df_analisis['Close'].pct_change()
            
            retornos_dict[nombre[:20]] = df_analisis['Retornos']
        
        if len(retornos_dict) < 2:
            return
        
        # Crear DataFrame de retornos
        df_retornos = pd.DataFrame(retornos_dict).dropna()
        
        # Calcular correlación
        corr_matrix = df_retornos.corr()
        
        # Guardar en métricas
        self.metricas_cartera['correlacion_media'] = (
            corr_matrix.values[np.triu_indices_from(corr_matrix.values, k=1)].mean()
        )
        
        print(f"  ✅ Matriz de correlaciones calculada")
        print(f"     🔗 Correlación media: {self.metricas_cartera['correlacion_media']:.3f}")
        
        # Visualizar
        self._plot_correlacion(corr_matrix)
    
    def _plot_correlacion(self, corr_matrix):
        """Visualiza matriz de correlaciones."""
        plt.figure(figsize=(12, 10))
        sns.heatmap(
            corr_matrix,
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            center=0,
            square=True,
            linewidths=1,
            cbar_kws={"shrink": 0.8}
        )
        plt.title('Matriz de Correlaciones de Retornos', fontsize=16, fontweight='bold')
        plt.tight_layout()
        plt.show()
    
    def _generar_diagnostico_pedagogico(self):
        """Genera diagnóstico con recomendaciones pedagógicas."""
        activos_ok = [r for r in self.resultados_activos if r['estado'] == 'ok']
        
        if not activos_ok:
            return
        
        diagnostico = []
        
        # 1. Análisis de rentabilidad
        print("\n  📊 RENTABILIDAD:")
        cagr_cartera = self.metricas_cartera['cagr_ponderado'] * 100
        
        if cagr_cartera > 10:
            diagnostico.append("✅ EXCELENTE: CAGR superior al 10% anual")
        elif cagr_cartera > 5:
            diagnostico.append("👍 BUENO: CAGR entre 5-10% anual, sólido retorno")
        elif cagr_cartera > 0:
            diagnostico.append("⚠️  MODERADO: CAGR positivo pero bajo (<5%)")
        else:
            diagnostico.append("🔴 ATENCIÓN: CAGR negativo, revisar estrategia")
        
        # 2. Análisis de riesgo
        print("\n  ⚠️  RIESGO:")
        vol = self.metricas_cartera['volatilidad_ponderada'] * 100
        
        if vol < 10:
            diagnostico.append("✅ Volatilidad baja (<10%), cartera conservadora")
        elif vol < 20:
            diagnostico.append("👍 Volatilidad moderada (10-20%), perfil equilibrado")
        else:
            diagnostico.append("⚠️  Volatilidad alta (>20%), cartera agresiva")
        
        # 3. Análisis Sharpe
        print("\n  📈 EFICIENCIA:")
        sharpe = self.metricas_cartera['sharpe_ponderado']
        
        if sharpe > 1.5:
            diagnostico.append("✅ Sharpe >1.5: Excelente rentabilidad ajustada por riesgo")
        elif sharpe > 1:
            diagnostico.append("👍 Sharpe >1: Buena relación rentabilidad/riesgo")
        elif sharpe > 0.5:
            diagnostico.append("⚠️  Sharpe 0.5-1: Relación rentabilidad/riesgo aceptable")
        else:
            diagnostico.append("🔴 Sharpe <0.5: Riesgo excesivo para la rentabilidad obtenida")
        
        # 4. Diversificación
        print("\n  🔗 DIVERSIFICACIÓN:")
        if 'correlacion_media' in self.metricas_cartera:
            corr = self.metricas_cartera['correlacion_media']
            
            if corr < 0.3:
                diagnostico.append("✅ Baja correlación (<0.3): Excelente diversificación")
            elif corr < 0.6:
                diagnostico.append("👍 Correlación moderada (0.3-0.6): Buena diversificación")
            else:
                diagnostico.append("⚠️  Alta correlación (>0.6): Diversificación limitada")
        
        # 5. Distribución de resultados
        print("\n  📊 DISTRIBUCIÓN:")
        tasa_exito = (self.metricas_cartera['num_positivos'] / 
                     self.metricas_cartera['num_activos'] * 100)
        
        if tasa_exito >= 75:
            diagnostico.append(f"✅ {tasa_exito:.0f}% activos positivos: Excelente selección")
        elif tasa_exito >= 50:
            diagnostico.append(f"👍 {tasa_exito:.0f}% activos positivos: Selección equilibrada")
        else:
            diagnostico.append(f"⚠️  {tasa_exito:.0f}% activos positivos: Revisar selección")
        
        # 6. Activos problemáticos
        print("\n  🔍 ACTIVOS A REVISAR:")
        problematicos = [
            r for r in activos_ok
            if r['sharpe'] < 0.5 or r['cagr'] < 0
        ]
        
        if problematicos:
            diagnostico.append(f"⚠️  {len(problematicos)} activo(s) con bajo rendimiento/riesgo elevado")
        
        # Guardar diagnóstico
        self.metricas_cartera['diagnostico'] = diagnostico
        
        # Mostrar
        print("\n" + "="*100)
        print("💡 DIAGNÓSTICO PEDAGÓGICO")
        print("="*100)
        for d in diagnostico:
            print(f"  {d}")
        print("="*100)
    
    def _mostrar_resumen_final(self):
        """Muestra resumen ejecutivo del análisis."""
        if not self.metricas_cartera:
            return
        
        m = self.metricas_cartera
        
        print("\n" + "="*100)
        print("📋 RESUMEN EJECUTIVO DE CARTERA")
        print("="*100)
        
        print(f"\n💰 CAPITAL:")
        print(f"   Invertido:              {m['total_invertido']:>15,.2f} €")
        print(f"   Valor Actual:           {m['total_actual']:>15,.2f} €")
        print(f"   Ganancia/Pérdida:       {m['ganancia_total']:>+15,.2f} €")
        print(f"   Rentabilidad Total:     {m['rentabilidad_total']:>+15.2f} %")
        
        print(f"\n📊 MÉTRICAS DE RENDIMIENTO:")
        print(f"   CAGR (ponderado):       {m['cagr_ponderado']*100:>+15.2f} %")
        print(f"   Volatilidad:            {m['volatilidad_ponderada']*100:>15.2f} %")
        print(f"   Sharpe Ratio:           {m['sharpe_ponderado']:>15.2f}")
        print(f"   Sortino Ratio:          {m['sortino_ponderado']:>15.2f}")
        print(f"   Max Drawdown:           {m['max_drawdown_ponderado']*100:>15.2f} %")
        
        if 'correlacion_media' in m:
            print(f"\n🔗 DIVERSIFICACIÓN:")
            print(f"   Correlación Media:      {m['correlacion_media']:>15.3f}")
        
        print(f"\n📈 DISTRIBUCIÓN:")
        print(f"   Activos Totales:        {m['num_activos']:>15}")
        print(f"   Activos Positivos:      {m['num_positivos']:>15}")
        print(f"   Activos Negativos:      {m['num_negativos']:>15}")
        print(f"   Tasa de Éxito:          {(m['num_positivos']/m['num_activos']*100):>15.1f} %")
        
        print("\n" + "="*100)

    def _calcular_alpha_beta(self, df_activo, df_benchmark):
        """
        Calcula Alpha, Beta y Ratio de Información.
        
        Args:
            df_activo: DataFrame con datos del activo
            df_benchmark: DataFrame con datos del benchmark
            
        Returns:
            dict: {'alpha', 'beta', 'ratio_informacion', 'r_squared'}
        """
        try:
            # Alinear fechas
            df_merged = pd.DataFrame({
                'activo': df_activo['Close'],
                'benchmark': df_benchmark['Close']
            }).dropna()
            
            if df_merged.empty or len(df_merged) < 10:
                return {'alpha': np.nan, 'beta': np.nan, 'ratio_informacion': np.nan, 'r_squared': np.nan}
            
            # Calcular retornos
            ret_activo = df_merged['activo'].pct_change().dropna()
            ret_benchmark = df_merged['benchmark'].pct_change().dropna()
            
            # Alinear retornos
            df_ret = pd.DataFrame({
                'activo': ret_activo,
                'benchmark': ret_benchmark
            }).dropna()
            
            if df_ret.empty or len(df_ret) < 10:
                return {'alpha': np.nan, 'beta': np.nan, 'ratio_informacion': np.nan, 'r_squared': np.nan}
            
            # Regresión lineal: retorno_activo = alpha + beta * retorno_benchmark
            from scipy.stats import linregress
            
            slope, intercept, r_value, p_value, std_err = linregress(
                df_ret['benchmark'],
                df_ret['activo']
            )
            
            beta = slope
            alpha = intercept * 252  # Anualizado (asumiendo datos diarios)
            r_squared = r_value ** 2
            
            # Ratio de Información = (Retorno_activo - Retorno_benchmark) / Tracking_Error
            excess_returns = df_ret['activo'] - df_ret['benchmark']
            tracking_error = excess_returns.std() * np.sqrt(252)  # Anualizado
            
            mean_excess_return = excess_returns.mean() * 252  # Anualizado
            
            ratio_informacion = mean_excess_return / tracking_error if tracking_error != 0 else np.nan
            
            return {
                'alpha': alpha,
                'beta': beta,
                'ratio_informacion': ratio_informacion,
                'r_squared': r_squared,
                'tracking_error': tracking_error
            }
            
        except Exception as e:
            print(f"      ⚠️  Error calculando Alpha/Beta: {e}")
            return {'alpha': np.nan, 'beta': np.nan, 'ratio_informacion': np.nan, 'r_squared': np.nan}

# Ejemplo de uso
# Listar clases y métodos
import inspect

def listar_clases_y_metodos():
    clases = []
    for nombre, objeto in globals().items():
        if inspect.isclass(objeto) and objeto.__module__ == "__main__":
            clases.append(objeto)

    print("\n📦 Clases definidas en esta celda:")
    for clase in sorted(clases, key=lambda c: c.__name__):
        # Obtener el número de línea de la clase (si está disponible)
        try:
            line_number = inspect.getsourcelines(clase)[1]
            print(f"\n🔹 {clase.__name__} (línea {line_number})")
        except (OSError, TypeError):
            print(f"\n🔹 {clase.__name__} (línea desconocida)")

        # Listar todos los métodos (públicos y privados)
        metodos = [m for m in inspect.getmembers(clase, predicate=inspect.isfunction)]
        if metodos:
            for metodo_name, metodo_obj in metodos:
                # Obtener el número de línea del método (si está disponible)
                try:
                    metodo_line = inspect.getsourcelines(metodo_obj)[1]
                    print(f"   • {metodo_name} (línea {metodo_line})")
                except (OSError, TypeError):
                    print(f"   • {metodo_name} (línea desconocida)")
        else:
            print("   (sin métodos definidos)")

# Ejecuta la función
listar_clases_y_metodos()

```

```python
"""
CELDA DE INICIALIZACIÓN
Usa asset_data de la celda anterior (no de archivo .py)
"""

# ============================================================================
# 1. VERIFICAR ASSET DATA
# ============================================================================

print("="*70)
print("🚀 INICIALIZANDO GESTOR DE CARTERA")
print("="*70)

# Verificar que asset_data existe en el entorno
if 'asset_data' not in globals():
    print("\n⚠️  ERROR: asset_data no encontrado")
    print("   Asegúrate de ejecutar la celda con el diccionario asset_data ANTES de esta")
    raise NameError("asset_data no está definido. Ejecuta la celda con el diccionario primero.")
else:
    # Contar activos
    total_activos = sum(len(v) for k, v in asset_data.items() if k != 'Indices')
    total_indices = len(asset_data.get('Indices', {}))
    
    print(f"\n✅ Asset data encontrado:")
    print(f"   📊 {total_activos} activos")
    print(f"   📈 {total_indices} índices")

# ============================================================================
# 2. CARGAR ARCHIVOS CSV
# ============================================================================

print(f"\n📥 Cargando archivos CSV desde: {CONFIGURACION['dir_datos']}")

loader = FondosLoader(CONFIGURACION['dir_datos'])
lista_df = loader.cargar_todos()

print(f"   ✅ {len(lista_df)} archivos CSV cargados")

if len(lista_df) == 0:
    print(f"   ⚠️  No hay archivos CSV")
    print(f"      💡 Añade archivos CSV a: {CONFIGURACION['dir_datos']}")

# ============================================================================
# 3. INICIALIZAR GESTOR DE CARTERA
# ============================================================================
print(f"\n🎯 Inicializando GestorCartera...")

gestor = GestorCartera(
    directorio_datos=CONFIGURACION['dir_datos'],
    directorio_carteras=CONFIGURACION['dir_carteras'],
    directorio_exports=CONFIGURACION['dir_exports'],
    benchmark_ticker='^STOXX50E'
)

# Asignar datos cargados
gestor.asset_data = asset_data
gestor.lista_df = lista_df

print(f"   ✅ Gestor inicializado")

# ============================================================================
# 3.5 CARGAR O CREAR CARTERA
# ============================================================================
print(f"\n💼 Gestionando cartera...")

def crear_cartera_nueva(nombre_cartera):
    """Crea una cartera vacía y la guarda."""
    if nombre_cartera.endswith('.xlsx'):
        nombre_cartera = nombre_cartera.replace('.xlsx', '')
    
    gestor.nombre_cartera_actual = nombre_cartera
    gestor.archivo_cartera_actual = CONFIGURACION['dir_carteras'] / f'{nombre_cartera}.xlsx'
    gestor.archivo_bajas_actual   = CONFIGURACION['dir_carteras'] / f'{nombre_cartera}_bajas.xlsx'
    gestor.df_cartera = pd.DataFrame(columns=[
        'Valor', 'Categoria', 'ISIN', 'Fichero_CSV',
        'Fecha_Inversion', 'Cantidad_Invertida', 'Precio_Compra',
        'Operación', 'Num_Acciones', 'Fecha_Actual', 'Precio_Actual',
        'Valor_Actual', 'Beneficio', 'Plusv%',
        'Peso_Objetivo', 'Peso_Actual', 'Observaciones'
    ])
    gestor.guardar_cartera()
    print(f"   ✅ Cartera '{nombre_cartera}' creada")

# Verificar si hay carteras existentes
carteras_existentes = list(CONFIGURACION['dir_carteras'].glob('*.xlsx'))

if carteras_existentes:
    print(f"\n📁 Carteras disponibles: {len(carteras_existentes)}")
    for i, cartera in enumerate(carteras_existentes, 1):
        print(f"   {i}. {cartera.stem}")
    
    print(f"\n¿Qué deseas hacer?")
    print(f"   1. Cargar una cartera existente")
    print(f"   2. Crear una nueva cartera")
    
    opcion = input("\nOpción (1-2): ").strip()
    
    if opcion == '1':
        print(f"\nSelecciona una cartera:")
        for i, cartera in enumerate(carteras_existentes, 1):
            print(f"   {i}. {cartera.stem}")
        
        try:
            seleccion = int(input("\nNúmero de cartera: ").strip())
            if 1 <= seleccion <= len(carteras_existentes):
                cartera_seleccionada = carteras_existentes[seleccion - 1]
                if gestor.cargar_cartera(cartera_seleccionada.stem):
                    print(f"   ✅ Cartera '{cartera_seleccionada.stem}' cargada")
                else:
                    print(f"   ⚠️  Error al cargar")
            else:
                print(f"   ⚠️  Selección inválida")
        except:
            print(f"   ⚠️  Entrada inválida")
    
    else:
        nombre_cartera = input("\n   Nombre de la nueva cartera: ").strip()
        crear_cartera_nueva(nombre_cartera)

else:
    print(f"\n   ℹ️  No hay carteras existentes")
    nombre_cartera = input("\n   Nombre de la nueva cartera: ").strip()
    crear_cartera_nueva(nombre_cartera)
# ============================================================================
# 4. INICIALIZAR INTERFAZ DE USUARIO
# ============================================================================

print(f"\n🎨 Inicializando InterfazUsuario...")

app = InterfazUsuario(gestor)

print(f"   ✅ Interfaz lista")

# ============================================================================
# 5. RESUMEN FINAL
# ============================================================================

print("\n" + "="*70)
print("✅ SISTEMA LISTO PARA USAR")
print("="*70)
print(f"\n📊 INFORMACIÓN:")
print(f"   🌍 Entorno: {CONFIGURACION['entorno'].upper()}")
print(f"   📁 Archivos CSV: {len(lista_df)}")
print(f"   📊 Activos: {total_activos}")
print(f"   📈 Índices: {total_indices}")

# Verificar carteras existentes
if CONFIGURACION['dir_carteras'].exists():
    carteras = list(CONFIGURACION['dir_carteras'].glob('*.xlsx'))
    if carteras:
        print(f"   💼 Carteras: {len(carteras)}")
        for c in carteras[:3]:
            print(f"      - {c.stem}")
        if len(carteras) > 3:
            print(f"      ... y {len(carteras)-3} más")
    else:
        print(f"   💼 No hay carteras (crea una con opción 2)")

print("\n" + "="*70)
print("💡 PARA COMENZAR:")
print("   >>> app.menu_principal()")
print("="*70)

# ============================================================================
# FUNCIÓN AUXILIAR PARA COLAB
# ============================================================================

if CONFIGURACION['entorno'] == 'colab':
    from google.colab import files
    
    def descargar_archivo(nombre_archivo, tipo='cartera'):
        """Descarga archivo desde Colab."""
        if tipo == 'cartera':
            ruta = CONFIGURACION['dir_carteras'] / nombre_archivo
        elif tipo == 'export':
            ruta = CONFIGURACION['dir_exports'] / nombre_archivo
        else:
            print(f"❌ Tipo '{tipo}' no válido")
            return
        
        if ruta.exists():
            files.download(str(ruta))
            print(f"✅ Descargando: {nombre_archivo}")
        else:
            print(f"❌ No existe: {ruta}")
    
    print("\n💾 Función de descarga disponible (Colab):")
    print("   >>> descargar_archivo('mi_cartera.xlsx', tipo='cartera')")
```
